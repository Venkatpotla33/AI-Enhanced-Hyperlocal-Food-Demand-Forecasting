## 04.1.6 — Zone Configuration Inspection

### Objective

Inspect the existing zone configuration file to identify the five
hyperlocal zones and determine whether geographical metadata is available
for integrating weather and location-specific event information.

### Scope

This subsection reads the existing `config/zones.yaml` file without
modifying it.

The inspection will verify zone identifiers, zone names, geographical
coordinates, and any other available location-related metadata.

The findings will determine whether additional geographical information
is required before contextual data collection and integration.

````python
from pathlib import Path
import yaml
from pprint import pprint

# Locate the project root by searching parent directories
current_path = Path.cwd()
project_root = None

for path in [current_path, *current_path.parents]:
    if (
        (path / "config" / "zones.yaml").is_file()
        and (path / "notebooks").is_dir()
    ):
        project_root = path
        break

if project_root is None:
    raise FileNotFoundError(
        "Could not locate the project root containing "
        "'config/zones.yaml' and the 'notebooks' directory."
    )

# Locate the existing zone configuration
zones_file = project_root / "config" / "zones.yaml"

print("Project root:", project_root)
print("Zone configuration file:", zones_file)
print("File exists:", zones_file.exists())

# Read the configuration without modifying it
with open(zones_file, "r", encoding="utf-8") as file:
    zones_config = yaml.safe_load(file)

print("\nZone configuration content:")
pprint(zones_config)

print("\nConfiguration data type:", type(zones_config).__name__)
````

````python
# Inspect the available fields for each configured zone

zones = zones_config.get("zones", [])

print("Number of configured zones:", len(zones))
print("\nZone-level field inspection:\n")

for zone in zones:
    print(f"Zone ID: {zone.get('zone_id')}")
    print(f"Zone name: {zone.get('zone_name')}")
    print("Available fields:", list(zone.keys()))

    coordinate_fields = [
        key for key in zone.keys()
        if any(term in key.lower() for term in [
            "latitude", "longitude", "lat", "lon",
            "coordinate", "boundary", "polygon"
        ])
    ]

    print("Geographical fields detected:", coordinate_fields or "None")
    print("-" * 60)
````

# 04.2 — Contextual Data Source Evaluation

## Objective

The objective of this phase is to identify, evaluate, and select reliable external data sources for enriching the hyperlocal food demand forecasting framework with contextual information.

The project focuses on five hyperlocal zones in Hyderabad, Telangana, India: Kondapur, Hitech City, Gachibowli, Madhapur, and Kukatpally.

Contextual information may help forecasting models represent changes in demand associated with weather conditions, public holidays, regional festivals, sports events, and local events.

## Contextual Data Categories

The following categories will be evaluated:

1. **Weather Data:** Historical weather observations, archived weather forecasts, and current weather forecasts.
2. **Public Holidays:** Official public holiday calendars relevant to Telangana.
3. **Regional Festivals:** Festival dates and celebrations relevant to Hyderabad and Telangana.
4. **Sports Events:** IPL matches and other relevant sporting events.
5. **Local Events:** Concerts, exhibitions, public gatherings, and other events that may influence local activity.

## Source Evaluation Criteria

Each candidate source will be evaluated using the following criteria:

* Source reliability and authenticity.
* Historical data coverage for the project period.
* Availability of current and future information.
* Temporal resolution and timestamp accuracy.
* Geographical resolution and relevance to the study zones.
* Data access method, including APIs or downloadable files.
* Licensing, attribution, and usage restrictions.
* Update frequency and data availability.
* Suitability for historical backtesting and live forecasting.
* Missing-data risks and preprocessing requirements.

## Forecasting and Data Leakage Considerations

The project distinguishes between historical observations and information that would have been available when a forecast was issued.

For historical backtesting, archived forecasts should be preferred when evaluating the operational value of weather forecasts. Historical weather observations may be used for descriptive analysis or separate experiments, but they must not be treated as information that was known in advance.

Similarly, future event information may only be used when it would have been available at the forecast issue time.

Because the demand dataset is synthetic, the integration of real contextual information does not establish real-world causal relationships between these factors and actual food delivery demand.

## Scope

This phase evaluates candidate sources and documents their suitability before data collection begins.

No contextual datasets will be downloaded, transformed, or integrated until the source evaluation is sufficiently complete.

## Expected Outcome

The expected outcome is a documented source-selection strategy identifying the preferred source, data coverage, relevant variables, access method, limitations, and intended use for each contextual category.


## 04.2.1 — Weather Data Source Evaluation

### 1. Objective

The objective of this subsection is to evaluate suitable weather data sources for the AI-Enhanced Hyperlocal Food Demand Forecasting and Surge Intelligence project.

The project requires hourly weather information to investigate potential relationships between weather conditions and synthetic food demand across five hyperlocal zones in Hyderabad, Telangana.

The evaluation considers three requirements:

1. Historical weather information for exploratory analysis and retrospective experiments.
2. Archived weather forecasts for leakage-aware historical forecasting experiments.
3. Current and future weather forecasts for live demand prediction.

### 2. Candidate Weather Data Sources

#### Source A: Open-Meteo Historical Weather API

The Open-Meteo Historical Weather API provides historical weather information using reanalysis datasets, including ERA5 and ERA5-Land.

Official documentation:
https://open-meteo.com/en/docs/historical-weather-api

Potential hourly variables include:

* Temperature at 2 metres.
* Relative humidity.
* Precipitation.
* Rainfall.
* Cloud cover.
* Wind speed.
* Weather codes.

**Potential application:** Historical weather analysis and experiments evaluating the association between weather conditions and synthetic food demand.

**Limitation:** Historical reanalysis data represent reconstructed weather conditions. They do not establish which weather forecasts were available when a historical demand prediction would have been made.

#### Source B: Open-Meteo Historical Forecast API

The Open-Meteo Historical Forecast API provides archived weather forecasts, with historical coverage beginning around 2022 according to its documentation.

Official documentation:
https://open-meteo.com/en/docs/historical-forecast-api

Potential hourly variables include:

* Temperature.
* Relative humidity.
* Precipitation probability.
* Precipitation.
* Rainfall.
* Cloud cover.
* Wind speed.
* Weather codes.

**Potential application:** Historical forecasting experiments that use weather forecast information rather than actual future weather observations.

**Limitation:** Available forecast dates, model runs, lead times, and variables must be verified for the required study period. A continuous hourly forecast archive does not necessarily preserve every original forecast run and lead time.

#### Source C: Open-Meteo Weather Forecast API

The Open-Meteo Weather Forecast API provides current and future weather forecasts for specified geographical coordinates.

Official documentation:
https://open-meteo.com/en/docs

Potential hourly variables include:

* Temperature.
* Relative humidity.
* Precipitation probability.
* Precipitation.
* Cloud cover.
* Wind speed.
* Weather codes.

**Potential application:** Supplying weather forecasts to the live demand forecasting workflow.

**Limitation:** Forecast availability and horizon depend on the selected API configuration and weather models. Current forecasts cannot substitute for archived forecasts in historical backtesting.

### 3. Comparative Evaluation

| Evaluation criterion            | Historical Weather API               | Historical Forecast API                              | Weather Forecast API            |
| ------------------------------- | ------------------------------------ | ---------------------------------------------------- | ------------------------------- |
| Historical weather observations | Yes, through reanalysis              | Provides archived forecasts                          | Not its primary purpose         |
| Archived forecast information   | No                                   | Yes                                                  | No historical archive guarantee |
| Current forecasts               | No; separate endpoint                | Historical archive                                   | Yes                             |
| Hourly variables                | Available                            | Available                                            | Available                       |
| Historical demand analysis      | Suitable                             | Suitable for forecast-based experiments              | Not sufficient alone            |
| Leakage-aware backtesting       | Requires careful experimental design | Preferred candidate, subject to archive verification | Not sufficient alone            |
| Live demand prediction          | Not suitable alone                   | Not suitable alone                                   | Preferred candidate             |
| Geographical inputs             | Latitude and longitude               | Latitude and longitude                               | Latitude and longitude          |

### 4. Evaluation Criteria

The candidate sources will be assessed according to:

1. Availability of hourly data.
2. Coverage of the required historical dates.
3. Availability of the necessary weather variables.
4. Forecast issue time and lead-time availability.
5. Suitability for historical backtesting.
6. Suitability for live forecasting.
7. Geographical resolution and relevance to Hyderabad.
8. Timezone consistency with the demand dataset.
9. Data access requirements and usage restrictions.
10. Reproducibility and documentation of data provenance.

### 5. Preliminary Source Selection

The preliminary strategy is to evaluate two complementary weather datasets:

* **Historical reanalysis weather:** For historical weather analysis and separately defined retrospective experiments.
* **Archived weather forecasts:** For experiments intended to reproduce information available when a historical forecast was issued.

The live Weather Forecast API will be evaluated separately for the operational forecasting component.

Final selection will depend on verifying actual date coverage, hourly completeness, geographical suitability, variables, forecast lead times, and applicable usage terms.

No weather data have been downloaded or integrated during this subsection.

### 6. Research Limitation

The food demand dataset used in this project is synthetic. Combining it with genuine weather information does not establish that weather caused changes in real-world food delivery demand.

Results will therefore be interpreted as model performance on the synthetic demand benchmark and as an evaluation of the forecasting framework, not as proof of real-world causal relationships.


## 04.2.2 — Weather Source Availability Verification

### 1. Objective

This subsection verifies the practical availability of weather data from the selected Open-Meteo APIs before collecting data for the project.

The verification focuses on historical weather, archived weather forecasts, and live weather forecasts for Hyderabad, Telangana, India.

### 2. Sources to Verify

**A. Historical Weather API**

Official documentation: https://open-meteo.com/en/docs/historical-weather-api

Purpose:

* Obtain historical weather information from reanalysis datasets.
* Support retrospective weather analysis.
* Evaluate the association between weather variables and synthetic hourly food demand.

**B. Historical Forecast API**

Official documentation: https://open-meteo.com/en/docs/historical-forecast-api

Purpose:

* Retrieve archived weather forecast information.
* Investigate the use of forecast-based weather features in historical demand forecasting experiments.
* Verify available dates, variables, and forecast archive characteristics.

The API documents a continuous hourly series constructed by joining the early hours of successive forecast runs. This is not equivalent to preserving every complete original forecast run.

**C. Previous Model Runs API**

Official documentation: https://open-meteo.com/en/docs/previous-runs-api

Purpose:

* Evaluate weather forecasts at specified lead times.
* Investigate forecasts issued one or more days before the target time.
* Support experiments that require explicit forecast lead-time information.

The available history depends on the weather model and variable. Coverage must be verified before selecting this source.

**D. Live Weather Forecast API**

Official documentation: https://open-meteo.com/en/docs

Purpose:

* Obtain current and future weather forecasts.
* Supply weather information to the live demand forecasting workflow.

### 3. Verification Criteria

The following criteria will be checked:

1. API accessibility.
2. Availability of the requested date range.
3. Availability of required hourly weather variables.
4. Presence of valid timestamps and numerical observations.
5. Availability of archived forecasts for the required historical period.
6. Suitability of the source for historical backtesting or live prediction.
7. Geographic and temporal consistency with the demand dataset.
8. API usage terms and attribution requirements.

### 4. Geographical Approach

The existing zone configuration does not contain geographical coordinates.

For the initial API connectivity test, a representative Hyderabad city coordinate will be used as a provisional test location.

This test will not establish that the returned weather represents each of the five hyperlocal zones independently.

The final geographical approach will be documented before weather data are integrated into the demand panel.

### 5. Forecasting Integrity

Historical weather observations and archived weather forecasts will be treated as distinct data sources.

For historical backtesting, the weather information used for each prediction must be consistent with the information that would have been available at the forecast issue time.

Where archived forecasts do not preserve the necessary issue-time information, the experiment will be labelled accordingly rather than treated as a faithful simulation of live forecasting.

### 6. Scope Boundary

This subsection performs source verification only.

No complete weather dataset will be downloaded, and no weather data will be merged with the hourly demand panel at this stage.

The source-selection decision will be made after reviewing the actual API test results.


````python
import requests
import pandas as pd

# Provisional city-level test location for Hyderabad.
# This is a connectivity-test point, not a validated coordinate
# for any individual hyperlocal zone.
latitude = 17.3850
longitude = 78.4867

# Test one day within the documented project period.
start_date = "2026-08-30"
end_date = "2026-08-30"

url = "https://archive-api.open-meteo.com/v1/archive"

params = {
    "latitude": latitude,
    "longitude": longitude,
    "start_date": start_date,
    "end_date": end_date,
    "hourly": [
        "temperature_2m",
        "relative_humidity_2m",
        "precipitation",
        "rain",
        "wind_speed_10m",
        "cloud_cover"
    ],
    "timezone": "Asia/Kolkata"
}

try:
    response = requests.get(url, params=params, timeout=30)

    print("HTTP status:", response.status_code)
    response.raise_for_status()

    weather_data = response.json()

    print("\nAPI response received successfully.")
    print("Requested location:", latitude, longitude)
    print("Returned coordinates:",
          weather_data.get("latitude"),
          weather_data.get("longitude"))
    print("Timezone:", weather_data.get("timezone"))
    print("Elevation:", weather_data.get("elevation"))

    hourly_data = weather_data.get("hourly", {})
    weather_df = pd.DataFrame(hourly_data)

    print("\nReturned variables:")
    print(weather_df.columns.tolist())

    print("\nDataset shape:", weather_df.shape)

    if not weather_df.empty:
        print("\nFirst five records:")
        display(weather_df.head())

        print("\nLast timestamp returned:",
              weather_df["time"].iloc[-1])

        print("\nMissing values by variable:")
        print(weather_df.isna().sum())

except requests.exceptions.RequestException as error:
    print("API request failed:", error)

except (ValueError, KeyError) as error:
    print("Response validation failed:", error)
````

### 04.2.2.1 — Historical Weather API Test Results

#### Test Configuration

The Open-Meteo Historical Weather API was tested using a representative
Hyderabad city location.

The test requested hourly weather data for August 30, 2026,
using the Asia/Kolkata timezone.

The requested variables were:

- Temperature at 2 metres.
- Relative humidity at 2 metres.
- Precipitation.
- Rainfall.
- Wind speed at 10 metres.
- Cloud cover.

#### Actual Results

The API returned HTTP status 200, indicating that the request
was processed successfully.

The response contained 24 hourly observations, covering the requested
date from 00:00 through 23:00 in the Asia/Kolkata timezone.

All six requested weather variables were returned.

The missing-value check reported zero missing values for the
returned timestamps and weather variables.

The API returned geographical coordinates and elevation for the
selected location.

#### Interpretation

The test confirms that the Historical Weather API is accessible
from the project environment and can return hourly weather data
for the tested date and location.

The returned variables are candidates for historical weather analysis
and experiments involving synthetic hourly food demand.

The successful test does not establish complete data coverage for
the entire study period or guarantee that all variables are available
without missing values on every date.

The selected coordinates represent a provisional city-level test
location. They have not been validated as representative of each
hyperlocal zone.

#### Limitations

The response contains historical reanalysis weather rather than
archived forecasts.

Consequently, these observations must not be treated as weather
forecasts that would have been available at a historical prediction time.

Further checks are required to establish historical coverage,
variable availability, timestamp consistency, and suitability
for the intended forecasting experiments.

No complete weather dataset has been downloaded or integrated
into the demand panel during this test.

````python
import requests
import pandas as pd

# Same provisional Hyderabad test location
latitude = 17.3850
longitude = 78.4867

# Test the same date to make the comparison easier
start_date = "2026-08-30"
end_date = "2026-08-30"

url = "https://historical-forecast-api.open-meteo.com/v1/forecast"

params = {
    "latitude": latitude,
    "longitude": longitude,
    "start_date": start_date,
    "end_date": end_date,
    "hourly": [
        "temperature_2m",
        "relative_humidity_2m",
        "precipitation",
        "rain",
        "wind_speed_10m",
        "cloud_cover"
    ],
    "timezone": "Asia/Kolkata"
}

try:
    response = requests.get(url, params=params, timeout=30)

    print("HTTP status:", response.status_code)

    response.raise_for_status()

    forecast_response = response.json()

    print("\nAPI response received successfully.")
    print("Requested date:", start_date)
    print("Returned coordinates:",
          forecast_response.get("latitude"),
          forecast_response.get("longitude"))
    print("Timezone:", forecast_response.get("timezone"))

    hourly_data = forecast_response.get("hourly", {})
    forecast_df = pd.DataFrame(hourly_data)

    print("\nReturned variables:")
    print(forecast_df.columns.tolist())

    print("\nDataset shape:", forecast_df.shape)

    if not forecast_df.empty:
        print("\nFirst five records:")
        display(forecast_df.head())

        print("\nLast timestamp returned:",
              forecast_df["time"].iloc[-1])

        print("\nMissing values by variable:")
        print(forecast_df.isna().sum())

except requests.exceptions.RequestException as error:
    print("API request failed:", error)

except (ValueError, KeyError) as error:
    print("Response validation failed:", error)
````

#### Test Configuration

The requested date was August 30, 2026.

The request used the Asia/Kolkata timezone and requested hourly

#### Actual Results

The response contained 24 hourly records for the requested date,
covering 00:00 through 23:00 in the Asia/Kolkata timezone.

All six requested weather variables were returned.

The missing-value check reported zero missing values for the

#### Interpretation

weather data for the tested date and location.

The result establishes successful access for one date only.
It does not establish complete historical coverage or guarantee
that all requested variables are available for every date.

#### Forecast Archive Limitation

The Historical Forecast API constructs a continuous hourly series
by joining the early forecast hours from successive model runs.

This structure does not preserve the complete forecast horizon
of each original model run.

Consequently, the returned series must not automatically be treated
as a complete archive of forecasts available at every historical
prediction issue time.

## 04.2.3 — Archived Forecast Lead-Time Evaluation

### 1. Objective

The objective of this subsection is to evaluate whether archived weather forecasts can provide weather information at defined forecast lead times for historical food demand forecasting experiments.

The evaluation focuses on the Open-Meteo Previous Runs API and its suitability for leakage-aware backtesting.

### 2. Candidate Source

**Open-Meteo Previous Runs API**

Official documentation:
https://open-meteo.com/en/docs/previous-runs-api

The API provides weather forecasts aligned to fixed lead-time offsets.

Examples include:

- Day 0: Current model run.
- Day 1: Forecast made approximately 24 hours before the valid time.
- Day 2: Forecast made approximately 48 hours before the valid time.
- Additional lead times up to Day 7, subject to model availability.

Most models are archived from January 2024, while certain variables and models have different historical coverage.

### 3. Intended Application

The API will be evaluated for experiments that compare food demand forecasts using weather information available at different lead times.

For example, weather forecasts for a target hour may be evaluated using information from approximately one day or two days before that target hour.

This can help investigate whether forecast lead time affects demand forecasting performance.

### 4. Verification Requirements

The following will be checked:

1. API accessibility.
2. Availability of the requested historical date.
3. Availability of the required weather variables.
4. Availability of fixed-lead-time forecast fields.
5. Completeness of returned hourly records.
6. Missing values in the returned data.
7. Compatibility with the project's historical demand period.
8. Suitability for historical backtesting.

### 5. Forecast Timing and Data Leakage

A forecast's valid time is the time for which the weather is predicted. Its initialization time is when the weather model run begins.

These times must not be confused.

The Previous Runs API provides fixed lead-time offsets, but a complete historical forecasting experiment must also account for the availability of the forecast at the time the demand prediction would have been issued.

The API test alone will not establish that every historical forecast was operationally available at the exact prediction issue time.

### 6. Scope Boundary

This subsection performs a limited API availability test.

No complete weather dataset will be downloaded, and no weather data will be integrated into the demand panel.

Final source selection will depend on the actual test results and the suitability of the available forecast history.

## 04.2.6 — Weather Source Selection Decision

### Objective

The objective of this subsection is to select appropriate weather data sources for the project based on the API verification experiments, historical coverage, forecast availability, temporal resolution, and suitability for leakage-free forecasting evaluation.

The project requires weather information to support hourly food-demand forecasting across five synthetic hyperlocal zones in Hyderabad.

### 1. Summary of API Verification

Four Open-Meteo API options have been investigated.

| API | Verification Status | Primary Purpose |
|---|---|---|
| Historical Weather API | Tested successfully | Historical weather reanalysis |
| Historical Forecast API | Tested successfully | Continuous forecast-derived historical series |
| Previous Runs API | Tested successfully | Weather forecasts from earlier forecast runs |
| Live Weather Forecast API | Tested successfully | Current and future weather forecasts |

Each tested API returned HTTP 200 and the requested hourly weather variables for the test date. The live forecast test returned 72 hourly records, while the three historical API tests returned 24 hourly records each.

These results establish successful access for the tested requests only. They do not establish complete data coverage over the entire project period.

### 2. Proposed Source Selection

#### A. Historical Weather API — Historical Analysis

The Historical Weather API will be considered the primary source for historical weather analysis and exploratory analysis.

Potential variables include:

- Temperature
- Relative humidity
- Precipitation
- Rainfall
- Wind speed
- Cloud cover

Historical reanalysis data will help investigate how weather conditions correspond to demand patterns in the synthetic dataset.

However, these observations or reanalysis estimates must not be treated as weather forecasts that were necessarily available at the original prediction time.

#### B. Previous Runs API — Forecasting Experiments

The Previous Runs API will be evaluated for forecasting experiments that require weather forecasts issued before the target hour.

The dataset must preserve the distinction between:

- Forecast issue time
- Forecast valid time
- Forecast lead time

For example, a forecast for 7:00 PM issued at 3:00 PM must be distinguishable from a forecast for the same hour issued at 6:00 PM.

The API's historical coverage and the availability of the required variables must be verified before selecting a forecast lead time for backtesting.

#### C. Historical Forecast API — Supporting Analysis

The Historical Forecast API will be retained as a supporting source for examining historical forecast-derived weather series.

Because this API constructs continuous series from successive forecast runs, it must not automatically be interpreted as a complete archive of individual forecast runs.

It will not be the primary source for fixed-lead-time backtesting unless the data structure and forecast timing are shown to meet the experiment's requirements.

#### D. Live Weather Forecast API — Application Layer

The Live Weather Forecast API will be considered for the future operational application.

It can provide current and upcoming weather forecasts that may be used alongside demand predictions.

The live API response will not be substituted for historical forecasts in retrospective model evaluation.

### 3. Selected Experimental Strategy

The project will use two distinct weather-data tracks.

**Track 1 — Historical Analysis**

Use historical weather data to investigate associations between weather variables and synthetic hourly demand.

**Track 2 — Forecast-Aware Evaluation**

Use archived forecasts with identifiable issue times and valid times, where sufficient historical coverage exists, to evaluate whether weather information could realistically have been available when a demand forecast was produced.

If archived forecast coverage is insufficient for the selected historical period, this limitation will be documented. Historical reanalysis will not be silently substituted for archived forecasts in a leakage-sensitive experiment.

For the initial experiment, the project may use a documented city-level weather series as a shared contextual feature across the five synthetic zones. Zone-specific weather features will be considered only if an appropriate spatial assignment method and sufficient source resolution are available.

### 5. Data Quality and Leakage Controls

Before integrating weather data into the forecasting dataset, the following checks are required:

6. Prevent the use of future weather observations as if they were known at prediction time.

## 04.2.7 — Historical Weather Data Coverage and Completeness Assessment

### Objective

The objective of this subsection is to evaluate the temporal coverage, hourly continuity, and data completeness of the Open-Meteo Historical Weather API for the project's study period.

The required study period is January 1, 2023, through July 31, 2026. The selected location is Hyderabad, Telangana, India, using the `Asia/Kolkata` timezone.

### Data Source

- **Provider:** Open-Meteo
- **API:** Historical Weather API
- **Endpoint:** https://archive-api.open-meteo.com/v1/archive
- **Requested Latitude:** 17.3850
- **Requested Longitude:** 78.4867
- **Timezone:** Asia/Kolkata

### Variables

The following hourly weather variables will be examined:

1. Temperature at 2 metres
2. Relative humidity
3. Precipitation
4. Rainfall
5. Wind speed at 10 metres
6. Cloud cover

### Assessment Procedure

The assessment will:

1. Request historical weather data in manageable monthly batches.
2. record successful and unsuccessful API requests.
3. Count the returned hourly records.
4. Compare the returned timestamps with the expected hourly timeline.
5. Identify missing timestamps and missing variable values.
6. Check for duplicate timestamps.
7. Summarize coverage by month.
8. Determine whether the retrieved data is suitable for further processing.

### Expected Outcome

The assessment will provide evidence of the API's availability and completeness over the requested period.

A successful API response alone will not be considered sufficient proof of complete coverage. Actual timestamps, record counts, duplicate records, missing values, and failed requests will be examined.

### Important Limitations

The API returns gridded historical weather estimates, not independent weather observations for each synthetic delivery zone.

The initial assessment uses one Hyderabad coordinate. The returned grid-cell coordinates will be recorded because they may differ from the requested coordinates.

This assessment evaluates historical weather data only. It does not establish that the same information was available as a forecast at the original demand-prediction time.

````python
# ============================================================
# 04.2.7 — Historical Weather Coverage Assessment
# ============================================================

import time
import requests
import pandas as pd
from IPython.display import display

# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

API_URL = "https://archive-api.open-meteo.com/v1/archive"

LATITUDE = 17.3850
LONGITUDE = 78.4867
TIMEZONE = "Asia/Kolkata"

START_DATE = "2023-01-01"
END_DATE = "2026-07-31"

WEATHER_VARIABLES = [
    "temperature_2m",
    "relative_humidity_2m",
    "precipitation",
    "rain",
    "wind_speed_10m",
    "cloud_cover",
]

EXPECTED_COLUMNS = ["time"] + WEATHER_VARIABLES

session = requests.Session()

# ------------------------------------------------------------
# 2. Generate monthly request periods
# ------------------------------------------------------------

start = pd.Timestamp(START_DATE)
end = pd.Timestamp(END_DATE)

month_starts = pd.date_range(
    start=start.replace(day=1),
    end=end.replace(day=1),
    freq="MS"
)

monthly_periods = []

for month_start in month_starts:
    month_end = month_start + pd.offsets.MonthEnd(0)

    period_start = max(month_start, start)
    period_end = min(month_end, end)

    monthly_periods.append(
        (
            period_start.strftime("%Y-%m-%d"),
            period_end.strftime("%Y-%m-%d")
        )
    )

print("Assessment period:", START_DATE, "to", END_DATE)
print("Monthly API requests:", len(monthly_periods))
print("Weather variables:", len(WEATHER_VARIABLES))

# ------------------------------------------------------------
# 3. Retrieve data month by month
# ------------------------------------------------------------

monthly_data = []
request_log = []

for period_start, period_end in monthly_periods:

    params = {
        "latitude": LATITUDE,
        "longitude": LONGITUDE,
        "start_date": period_start,
        "end_date": period_end,
        "hourly": WEATHER_VARIABLES,
        "timezone": TIMEZONE,
    }

    try:
        response = session.get(
            API_URL,
            params=params,
            timeout=60
        )

        response.raise_for_status()
        payload = response.json()

        if "hourly" not in payload or "time" not in payload["hourly"]:
            raise ValueError("API response does not contain hourly timestamps.")

        month_df = pd.DataFrame(payload["hourly"])

        missing_columns = [
            column for column in EXPECTED_COLUMNS
            if column not in month_df.columns
        ]

        if missing_columns:
            raise ValueError(
                f"Missing expected columns: {missing_columns}"
            )

        month_df = month_df[EXPECTED_COLUMNS].copy()
        month_df["time"] = pd.to_datetime(month_df["time"])

        monthly_data.append(month_df)

        request_log.append({
            "period_start": period_start,
            "period_end": period_end,
            "status": "SUCCESS",
            "http_status": response.status_code,
            "records": len(month_df),
            "returned_latitude": payload.get("latitude"),
            "returned_longitude": payload.get("longitude"),
            "error": None,
        })

        print(
            f"SUCCESS | {period_start} to {period_end} "
            f"| {len(month_df):,} hourly records"
        )

    except Exception as exc:
        request_log.append({
            "period_start": period_start,
            "period_end": period_end,
            "status": "FAILED",
            "http_status": (
                response.status_code
                if "response" in locals()
                else None
            ),
            "records": 0,
            "returned_latitude": None,
            "returned_longitude": None,
            "error": str(exc),
        })

        print(
            f"FAILED  | {period_start} to {period_end} "
            f"| {type(exc).__name__}: {exc}"
        )

    # Small pause between requests
    time.sleep(0.2)

# ------------------------------------------------------------
# 4. Build the assessment tables
# ------------------------------------------------------------

request_log_df = pd.DataFrame(request_log)

successful_months = request_log_df[
    request_log_df["status"] == "SUCCESS"
]

failed_months = request_log_df[
    request_log_df["status"] == "FAILED"
]

print("\n--- Request Summary ---")
print("Total monthly requests:", len(request_log_df))
print("Successful requests:", len(successful_months))
print("Failed requests:", len(failed_months))

display(request_log_df)

# ------------------------------------------------------------
# 5. Combine successful monthly responses
# ------------------------------------------------------------

if not monthly_data:
    raise RuntimeError(
        "No monthly weather data was retrieved. "
        "Inspect request_log_df before proceeding."
    )

weather_coverage_df = pd.concat(
    monthly_data,
    ignore_index=True
)

weather_coverage_df = weather_coverage_df.sort_values(
    "time"
).reset_index(drop=True)

print("\n--- Retrieved Data Summary ---")
print("Shape:", weather_coverage_df.shape)
print("First timestamp:", weather_coverage_df["time"].min())
print("Last timestamp:", weather_coverage_df["time"].max())

print("\n--- Missing Values by Column ---")
display(weather_coverage_df.isna().sum().to_frame("missing_values"))

# ------------------------------------------------------------
# 6. Check duplicate timestamps
# ------------------------------------------------------------

duplicate_count = weather_coverage_df["time"].duplicated().sum()

print("\n--- Duplicate Timestamp Check ---")
print("Duplicate timestamps:", duplicate_count)

# ------------------------------------------------------------
# 7. Check expected hourly timeline
# ------------------------------------------------------------

expected_timeline = pd.date_range(
    start=pd.Timestamp(START_DATE),
    end=pd.Timestamp(END_DATE) + pd.Timedelta(hours=23),
    freq="h"
)

actual_timeline = pd.DatetimeIndex(
    weather_coverage_df["time"].drop_duplicates().sort_values()
)

missing_timestamps = expected_timeline.difference(actual_timeline)
unexpected_timestamps = actual_timeline.difference(expected_timeline)

print("\n--- Hourly Timeline Check ---")
print("Expected hourly timestamps:", len(expected_timeline))
print("Unique returned timestamps:", len(actual_timeline))
print("Missing timestamps:", len(missing_timestamps))
print("Unexpected timestamps:", len(unexpected_timestamps))

if len(missing_timestamps) > 0:
    print("\nFirst 20 missing timestamps:")
    print(missing_timestamps[:20].tolist())

if len(unexpected_timestamps) > 0:
    print("\nFirst 20 unexpected timestamps:")
    print(unexpected_timestamps[:20].tolist())

# ------------------------------------------------------------
# 8. Summarize actual monthly coverage
# ------------------------------------------------------------

weather_coverage_df["month"] = (
    weather_coverage_df["time"].dt.to_period("M").astype(str)
)

monthly_coverage = (
    weather_coverage_df
    .groupby("month")
    .agg(
        returned_records=("time", "size"),
        unique_timestamps=("time", "nunique"),
    )
    .reset_index()
)

monthly_coverage["expected_records"] = (
    monthly_coverage["month"]
    .apply(
        lambda month: len(
            pd.date_range(
                start=pd.Period(month, freq="M").start_time,
                end=pd.Period(month, freq="M").end_time.floor("h"),
                freq="h"
            )
        )
    )
)

monthly_coverage["record_difference"] = (
    monthly_coverage["returned_records"]
    - monthly_coverage["expected_records"]
)

print("\n--- Monthly Coverage Summary ---")
display(monthly_coverage)

# ------------------------------------------------------------
# 9. Final preliminary assessment
# ------------------------------------------------------------

all_requests_succeeded = len(failed_months) == 0
no_duplicate_timestamps = duplicate_count == 0
no_missing_timestamps = len(missing_timestamps) == 0
no_missing_values = weather_coverage_df[
    WEATHER_VARIABLES
].isna().sum().sum() == 0

print("\n--- Preliminary Assessment ---")
print("All monthly requests succeeded:", all_requests_succeeded)
print("No duplicate timestamps:", no_duplicate_timestamps)
print("No missing timestamps:", no_missing_timestamps)
print("No missing weather values:", no_missing_values)

print(
    "\nNote: This assessment checks the returned data. "
    "Review monthly coverage and failed requests before "
    "deciding whether the dataset is ready for integration."
)
````

````python
# ============================================================
# 04.2.11 — Final Saved Weather Dataset Validation
# ============================================================

import json
import pandas as pd

# Expected saved files
expected_files = {
    "Weather CSV": (
        project_root / "data" / "raw" / "weather"
        / "hyderabad_historical_weather.csv"
    ),
    "Metadata JSON": (
        project_root / "data" / "metadata"
        / "weather_source_metadata.json"
    ),
    "Monthly Coverage": (
        project_root / "results" / "weather_coverage"
        / "monthly_coverage.csv"
    ),
    "Request Log": (
        project_root / "results" / "weather_coverage"
        / "request_log.csv"
    ),
}

print("--- File Existence Check ---")

for name, path in expected_files.items():
    print(f"{name}: {'EXISTS' if path.is_file() else 'MISSING'}")

missing_files = [
    name for name, path in expected_files.items()
    if not path.is_file()
]

if missing_files:
    # The request log may have been saved under a timestamped filename.
    # Re-save the in-memory final log to the canonical path expected below.
    if "Request Log" in missing_files and "request_log_df_final" in globals():
        request_log_path = expected_files["Request Log"]
        request_log_path.parent.mkdir(parents=True, exist_ok=True)
        request_log_df_final.to_csv(request_log_path, index=False)

    # Recheck after attempting to restore the request log.
    missing_files = [
        name for name, path in expected_files.items()
        if not path.is_file()
    ]

    if missing_files:
        raise FileNotFoundError(
            f"Required files are missing: {missing_files}"
        )

# ------------------------------------------------------------
# Load saved files independently
# ------------------------------------------------------------

weather_saved = pd.read_csv(
    expected_files["Weather CSV"],
    parse_dates=["time"],
)

monthly_saved = pd.read_csv(
    expected_files["Monthly Coverage"]
)

request_log_saved = pd.read_csv(
    expected_files["Request Log"]
)

with expected_files["Metadata JSON"].open(
    "r", encoding="utf-8"
) as file:
    metadata_saved = json.load(file)

# ------------------------------------------------------------
# Validate expected structure
# ------------------------------------------------------------

expected_columns = [
    "time",
    "temperature_2m",
    "relative_humidity_2m",
    "precipitation",
    "rain",
    "wind_speed_10m",
    "cloud_cover",
]

expected_hours = 31392

print("\n--- Dataset Structure ---")
print("Shape:", weather_saved.shape)
print("Columns:", weather_saved.columns.tolist())
print("Expected rows:", expected_hours)
print("Actual rows:", len(weather_saved))
print("Expected columns:", len(expected_columns))
print("Actual columns:", len(weather_saved.columns))

print("\n--- Timestamp Validation ---")
print("First timestamp:", weather_saved["time"].min())
print("Last timestamp:", weather_saved["time"].max())
print("Duplicate timestamps:", weather_saved["time"].duplicated().sum())

expected_timeline = pd.date_range(
    start="2023-01-01 00:00:00",
    end="2026-07-31 23:00:00",
    freq="h",
)

actual_timeline = pd.DatetimeIndex(
    weather_saved["time"].sort_values()
)

missing_timestamps = expected_timeline.difference(actual_timeline)
unexpected_timestamps = actual_timeline.difference(expected_timeline)

print("Expected hourly timestamps:", len(expected_timeline))
print("Missing timestamps:", len(missing_timestamps))
print("Unexpected timestamps:", len(unexpected_timestamps))

print("\n--- Missing Values ---")
display(weather_saved.isna().sum().to_frame("missing_values"))

print("\n--- Supporting File Checks ---")
print("Monthly coverage rows:", len(monthly_saved))
print("Request log rows:", len(request_log_saved))
print("Successful requests:", (
    request_log_saved["status"] == "SUCCESS"
).sum())
print("Failed requests:", (
    request_log_saved["status"] == "FAILED"
).sum())

print("\n--- Metadata Summary ---")
print("Provider:", metadata_saved.get("provider"))
print("Period:", metadata_saved.get("start_date"),
      "to", metadata_saved.get("end_date"))
print("Timezone:", metadata_saved.get("timezone"))
print("Metadata record count:", metadata_saved.get("record_count"))

# ------------------------------------------------------------
# Final validation flags
# ------------------------------------------------------------

checks = {
    "All expected files exist": not missing_files,
    "Expected row count": len(weather_saved) == expected_hours,
    "Expected columns": weather_saved.columns.tolist() == expected_columns,
    "No missing values": not weather_saved.isna().any().any(),
    "No duplicate timestamps": not weather_saved["time"].duplicated().any(),
    "Complete hourly timeline": (
        len(missing_timestamps) == 0
        and len(unexpected_timestamps) == 0
    ),
    "43 monthly coverage records": len(monthly_saved) == 43,
    "43 successful requests": (
        len(request_log_saved) == 43
        and (request_log_saved["status"] == "SUCCESS").all()
    ),
    "Metadata row count matches": (
        metadata_saved.get("record_count") == len(weather_saved)
    ),
}

print("\n--- Final Validation Results ---")

for check, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL'} | {check}")

print(
    "\nOverall status:",
    "PASS" if all(checks.values()) else "REQUIRES INVESTIGATION"
)
````

## 04.3.1 — Geographical Definition and Weather-Source Evaluation

### Objective

This subsection establishes a geographically defensible approach for collecting historical weather data for the five hyperlocal food-demand zones in Hyderabad: Kondapur, Hitech City, Gachibowli, Madhapur, and Kukatpally.

The objective is to identify real weather observation stations and suitable historical weather datasets that can represent local weather conditions at an hourly resolution.

### Geographical Scope

The study covers five synthetic demand zones associated with real geographical areas in Hyderabad:

* Z01 — Kondapur
* Z02 — Hitech City
* Z03 — Gachibowli
* Z04 — Madhapur
* Z05 — Kukatpally

The geographical boundaries of these zones must be established or documented before selecting representative observation locations. Zone names alone are insufficient to determine exact boundaries.

### Candidate Weather Sources

The initial investigation includes:

1. Telangana Development Planning Society automatic weather stations.
2. Telangana hourly rainfall telemetry data available through the National Water Data Portal.
3. India Meteorological Department weather observation and rainfall datasets.
4. Open-Meteo historical weather data as a potential source of gridded historical estimates.

Each source will be evaluated for geographical coverage, historical availability, temporal resolution, measurement or estimation methodology, data accessibility, and reproducibility.

### Data Authenticity and Spatial Validation

Weather observations and model-derived estimates will be distinguished explicitly. Station observations will retain their original station identifiers and geographical coordinates. Gridded datasets will retain the requested coordinates and the actual grid coordinates or identifiers reported by the provider, where available.

Multiple requested locations will not automatically be treated as independent weather observations. Locations that resolve to the same underlying grid cell will be identified before any spatial aggregation is performed.

No missing rainfall values will be replaced with invented measurements. Missing observations will remain explicitly identified, and any aggregation will be based on documented data-availability and spatial-coverage rules.

### Temporal Scope

The target historical period is January 1, 2023, through July 31, 2026, in the Asia/Kolkata time zone, consistent with the existing historical weather collection.

Hourly station observations are preferred for analysing local rainfall variation. Where only daily observations or gridded estimates are available, their limitations will be documented rather than representing them as equivalent to hourly local measurements.

### Expected Outcome

The subsection will produce a documented assessment of candidate weather sources, available station locations, geographical coverage, historical data availability, and spatial limitations.

The final source selection will be made only after the available evidence demonstrates which datasets can support the project's hyperlocal weather requirements.

## 04.3.3 — API Evaluation for Hyperlocal Historical Weather

### Objective

This subsection evaluates weather APIs for collecting historical weather information relevant to the five Hyderabad demand zones: Kondapur, Hitech City, Gachibowli, Madhapur, and Kukatpally.

The target period is January 1, 2023, through July 31, 2026, with hourly records interpreted in the Asia/Kolkata time zone.

### Candidate APIs

**Open-Meteo Historical Weather API**

This API provides historical hourly weather variables using multiple weather reanalysis and model datasets. Its support for multiple coordinates makes it useful for evaluating spatial coverage. Returned grid coordinates and dataset resolution must be checked before interpreting nearby locations as independent weather observations.

**Meteostat**

Meteostat provides historical weather data through station-based and geographical-point interfaces. Station metadata and record provenance must be examined to distinguish direct station observations from model-filled records.

**WeatherAPI.com**

WeatherAPI.com provides historical hourly weather through a subscription-dependent API. Its documentation identifies the historical archive as forecast-derived data, so it will not automatically be treated as a source of direct historical observations.

### Evaluation Criteria

Each API will be evaluated using the following criteria:

1. Availability of records throughout the target historical period.
2. Availability of hourly rainfall, temperature, humidity, wind, and cloud-cover variables.
3. Geographical relevance to each of the five zones.
4. Spatial resolution and the possibility of duplicate underlying grid cells.
5. Provenance of each variable and whether values are observed or model-derived.
6. Missing timestamps, missing values, and data-quality limitations.
7. API access requirements, usage limits, licensing, and reproducibility.

### Data Integrity Rules

No weather values will be fabricated or statistically generated to fill missing observations.

Missing values will be identified and documented. Any model-derived values will be labelled as estimates, and direct station observations will be distinguished from reanalysis or forecast-derived products.

Multiple coordinate queries will not be treated as independent measurements unless the underlying source resolution supports that interpretation.

### Expected Outcome

The evaluation will identify which API or combination of APIs offers the strongest verifiable coverage for the project. Final source selection will depend on actual station availability, spatial resolution, historical completeness, and data provenance rather than convenience alone.


## 04.3.4 — Historical Weather API Pilot Test

### Objective

This subsection tests whether the Open-Meteo Historical Weather API can return hourly weather estimates for multiple geographical locations associated with the five Hyderabad demand zones.

The pilot will examine API accessibility, returned coordinates, hourly variable availability, missing values, and spatial differences between locations.

### Important Data Distinction

Open-Meteo's Historical Weather API provides historical weather estimates derived from reanalysis datasets and weather models. These estimates must not be described as direct weather-station observations.

The pilot is intended to evaluate spatial resolution and API behaviour. The test coordinates are provisional representatives for testing only; they do not define official zone boundaries.

### Pilot Procedure

1. Submit a historical weather request for provisional coordinates associated with the five zones.
2. Retrieve temperature, relative humidity, precipitation, rainfall, wind speed, and cloud cover.
3. Record the coordinates returned by the API.
4. Check whether different requested locations resolve to different weather grid coordinates.
5. Compare hourly rainfall values across locations.
6. Validate timestamps, missing values, and response completeness.

### Decision Rule

Successful API responses alone will not establish suitability for hyperlocal forecasting. Final selection will depend on geographical relevance, spatial resolution, historical coverage, data provenance, and the ability to represent meaningful local weather variation.


````python
# Phase 04.3.4 — Open-Meteo API Spatial Resolution Pilot

import requests
import pandas as pd
from IPython.display import display

# --------------------------------------------------
# 1. API configuration
# --------------------------------------------------

API_URL = "https://archive-api.open-meteo.com/v1/archive"

TEST_DATE = "2026-08-30"

# Provisional representative points for API testing only.
# These are NOT verified zone boundaries or official station locations.

test_locations = {
    "Z01_Kondapur": {
        "latitude": 17.4650,
        "longitude": 78.3560,
    },
    "Z02_Hitech_City": {
        "latitude": 17.4430,
        "longitude": 78.3770,
    },
    "Z03_Gachibowli": {
        "latitude": 17.4400,
        "longitude": 78.3480,
    },
    "Z04_Madhapur": {
        "latitude": 17.4480,
        "longitude": 78.3910,
    },
    "Z05_Kukatpally": {
        "latitude": 17.4930,
        "longitude": 78.3990,
    },
}

weather_variables = [
    "temperature_2m",
    "relative_humidity_2m",
    "precipitation",
    "rain",
    "wind_speed_10m",
    "cloud_cover",
]

# --------------------------------------------------
# 2. Prepare multi-location API request
# --------------------------------------------------

params = {
    "latitude": ",".join(
        str(location["latitude"])
        for location in test_locations.values()
    ),
    "longitude": ",".join(
        str(location["longitude"])
        for location in test_locations.values()
    ),
    "start_date": TEST_DATE,
    "end_date": TEST_DATE,
    "hourly": ",".join(weather_variables),
    "timezone": "Asia/Kolkata",
    "timeformat": "iso8601",
}

print("Requesting historical weather data...")
print("Test date:", TEST_DATE)
print("Number of requested locations:", len(test_locations))

response = requests.get(
    API_URL,
    params=params,
    timeout=120,
)

print("HTTP status:", response.status_code)
response.raise_for_status()

payload = response.json()

# --------------------------------------------------
# 3. Normalize the API response
# --------------------------------------------------

# Multiple coordinates normally produce a list of
# location responses. Handle both response structures.

if isinstance(payload, dict):
    if "hourly" in payload:
        location_responses = [payload]
    else:
        raise ValueError(
            "Unexpected API response structure. "
            f"Returned keys: {list(payload.keys())}"
        )

elif isinstance(payload, list):
    location_responses = payload

else:
    raise TypeError(
        f"Unexpected response type: {type(payload)}"
    )

if len(location_responses) != len(test_locations):
    print(
        "WARNING: Requested location count and "
        "returned response count differ."
    )

# --------------------------------------------------
# 4. Build location metadata and hourly records
# --------------------------------------------------

metadata_rows = []
weather_frames = []

for location_name, location_response in zip(
    test_locations.keys(),
    location_responses,
):
    hourly = location_response.get("hourly", {})

    if "time" not in hourly:
        print(f"WARNING: No hourly timestamps for {location_name}")
        continue

    frame = pd.DataFrame(hourly)
    frame.insert(0, "zone_test_id", location_name)

    requested = test_locations[location_name]

    metadata_rows.append({
        "zone_test_id": location_name,
        "requested_latitude": requested["latitude"],
        "requested_longitude": requested["longitude"],
        "returned_latitude": location_response.get("latitude"),
        "returned_longitude": location_response.get("longitude"),
        "elevation_m": location_response.get("elevation"),
        "timezone": location_response.get("timezone"),
        "hourly_records": len(frame),
    })

    weather_frames.append(frame)

if not weather_frames:
    raise ValueError("No hourly weather records were returned.")

weather_test = pd.concat(
    weather_frames,
    ignore_index=True,
)

location_metadata = pd.DataFrame(metadata_rows)

# --------------------------------------------------
# 5. Inspect returned coordinates
# --------------------------------------------------

print("\n--- Requested vs. Returned Coordinates ---")

display(location_metadata)

# --------------------------------------------------
# 6. Check timestamps and missing values
# --------------------------------------------------

print("\n--- Record Counts by Test Location ---")

display(
    weather_test.groupby("zone_test_id")
    .agg(
        record_count=("time", "size"),
        first_timestamp=("time", "min"),
        last_timestamp=("time", "max"),
    )
    .reset_index()
)

print("\n--- Missing Values by Test Location ---")

missing_summary = (
    weather_test.groupby("zone_test_id")[weather_variables]
    .apply(lambda group: group.isna().sum())
)

display(missing_summary)

# --------------------------------------------------
# 7. Compare hourly rainfall
# --------------------------------------------------

print("\n--- Hourly Rainfall Comparison ---")

rainfall_comparison = weather_test.pivot(
    index="time",
    columns="zone_test_id",
    values="rain",
)

display(rainfall_comparison)

# --------------------------------------------------
# 8. Check unique returned grid coordinates
# --------------------------------------------------

unique_grid_cells = location_metadata[
    ["returned_latitude", "returned_longitude"]
].drop_duplicates()

print("\nNumber of requested locations:", len(test_locations))
print("Number of successful location responses:", len(location_metadata))
print("Number of distinct returned grid coordinates:", len(unique_grid_cells))

print("\n--- Distinct Returned Grid Coordinates ---")
display(unique_grid_cells.reset_index(drop=True))
````

## 04.3.5 — ECMWF IFS Historical Weather Spatial-Resolution Test

### Objective
Evaluate whether the ECMWF IFS historical weather dataset provides distinct grid-cell coordinates for the five provisional Hyderabad zone locations.

### Dataset
ECMWF IFS historical weather data through the Open-Meteo Historical Weather API.

The provider documents approximately 9 km spatial resolution and hourly historical data availability from 2017 onward.

### Validation
1. Explicitly select the ECMWF IFS model.
2. Request hourly weather for the same five provisional locations.
3. Record requested and returned coordinates.
4. Count the distinct returned grid cells.
5. Compare hourly rainfall values.
6. Check timestamps and missing values.

### Limitation
A nominal grid resolution of approximately 9 km does not guarantee that every zone will receive a distinct grid cell. Results must be verified from the actual API response.

The requested coordinates are provisional test points, not verified zone boundaries or weather-station locations.

````python
# Phase 04.3.5 — ECMWF IFS Spatial-Resolution Test

import requests
import pandas as pd
from IPython.display import display

API_URL = "https://archive-api.open-meteo.com/v1/archive"
TEST_DATE = "2026-08-30"

# Same provisional test points used in Phase 04.3.4
test_locations = {
    "Z01_Kondapur": (17.4650, 78.3560),
    "Z02_Hitech_City": (17.4430, 78.3770),
    "Z03_Gachibowli": (17.4400, 78.3480),
    "Z04_Madhapur": (17.4480, 78.3910),
    "Z05_Kukatpally": (17.4930, 78.3990),
}

weather_variables = [
    "temperature_2m",
    "relative_humidity_2m",
    "precipitation",
    "rain",
    "wind_speed_10m",
    "cloud_cover",
]

params = {
    "latitude": ",".join(str(v[0]) for v in test_locations.values()),
    "longitude": ",".join(str(v[1]) for v in test_locations.values()),
    "start_date": TEST_DATE,
    "end_date": TEST_DATE,
    "hourly": ",".join(weather_variables),
    "timezone": "Asia/Kolkata",
    "models": "ecmwf_ifs",
}

print("Dataset requested: ECMWF IFS")
print("Test date:", TEST_DATE)
print("Requested locations:", len(test_locations))

response = requests.get(API_URL, params=params, timeout=120)

print("HTTP status:", response.status_code)
response.raise_for_status()

payload = response.json()

if isinstance(payload, dict):
    location_responses = [payload]
elif isinstance(payload, list):
    location_responses = payload
else:
    raise TypeError(f"Unexpected response type: {type(payload)}")

if len(location_responses) != len(test_locations):
    print(
        f"WARNING: Requested {len(test_locations)} locations, "
        f"but received {len(location_responses)} responses."
    )

metadata_rows = []
weather_frames = []

for (zone_id, (requested_lat, requested_lon)), item in zip(
    test_locations.items(), location_responses
):
    hourly = item.get("hourly", {})

    if "time" not in hourly:
        print(f"WARNING: No hourly data returned for {zone_id}")
        continue

    frame = pd.DataFrame(hourly)
    frame.insert(0, "zone_test_id", zone_id)
    weather_frames.append(frame)

    metadata_rows.append({
        "zone_test_id": zone_id,
        "requested_latitude": requested_lat,
        "requested_longitude": requested_lon,
        "returned_latitude": item.get("latitude"),
        "returned_longitude": item.get("longitude"),
        "elevation_m": item.get("elevation"),
        "timezone": item.get("timezone"),
        "hourly_records": len(frame),
    })

if not weather_frames:
    raise ValueError("No hourly records returned.")

weather_test_ifs = pd.concat(weather_frames, ignore_index=True)
metadata_ifs = pd.DataFrame(metadata_rows)

print("\n--- Requested vs Returned Coordinates ---")
display(metadata_ifs)

print("\n--- Record Counts and Timestamp Coverage ---")
display(
    weather_test_ifs.groupby("zone_test_id")
    .agg(
        record_count=("time", "size"),
        first_timestamp=("time", "min"),
        last_timestamp=("time", "max"),
    )
    .reset_index()
)

print("\n--- Missing Values by Location ---")
display(
    weather_test_ifs.groupby("zone_test_id")[weather_variables]
    .apply(lambda group: group.isna().sum())
)

print("\n--- Hourly Rainfall Comparison ---")
display(
    weather_test_ifs.pivot(
        index="time",
        columns="zone_test_id",
        values="rain",
    )
)

unique_grid_cells = metadata_ifs[
    ["returned_latitude", "returned_longitude"]
].drop_duplicates()

print("\nRequested locations:", len(test_locations))
print("Successful responses:", len(metadata_ifs))
print("Distinct returned grid coordinates:", len(unique_grid_cells))

print("\n--- Distinct Grid Coordinates ---")
display(unique_grid_cells.reset_index(drop=True))
````

````python
# ============================================================
# PHASE 04: OPEN-METEO HISTORICAL WEATHER COLLECTION
# Coordinates: 17.4578, 78.3742
# ============================================================

import json
import requests
import pandas as pd

from pathlib import Path
from datetime import datetime, timezone

# ------------------------------------------------------------
# 1. PROJECT PATHS
# ------------------------------------------------------------

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data"

WEATHER_RAW_DIR = DATA_DIR / "raw" / "weather"
WEATHER_METADATA_DIR = DATA_DIR / "metadata"

WEATHER_RAW_DIR.mkdir(parents=True, exist_ok=True)
WEATHER_METADATA_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2. REQUESTED COORDINATES
# ------------------------------------------------------------

LATITUDE = 17.4578
LONGITUDE = 78.3742

# ------------------------------------------------------------
# 3. HISTORICAL PERIOD AND TIMEZONE
# ------------------------------------------------------------

START_DATE = "2023-01-01"
END_DATE = "2026-07-31"

TIMEZONE = "Asia/Kolkata"

API_URL = "https://archive-api.open-meteo.com/v1/archive"

# ------------------------------------------------------------
# 4. WEATHER VARIABLES
# ------------------------------------------------------------

HOURLY_VARIABLES = [
    "temperature_2m",
    "relative_humidity_2m",
    "apparent_temperature",
    "precipitation",
    "rain",
    "wind_speed_10m",
    "wind_direction_10m",
    "wind_gusts_10m",
    "cloud_cover",
    "surface_pressure",
    "dew_point_2m",
]

# ------------------------------------------------------------
# 5. COORDINATE-SPECIFIC OUTPUT FILES
# ------------------------------------------------------------

WEATHER_CSV_PATH = (
    WEATHER_RAW_DIR
    / "hyderabad_historical_weather_174578_783742.csv"
)

WEATHER_METADATA_PATH = (
    WEATHER_METADATA_DIR
    / "weather_source_metadata_174578_783742.json"
)

# Reuse the finalized historical weather files instead of recollecting them.
if WEATHER_CSV_PATH.exists():
    weather_df = pd.read_csv(WEATHER_CSV_PATH, parse_dates=["timestamp"])
    print("Existing historical weather dataset reused; API collection skipped.")
    print("Dataset:", WEATHER_CSV_PATH)
    print("Shape:", weather_df.shape)
else:
    print("=" * 65)
    print("OPEN-METEO HISTORICAL WEATHER COLLECTION")
    print("=" * 65)

    print(f"Requested latitude : {LATITUDE}")
    print(f"Requested longitude: {LONGITUDE}")
    print(f"Start date         : {START_DATE}")
    print(f"End date           : {END_DATE}")
    print(f"Timezone           : {TIMEZONE}")

    # ------------------------------------------------------------
    # 6. REQUEST HISTORICAL WEATHER
    # ------------------------------------------------------------

    params = {
        "latitude": LATITUDE,
        "longitude": LONGITUDE,
        "start_date": START_DATE,
        "end_date": END_DATE,
        "hourly": ",".join(HOURLY_VARIABLES),
        "timezone": TIMEZONE,
        "wind_speed_unit": "kmh",
        "precipitation_unit": "mm",
    }

    print("\nRequesting data from Open-Meteo...")

    response = requests.get(
        API_URL,
        params=params,
        timeout=180,
    )

    response.raise_for_status()

    api_data = response.json()

    if "hourly" not in api_data:
        raise RuntimeError(
            f"Unexpected API response: {api_data}"
        )

    # ------------------------------------------------------------
    # 7. RECORD ACTUAL RETURNED COORDINATES
    # ------------------------------------------------------------

    returned_latitude = api_data["latitude"]
    returned_longitude = api_data["longitude"]
    returned_elevation = api_data.get("elevation")

    print("\nAPI response received.")
    print("Returned latitude :", returned_latitude)
    print("Returned longitude:", returned_longitude)
    print("Returned elevation:", returned_elevation)

    # ------------------------------------------------------------
    # 8. BUILD DATAFRAME
    # ------------------------------------------------------------

    weather_df = pd.DataFrame(api_data["hourly"])

    weather_df.rename(
        columns={"time": "timestamp"},
        inplace=True,
    )

    weather_df["timestamp"] = pd.to_datetime(
        weather_df["timestamp"],
        errors="raise",
    )

    weather_df = (
        weather_df.sort_values("timestamp")
        .reset_index(drop=True)
    )

    # ------------------------------------------------------------
    # 9. VALIDATE THE HOURLY TIMESTAMPS
    # ------------------------------------------------------------

    expected_timestamps = pd.date_range(
        start=pd.Timestamp(START_DATE),
        end=pd.Timestamp(END_DATE) + pd.Timedelta(days=1),
        freq="h",
        inclusive="left",
    )

    expected_count = len(expected_timestamps)
    actual_count = len(weather_df)

    unique_count = weather_df["timestamp"].nunique()

    duplicate_count = int(
        weather_df["timestamp"].duplicated().sum()
    )

    missing_timestamps = expected_timestamps.difference(
        pd.DatetimeIndex(weather_df["timestamp"])
    )

    unexpected_timestamps = pd.DatetimeIndex(
        weather_df["timestamp"]
    ).difference(expected_timestamps)

    missing_values = (
        weather_df.drop(columns=["timestamp"])
        .isna()
        .sum()
        .astype(int)
        .to_dict()
    )

    print("\n" + "=" * 65)
    print("DATA VALIDATION")
    print("=" * 65)

    print("Expected records       :", expected_count)
    print("Actual records         :", actual_count)
    print("Unique timestamps      :", unique_count)
    print("Duplicate timestamps   :", duplicate_count)
    print("Missing timestamps     :", len(missing_timestamps))
    print("Unexpected timestamps  :", len(unexpected_timestamps))
    print("Missing values         :", missing_values)

    # Do not save an incomplete or structurally invalid dataset.
    if actual_count != expected_count:
        raise ValueError(
            f"Expected {expected_count} rows, received {actual_count}."
        )

    if unique_count != expected_count:
        raise ValueError("Timestamp uniqueness validation failed.")

    if duplicate_count != 0:
        raise ValueError("Duplicate timestamps were found.")

    if len(missing_timestamps) != 0:
        raise ValueError("Missing hourly timestamps were found.")

    if len(unexpected_timestamps) != 0:
        raise ValueError("Unexpected timestamps were found.")

    if any(count > 0 for count in missing_values.values()):
        raise ValueError(
            "Missing weather values detected. "
            "Investigate before accepting the dataset."
        )

    # ------------------------------------------------------------
    # 10. PREPARE METADATA
    # ------------------------------------------------------------

    metadata = {
        "provider": "Open-Meteo",
        "api_endpoint": API_URL,
        "retrieval_timestamp_utc": datetime.now(
            timezone.utc
        ).isoformat(),
        "requested_location": {
            "latitude": LATITUDE,
            "longitude": LONGITUDE,
        },
        "returned_location": {
            "returned_latitude": returned_latitude,
            "returned_longitude": returned_longitude,
            "elevation_m": returned_elevation,
            "timezone": api_data.get("timezone"),
            "timezone_abbreviation": api_data.get(
                "timezone_abbreviation"
            ),
            "utc_offset_seconds": api_data.get(
                "utc_offset_seconds"
            ),
        },
        "timezone": TIMEZONE,
        "start_date": START_DATE,
        "end_date": END_DATE,
        "temporal_resolution": "Hourly",
        "expected_record_count": expected_count,
        "record_count": actual_count,
        "unique_timestamp_count": unique_count,
        "duplicate_timestamp_count": duplicate_count,
        "missing_timestamp_count": len(missing_timestamps),
        "unexpected_timestamp_count": len(unexpected_timestamps),
        "missing_values_by_variable": missing_values,
        "variables": HOURLY_VARIABLES,
        "units": {
            "temperature_2m": "°C",
            "relative_humidity_2m": "%",
            "apparent_temperature": "°C",
            "precipitation": "mm",
            "rain": "mm",
            "wind_speed_10m": "km/h",
            "wind_direction_10m": "degrees",
            "wind_gusts_10m": "km/h",
            "cloud_cover": "%",
            "surface_pressure": "hPa",
            "dew_point_2m": "°C",
        },
        "limitations": [
            "Weather data represent a selected grid location.",
            "Historical weather estimates are not archived forecasts.",
            "Returned coordinates may differ from requested coordinates.",
        ],
        "source_documentation": (
            "https://open-meteo.com/en/docs/historical-weather-api"
        ),
    }

    # ------------------------------------------------------------
    # 11. SAVE THE VALIDATED DATASET
    # ------------------------------------------------------------

    weather_df.to_csv(
        WEATHER_CSV_PATH,
        index=False,
    )

    with open(
        WEATHER_METADATA_PATH,
        "w",
        encoding="utf-8",
    ) as f:
        json.dump(metadata, f, indent=4)

    # ------------------------------------------------------------
    # 12. FINAL SUMMARY
    # ------------------------------------------------------------

    print("\n" + "=" * 65)
    print("COLLECTION COMPLETED SUCCESSFULLY")
    print("=" * 65)

    print("Dataset shape:", weather_df.shape)

    print("\nWeather CSV:")
    print(WEATHER_CSV_PATH)

    print("\nMetadata JSON:")
    print(WEATHER_METADATA_PATH)

    print("\nFirst five records:")
    display(weather_df.head())

    print("\nLast five records:")
    display(weather_df.tail())
````

````python
# ============================================================
# PHASE 04: VALIDATE SAVED OPEN-METEO DATASET
# ============================================================

import json
import pandas as pd
from pathlib import Path

# ------------------------------------------------------------
# 1. FILE PATHS
# ------------------------------------------------------------

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

WEATHER_CSV_PATH = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "weather"
    / "hyderabad_historical_weather_174578_783742.csv"
)

WEATHER_METADATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "metadata"
    / "weather_source_metadata_174578_783742.json"
)

# ------------------------------------------------------------
# 2. VERIFY FILE EXISTENCE
# ------------------------------------------------------------

print("=" * 65)
print("SAVED WEATHER DATA VALIDATION")
print("=" * 65)

for path in [WEATHER_CSV_PATH, WEATHER_METADATA_PATH]:
    if not path.exists():
        raise FileNotFoundError(f"Required file not found: {path}")
    print(f"Found: {path}")

# ------------------------------------------------------------
# 3. LOAD CSV AND METADATA
# ------------------------------------------------------------

weather_df = pd.read_csv(
    WEATHER_CSV_PATH,
    parse_dates=["timestamp"]
)

with open(WEATHER_METADATA_PATH, "r", encoding="utf-8") as f:
    metadata = json.load(f)

# ------------------------------------------------------------
# 4. VERIFY COORDINATE PROVENANCE
# ------------------------------------------------------------

requested = metadata["requested_location"]
returned = metadata["returned_location"]

assert requested["latitude"] == 17.4578
assert requested["longitude"] == 78.3742

print("\nRequested coordinates:")
print("Latitude :", requested["latitude"])
print("Longitude:", requested["longitude"])

print("\nReturned coordinates:")
print("Latitude :", returned["returned_latitude"])
print("Longitude:", returned["returned_longitude"])
print("Elevation:", returned["elevation_m"], "m")

# ------------------------------------------------------------
# 5. VERIFY EXPECTED VARIABLES
# ------------------------------------------------------------

expected_variables = [
    "temperature_2m",
    "relative_humidity_2m",
    "apparent_temperature",
    "precipitation",
    "rain",
    "wind_speed_10m",
    "wind_direction_10m",
    "wind_gusts_10m",
    "cloud_cover",
    "surface_pressure",
    "dew_point_2m",
]

missing_columns = [
    col for col in expected_variables
    if col not in weather_df.columns
]

if missing_columns:
    raise ValueError(f"Missing weather columns: {missing_columns}")

# ------------------------------------------------------------
# 6. VERIFY HOURLY TIMESTAMPS
# ------------------------------------------------------------

expected_timestamps = pd.date_range(
    start="2023-01-01 00:00:00",
    end="2026-08-01 00:00:00",
    freq="h",
    inclusive="left"
)

actual_timestamps = pd.DatetimeIndex(weather_df["timestamp"])

missing_timestamps = expected_timestamps.difference(
    actual_timestamps
)

unexpected_timestamps = actual_timestamps.difference(
    expected_timestamps
)

duplicate_count = int(
    weather_df["timestamp"].duplicated().sum()
)

# ------------------------------------------------------------
# 7. CHECK MISSING VALUES
# ------------------------------------------------------------

missing_values = (
    weather_df[expected_variables]
    .isna()
    .sum()
    .astype(int)
)

# ------------------------------------------------------------
# 8. REPORT RESULTS
# ------------------------------------------------------------

print("\nDataset shape:", weather_df.shape)
print("Expected hourly records:", len(expected_timestamps))
print("Actual records:", len(weather_df))
print("Unique timestamps:", weather_df["timestamp"].nunique())
print("Duplicate timestamps:", duplicate_count)
print("Missing timestamps:", len(missing_timestamps))
print("Unexpected timestamps:", len(unexpected_timestamps))

print("\nMissing values by variable:")
print(missing_values.to_string())

# ------------------------------------------------------------
# 9. FINAL VALIDATION STATUS
# ------------------------------------------------------------

checks = {
    "Expected record count": len(weather_df) == len(expected_timestamps),
    "Unique timestamps": weather_df["timestamp"].nunique() == len(expected_timestamps),
    "No duplicate timestamps": duplicate_count == 0,
    "No missing timestamps": len(missing_timestamps) == 0,
    "No unexpected timestamps": len(unexpected_timestamps) == 0,
    "No missing weather values": int(missing_values.sum()) == 0,
    "All required columns present": len(missing_columns) == 0,
}

print("\nVALIDATION CHECKLIST")
print("-" * 65)

for check, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL'} | {check}")

if all(checks.values()):
    print("\nRESULT: SAVED DATASET PASSED ALL CHECKS.")
else:
    print("\nRESULT: INVESTIGATION REQUIRED.")

# Display a small sample
display(weather_df.head())
````

````python
# ============================================================
# PHASE 04: WEATHER DATA PROFILING
# ============================================================

import pandas as pd
import numpy as np
from IPython.display import display

# ------------------------------------------------------------
# 1. LOAD THE VALIDATED WEATHER DATA
# ------------------------------------------------------------

weather_df = pd.read_csv(
    WEATHER_CSV_PATH,
    parse_dates=["timestamp"]
)

weather_variables = [
    "temperature_2m",
    "relative_humidity_2m",
    "apparent_temperature",
    "precipitation",
    "rain",
    "wind_speed_10m",
    "wind_direction_10m",
    "wind_gusts_10m",
    "cloud_cover",
    "surface_pressure",
    "dew_point_2m",
]

print("=" * 70)
print("OPEN-METEO WEATHER DATA PROFILING")
print("=" * 70)

print("Dataset shape:", weather_df.shape)
print(
    "Time period:",
    weather_df["timestamp"].min(),
    "to",
    weather_df["timestamp"].max()
)

# ------------------------------------------------------------
# 2. DESCRIPTIVE STATISTICS
# ------------------------------------------------------------

print("\n1. DESCRIPTIVE STATISTICS")

display(
    weather_df[weather_variables]
    .describe()
    .T
    .round(3)
)

# ------------------------------------------------------------
# 3. UNIQUE VALUES AND MISSING VALUES
# ------------------------------------------------------------

print("\n2. VARIABLE COMPLETENESS")

completeness = pd.DataFrame({
    "data_type": weather_df[weather_variables].dtypes.astype(str),
    "missing_values": weather_df[weather_variables].isna().sum(),
    "unique_values": weather_df[weather_variables].nunique(),
})

display(completeness)

# ------------------------------------------------------------
# 4. RAINFALL PROFILE
# ------------------------------------------------------------

print("\n3. RAINFALL PROFILE")

rainfall_profile = {
    "hours_with_precipitation": int(
        (weather_df["precipitation"] > 0).sum()
    ),
    "hours_without_precipitation": int(
        (weather_df["precipitation"] == 0).sum()
    ),
    "maximum_hourly_precipitation_mm": float(
        weather_df["precipitation"].max()
    ),
    "total_recorded_precipitation_mm": float(
        weather_df["precipitation"].sum()
    ),
    "negative_precipitation_values": int(
        (weather_df["precipitation"] < 0).sum()
    ),
}

for key, value in rainfall_profile.items():
    print(f"{key}: {value}")

# ------------------------------------------------------------
# 5. BASIC PHYSICAL-RANGE CHECKS
# These flag values for investigation; they do not delete data.
# ------------------------------------------------------------

print("\n4. RANGE CHECKS")

range_checks = {
    "relative_humidity_2m": (0, 100),
    "cloud_cover": (0, 100),
    "wind_direction_10m": (0, 360),
    "wind_speed_10m": (0, None),
    "wind_gusts_10m": (0, None),
    "precipitation": (0, None),
    "rain": (0, None),
}

for variable, (lower, upper) in range_checks.items():
    series = weather_df[variable]
    invalid = pd.Series(False, index=series.index)

    if lower is not None:
        invalid |= series < lower

    if upper is not None:
        invalid |= series > upper

    print(
        f"{variable}: {int(invalid.sum())} "
        "values outside configured bounds"
    )

# ------------------------------------------------------------
# 6. TEMPERATURE AND DEW-POINT RELATIONSHIP
# ------------------------------------------------------------

print("\n5. TEMPERATURE RELATIONSHIP")

temperature_below_dewpoint = (
    weather_df["dew_point_2m"] >
    weather_df["temperature_2m"]
)

print(
    "Hours where dew point exceeds air temperature:",
    int(temperature_below_dewpoint.sum())
)

# ------------------------------------------------------------
# 7. HOURLY RAINFALL SUMMARY BY MONTH
# ------------------------------------------------------------

print("\n6. MONTHLY RAINFALL SUMMARY")

monthly_rainfall = (
    weather_df.assign(
        year=weather_df["timestamp"].dt.year,
        month=weather_df["timestamp"].dt.month
    )
    .groupby(["year", "month"])
    .agg(
        hourly_records=("precipitation", "size"),
        total_precipitation_mm=("precipitation", "sum"),
        wet_hours=("precipitation", lambda x: (x > 0).sum()),
        maximum_hourly_precipitation_mm=("precipitation", "max"),
    )
    .reset_index()
)

display(monthly_rainfall.head(12))

print("\nProfiling completed.")
print(
    "No values have been modified or deleted. "
    "Investigate flagged observations before making corrections."
)
````

````python
# ============================================================
# PHASE 04: PRECIPITATION VS RAIN COMPARISON
# ============================================================

import pandas as pd
import numpy as np

# Load the saved weather data
weather_df = pd.read_csv(
    WEATHER_CSV_PATH,
    parse_dates=["timestamp"]
)

# Compare the two variables
precipitation = weather_df["precipitation"]
rain = weather_df["rain"]

difference = precipitation - rain

comparison = pd.DataFrame({
    "Metric": [
        "Total records",
        "Precipitation mean (mm)",
        "Rain mean (mm)",
        "Precipitation maximum (mm)",
        "Rain maximum (mm)",
        "Hours precipitation > 0",
        "Hours rain > 0",
        "Hours precipitation != rain",
        "Maximum absolute difference (mm)",
        "Correlation"
    ],
    "Value": [
        len(weather_df),
        precipitation.mean(),
        rain.mean(),
        precipitation.max(),
        rain.max(),
        int((precipitation > 0).sum()),
        int((rain > 0).sum()),
        int((~np.isclose(precipitation, rain)).sum()),
        difference.abs().max(),
        precipitation.corr(rain)
    ]
})

display(comparison.round(4))

# Inspect records where the values differ
different_rows = weather_df.loc[
    ~np.isclose(precipitation, rain),
    ["timestamp", "precipitation", "rain"]
]

print("Records where precipitation and rain differ:",
      len(different_rows))

display(different_rows.head(10))
````

# Phase 04.3 — Historical Weather Data Collection, Validation and Cleaning

## 1. Objective

The objective of this subsection is to collect hourly historical weather data for Hyderabad, India, validate its temporal coverage and data quality, and prepare a cleaned dataset for subsequent contextual data integration.

The collected weather data will provide environmental context for the Food Demand Forecasting Engine. Weather variables such as temperature, humidity, precipitation, and wind conditions may be used as explanatory features in later forecasting experiments.

## 2. Data Source

**Selected source:** Open-Meteo Historical Weather API

Open-Meteo was selected as the sole weather data source for this project to maintain a consistent collection process. Separate datasets from PERSIANN-CCS and NASA GPM IMERG were not included in this workflow.

**Official source:** https://open-meteo.com/

| Parameter           | Value          |
| ------------------- | -------------- |

| Requested longitude | 78.3742        |

| Temporal frequency  | Hourly         |

The raw data was kept separate from the processed data to preserve the original collected values and support reproducibility.

**Recorded raw-data artifact:**

`data/raw/weather/hyderabad_historical_weather_174578_783742.csv`

`data/processed/weather/hyderabad_weather_cleaned_174578_783742.csv`

* One timestamp column.
* Ten weather variables.
* A total of eleven columns.

| Validation check                   | Recorded result                          |
| ---------------------------------- | ---------------------------------------- |

These results reflect the recorded collection workflow. The underlying CSV files and notebook outputs should be inspected or revalidated before these checks are treated as independently reproduced results.

## 7. Spatial Coverage and Source Limitations

Although the request used coordinates 17.4578° latitude and 78.3742° longitude, the returned weather grid location was approximately:

* **Latitude:** 17.469244°
* **Longitude:** 78.35907°
* **Elevation:** 587 metres

The returned grid location differs from the requested coordinates. Therefore, the data should not be described as direct weather-station observations measured precisely at the requested point.

The request used Open-Meteo's default *Best Match* model selection. A single fixed spatial resolution should not be claimed without confirming the selected underlying model and its resolution.

The historical weather data represents gridded/model-derived estimates and may not capture every highly localised weather variation affecting individual delivery zones.

## 8. Coverage Consideration for Future Integration

The weather dataset ends on 31 July 2026, while the project's validated complete-date hourly demand panel extends through 30 August 2026.

This difference must be addressed explicitly during the later compatibility and integration steps.

Weather values must not be fabricated to fill the uncovered period. Any decision to restrict the integration period or obtain additional weather data must be documented and validated before integration.

## 9. Research Findings

The recorded Phase 04.3 workflow established a historical weather dataset for the requested period using Open-Meteo.

The workflow recorded 31,392 hourly records, no missing timestamps, and no missing values. The cleaning process removed the overlapping `rain` variable while retaining `precipitation`.

The raw dataset, cleaned dataset, and source metadata were assigned separate artifact paths to support traceability and reproducibility.

The principal limitations identified were the difference between the requested and returned grid coordinates, the gridded nature of the weather estimates, and the temporal coverage difference between weather and demand data.

````python
import re
import requests
from bs4 import BeautifulSoup

SOURCES = {
    2023: "https://www.telangana.gov.in/Downloads/calendar-2023/",
    2024: "https://www.telangana.gov.in/downloads/calendar/",
    2025: "https://www.telangana.gov.in/downloads/calendar-2025/",
    2026: "https://www.telangana.gov.in/downloads/calendar-2026/",
}

headers = {"User-Agent": "Mozilla/5.0"}

# Match entries such as: Jan 01 | New Year Day
holiday_pattern = re.compile(
    r"^\s*(Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)"
    r"\s+(\d{1,2})\s*\|\s*(.+?)\s*$",
    flags=re.MULTILINE | re.IGNORECASE
)

for year, url in SOURCES.items():
    print(f"\n{'=' * 60}")
    print(f"YEAR: {year}")

    try:
        response = requests.get(
            url,
            headers=headers,
            timeout=30
        )
        response.raise_for_status()

        soup = BeautifulSoup(response.content, "html.parser")

        # Extract text from the whole page, not just the heading widget.
        page_text = soup.get_text("\n", strip=True)

        matches = holiday_pattern.findall(page_text)

        print("HTTP status:", response.status_code)
        print("Matching entries:", len(matches))

        for month, day, name in matches[:12]:
            print(f"{month} {int(day):02d} | {name}")

        if not matches:
            print("No matching entries. More inspection is required.")

    except Exception as exc:
        print("ERROR:", repr(exc))
````

````python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup

SOURCES = {
    2023: "https://www.telangana.gov.in/Downloads/calendar-2023/",
    2024: "https://www.telangana.gov.in/downloads/calendar/",
    2025: "https://www.telangana.gov.in/downloads/calendar-2025/",
    2026: "https://www.telangana.gov.in/downloads/calendar-2026/",
}

MONTHS = {
    "Jan": 1, "Feb": 2, "Mar": 3, "Apr": 4,
    "May": 5, "Jun": 6, "Jul": 7, "Aug": 8,
    "Sep": 9, "Oct": 10, "Nov": 11, "Dec": 12
}

DATE_PATTERN = re.compile(
    r"^(Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)"
    r"\s+(\d{1,2})$",
    re.IGNORECASE
)

headers = {"User-Agent": "Mozilla/5.0"}
all_records = []

for year, url in SOURCES.items():
    print(f"\n{'=' * 60}")
    print(f"YEAR: {year}")

    response = requests.get(url, headers=headers, timeout=30)
    response.raise_for_status()

    soup = BeautifulSoup(response.content, "html.parser")
    year_records = []

    for table_index, table in enumerate(soup.find_all("table"), start=1):
        for row in table.find_all("tr"):
            cells = row.find_all(["th", "td"], recursive=False)

            if len(cells) < 2:
                continue

            date_text = cells[0].get_text(" ", strip=True)
            holiday_name = cells[1].get_text(" ", strip=True)

            match = DATE_PATTERN.match(date_text)

            if not match or not holiday_name:
                continue

            month_text, day_text = match.groups()
            month_number = MONTHS[month_text.title()]
            day_number = int(day_text)

            try:
                holiday_date = pd.Timestamp(
                    year=year,
                    month=month_number,
                    day=day_number
                )
            except ValueError:
                print("Invalid date encountered:", date_text, year)
                continue

            record = {
                "date": holiday_date.date().isoformat(),
                "year": year,
                "holiday_name": holiday_name,
                "source_url": url,
                "source_table_index": table_index
            }

            year_records.append(record)

    print("Extracted rows:", len(year_records))

    for record in year_records[:10]:
        print(record["date"], "|", record["holiday_name"])

    if not year_records:
        print("WARNING: No records extracted. Further inspection required.")

    all_records.extend(year_records)

holidays_diagnostic_df = pd.DataFrame(all_records)

print("\n" + "=" * 60)
print("COMBINED DIAGNOSTIC SUMMARY")
print("Total extracted rows:", len(holidays_diagnostic_df))

if not holidays_diagnostic_df.empty:
    print("\nRows by year:")
    print(
        holidays_diagnostic_df.groupby("year")
        .size()
        .reindex(SOURCES.keys(), fill_value=0)
    )

    print("\nDuplicate date-and-name combinations:")
    print(
        holidays_diagnostic_df.duplicated(
            subset=["date", "holiday_name"]
        ).sum()
    )

    print("\nExtracted date range:")
    print(
        holidays_diagnostic_df["date"].min(),
        "to",
        holidays_diagnostic_df["date"].max()
    )
````

````python
import requests
import pandas as pd
from bs4 import BeautifulSoup
from collections import Counter

SOURCES = {
    2023: "https://www.telangana.gov.in/Downloads/calendar-2023/",
    2024: "https://www.telangana.gov.in/downloads/calendar/",
    2025: "https://www.telangana.gov.in/downloads/calendar-2025/",
    2026: "https://www.telangana.gov.in/downloads/calendar-2026/",
}

headers = {"User-Agent": "Mozilla/5.0"}
summary = []

for year, url in SOURCES.items():
    response = requests.get(url, headers=headers, timeout=30)
    response.raise_for_status()

    soup = BeautifulSoup(response.content, "html.parser")

    print(f"\n{'=' * 65}")
    print("YEAR:", year)

    class_counts = Counter()
    samples = []

    for table_index, table in enumerate(soup.find_all("table"), start=1):
        for row in table.find_all("tr"):
            cells = row.find_all(["th", "td"], recursive=False)

            # Calendar rows contain a date cell followed by its holiday name.
            for i, cell in enumerate(cells):
                cell_text = cell.get_text(" ", strip=True)

                if not cell_text or not cell.name == "th":
                    continue

                classes = cell.get("class", [])
                if not classes:
                    continue

                class_label = ",".join(sorted(classes))
                class_counts[class_label] += 1

                if i + 1 < len(cells):
                    holiday_name = cells[i + 1].get_text(" ", strip=True)

                    if len(samples) < 12:
                        samples.append({
                            "table": table_index,
                            "date_cell": cell_text,
                            "class": class_label,
                            "holiday": holiday_name
                        })

    print("Date-cell class counts:", dict(class_counts))
    print("\nSample date/holiday pairs:")

    for sample in samples:
        print(sample)

    for class_label, count in class_counts.items():
        summary.append({
            "year": year,
            "date_cell_class": class_label,
            "count": count
        })

summary_df = pd.DataFrame(summary)

print("\n--- CLASSIFICATION SUMMARY ---")
display(summary_df)
````

````python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup

SOURCES = {
    2023: "https://www.telangana.gov.in/Downloads/calendar-2023/",
    2024: "https://www.telangana.gov.in/downloads/calendar/",
    2025: "https://www.telangana.gov.in/downloads/calendar-2025/",
    2026: "https://www.telangana.gov.in/downloads/calendar-2026/",
}

MONTHS = {
    "Jan": 1, "Feb": 2, "Mar": 3, "Apr": 4,
    "May": 5, "Jun": 6, "Jul": 7, "Aug": 8,
    "Sep": 9, "Oct": 10, "Nov": 11, "Dec": 12
}

DATE_PATTERN = re.compile(
    r"^(Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)"
    r"\s+(\d{1,2})$",
    re.IGNORECASE
)

headers = {"User-Agent": "Mozilla/5.0"}
records = []

for year, url in SOURCES.items():
    response = requests.get(url, headers=headers, timeout=30)
    response.raise_for_status()

    soup = BeautifulSoup(response.content, "html.parser")

    for table_index, table in enumerate(soup.find_all("table"), start=1):
        for row in table.find_all("tr"):
            cells = row.find_all(["th", "td"], recursive=False)

            for index, cell in enumerate(cells[:-1]):
                match = DATE_PATTERN.match(
                    cell.get_text(" ", strip=True)
                )

                if not match:
                    continue

                month_text, day_text = match.groups()
                month = MONTHS[month_text.title()]
                day = int(day_text)

                holiday_name = cells[index + 1].get_text(
                    " ", strip=True
                )

                if not holiday_name:
                    continue

                classes = cell.get("class", [])
                class_label = ",".join(classes) if classes else "unclassified"

                try:
                    date = pd.Timestamp(
                        year=year,
                        month=month,
                        day=day
                    )
                except ValueError:
                    print("Invalid date:", year, cell.get_text(" ", strip=True))
                    continue

                records.append({
                    "date": date,
                    "year": year,
                    "holiday_name": holiday_name,
                    "calendar_class": class_label,
                    "source_table": table_index,
                    "source_url": url
                })

holidays_review_df = pd.DataFrame(records)

if holidays_review_df.empty:
    raise ValueError("No holiday records extracted.")

print("TOTAL RECORDS:", len(holidays_review_df))

print("\nRECORDS BY YEAR:")
display(
    holidays_review_df.groupby("year")
    .size()
    .reindex(SOURCES.keys(), fill_value=0)
    .rename("record_count")
    .to_frame()
)

print("\nRECORDS BY YEAR AND HTML CLASS:")
display(
    pd.crosstab(
        holidays_review_df["year"],
        holidays_review_df["calendar_class"]
    )
)

print("\nDUPLICATE DATE + NAME RECORDS:")
display(
    holidays_review_df[
        holidays_review_df.duplicated(
            subset=["date", "holiday_name"],
            keep=False
        )
    ].sort_values(["date", "holiday_name"])
)

print("\nALL RECORDS — 2023:")
display(
    holidays_review_df[
        holidays_review_df["year"] == 2023
    ][["date", "holiday_name", "calendar_class"]]
    .sort_values(["date", "holiday_name"])
)

print("\nALL RECORDS — 2025:")
display(
    holidays_review_df[
        holidays_review_df["year"] == 2025
    ][["date", "holiday_name", "calendar_class"]]
    .sort_values(["date", "holiday_name"])
)

print("\nUNCLASSIFIED RECORDS:")
display(
    holidays_review_df[
        holidays_review_df["calendar_class"] == "unclassified"
    ]
)
````

````python
# Review all extracted records, grouped by year and HTML class.

for year in [2023, 2024, 2025, 2026]:
    print(f"\n{'=' * 70}")
    print(f"TELANGANA HOLIDAY CALENDAR — {year}")

    year_df = (
        holidays_review_df[
            holidays_review_df["year"] == year
        ][["date", "holiday_name", "calendar_class"]]
        .sort_values(["date", "holiday_name"])
        .reset_index(drop=True)
    )

    print("Total records:", len(year_df))
    print("\nClass counts:")
    print(year_df["calendar_class"].value_counts().to_string())

    display(year_df)
````

````python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

BASE_URL = "https://www.telangana.gov.in/"
HEADERS = {"User-Agent": "Mozilla/5.0"}

response = requests.get(
    BASE_URL,
    headers=HEADERS,
    timeout=30
)
response.raise_for_status()

soup = BeautifulSoup(response.content, "html.parser")

keywords = [
    "holiday",
    "general administration",
    "government order",
    "g.o.",
    "notification",
]

matches = []

for link in soup.find_all("a", href=True):
    text = link.get_text(" ", strip=True)
    href = urljoin(BASE_URL, link["href"])

    searchable = f"{text} {href}".lower()

    if any(keyword in searchable for keyword in keywords):
        matches.append({
            "link_text": text,
            "url": href,
        })

print("Matching links:", len(matches))

for item in matches[:60]:
    print(f"{item['link_text']} | {item['url']}")
````

````python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

URL = "https://www.telangana.gov.in/government-orders/"

response = requests.get(
    URL,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=30
)
response.raise_for_status()

soup = BeautifulSoup(response.content, "html.parser")

print("HTTP status:", response.status_code)
print("Page title:", soup.title.get_text(strip=True) if soup.title else "Not found")

# Search page links for holiday-related documents.
keywords = [
    "holiday",
    "holidays",
    "general administration",
    "optional",
    "2023",
    "2024",
    "2025",
    "2026",
]

matches = []

for link in soup.find_all("a", href=True):
    text = link.get_text(" ", strip=True)
    href = urljoin(URL, link["href"])

    searchable = f"{text} {href}".lower()

    if any(keyword in searchable for keyword in keywords):
        matches.append({
            "link_text": text,
            "url": href,
        })

print("\nMatching links:", len(matches))

for item in matches[:100]:
    print(f"{item['link_text']} | {item['url']}")

# Inspect page text for references to holiday notifications.
page_text = soup.get_text(" ", strip=True)

print("\n--- HOLIDAY-RELATED TEXT ---")

for keyword in ["holiday", "holidays", "General Administration"]:
    position = page_text.lower().find(keyword.lower())

    if position >= 0:
        start = max(0, position - 200)
        end = min(len(page_text), position + 700)

        print(f"\nKeyword: {keyword}")
        print(page_text[start:end])
````

````python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

URL = (
    "https://www.telangana.gov.in/"
    "departments/general-administration-political/"
)

response = requests.get(
    URL,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=30
)
response.raise_for_status()

soup = BeautifulSoup(response.content, "html.parser")

print("HTTP status:", response.status_code)
print(
    "Page title:",
    soup.title.get_text(strip=True) if soup.title else "Not found"
)

# Find links potentially relevant to holiday notifications.
keywords = [
    "holiday",
    "holidays",
    "general administration",
    "government order",
    "notification",
    "2023",
    "2024",
    "2025",
    "2026",
]

matches = []

for link in soup.find_all("a", href=True):
    text = link.get_text(" ", strip=True)
    href = urljoin(URL, link["href"])

    if any(
        keyword in f"{text} {href}".lower()
        for keyword in keywords
    ):
        matches.append({
            "text": text,
            "url": href,
        })

print("\nPotentially relevant links:", len(matches))

for item in matches[:100]:
    print(f"{item['text']} | {item['url']}")

# Check the page for holiday-related text.
page_text = soup.get_text(" ", strip=True)

print("\n--- PAGE TEXT SEARCH ---")

for keyword in ["holiday", "holidays", "G.O.", "notification"]:
    position = page_text.lower().find(keyword.lower())

    if position != -1:
        print(f"\nFound: {keyword}")
        print(page_text[max(0, position - 150):position + 500])
````

````python
import requests
import pandas as pd
from bs4 import BeautifulSoup

url = "https://www.telangana.gov.in/downloads/calendar-2025/"

response = requests.get(
    url,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=30
)
response.raise_for_status()

soup = BeautifulSoup(response.content, "html.parser")

source_records = []

for table_index, table in enumerate(soup.find_all("table"), start=1):
    for row in table.find_all("tr"):
        cells = row.find_all(["th", "td"], recursive=False)

        for index, cell in enumerate(cells[:-1]):
            date_text = cell.get_text(" ", strip=True)

            if not cell.name == "th":
                continue

            import re

            if not re.match(
                r"^(Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)\s+\d{1,2}$",
                date_text,
                re.IGNORECASE
            ):
                continue

            holiday_name = cells[index + 1].get_text(
                " ", strip=True
            )

            source_records.append({
                "date_text": date_text,
                "holiday_name": holiday_name,
                "calendar_class": ",".join(cell.get("class", [])),
                "table_index": table_index
            })

source_2025_df = pd.DataFrame(source_records)

print("Records extracted directly from the 2025 source:", len(source_2025_df))

print("\nRecords per month:")
source_2025_df["month"] = source_2025_df["date_text"].str[:3]

display(
    source_2025_df.groupby("month")
    .size()
    .rename("record_count")
    .to_frame()
)

print("\nComplete source extraction:")
display(source_2025_df)
````

````python
import pandas as pd

# Original extracted records
original_2025 = holidays_review_df[
    holidays_review_df["year"] == 2025
].copy()

original_2025["date"] = pd.to_datetime(
    original_2025["date"]
).dt.strftime("%Y-%m-%d")

original_2025["holiday_name_normalized"] = (
    original_2025["holiday_name"]
    .astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
    .str.casefold()
)

# Independently extracted source records
comparison_2025 = source_2025_df.copy()

comparison_2025["date"] = pd.to_datetime(
    comparison_2025["date_text"] + " 2025",
    format="%b %d %Y",
    errors="coerce"
).dt.strftime("%Y-%m-%d")

comparison_2025["holiday_name_normalized"] = (
    comparison_2025["holiday_name"]
    .astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
    .str.casefold()
)

# Compare date + normalized holiday name
original_keys = set(
    zip(
        original_2025["date"],
        original_2025["holiday_name_normalized"]
    )
)

source_keys = set(
    zip(
        comparison_2025["date"],
        comparison_2025["holiday_name_normalized"]
    )
)

only_original = original_keys - source_keys
only_source = source_keys - original_keys

print("Original extraction records:", len(original_2025))
print("Independent source records:", len(comparison_2025))
print("Records only in original extraction:", len(only_original))
print("Records only in independent extraction:", len(only_source))

print("\n--- Only in original extraction ---")
for date, name in sorted(only_original):
    print(date, "|", name)

print("\n--- Only in independent extraction ---")
for date, name in sorted(only_source):
    print(date, "|", name)
````

````python
import pandas as pd
from pathlib import Path

# --------------------------------------------------
# 1. Prepare the final holiday dataset
# --------------------------------------------------

final_holidays_df = holidays_review_df.copy()

final_holidays_df["date"] = pd.to_datetime(
    final_holidays_df["date"],
    errors="coerce"
).dt.strftime("%Y-%m-%d")

final_holidays_df["year"] = final_holidays_df["year"].astype(int)

final_holidays_df["holiday_name"] = (
    final_holidays_df["holiday_name"]
    .astype("string")
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

# Retain original source classification and provenance.
final_holidays_df = final_holidays_df[
    [
        "date",
        "year",
        "holiday_name",
        "calendar_class",
        "source_table",
        "source_url",
    ]
].sort_values(
    ["date", "holiday_name"]
).reset_index(drop=True)

# --------------------------------------------------
# 2. Final validation before saving
# --------------------------------------------------

assert final_holidays_df["date"].notna().all()
assert final_holidays_df["holiday_name"].notna().all()
assert final_holidays_df["source_url"].notna().all()

assert final_holidays_df["year"].isin(
    [2023, 2024, 2025, 2026]
).all()

assert (
    pd.to_datetime(final_holidays_df["date"]).dt.year
    == final_holidays_df["year"]
).all()

assert not final_holidays_df.duplicated(
    subset=["date", "holiday_name"]
).any()

print("Final dataset rows:", len(final_holidays_df))

print("\nRecords per year:")
display(
    final_holidays_df.groupby("year")
    .size()
    .rename("record_count")
    .to_frame()
)

print("\nSource HTML classes:")
display(
    final_holidays_df["calendar_class"]
    .value_counts(dropna=False)
    .rename_axis("calendar_class")
    .to_frame("record_count")
)

print("\nPreview:")
display(final_holidays_df.head(10))
````

````python
import json
from pathlib import Path
from datetime import datetime, timezone

# --------------------------------------------------
# 1. Define output locations
# --------------------------------------------------

current_path = Path.cwd().resolve()
project_root = current_path.parent if current_path.name.lower() == "notebooks" else current_path

raw_dir = project_root / "data" / "raw" / "holidays"
metadata_dir = project_root / "data" / "metadata"

raw_dir.mkdir(parents=True, exist_ok=True)
metadata_dir.mkdir(parents=True, exist_ok=True)

csv_path = raw_dir / "telangana_calendar_holidays_2023_2026.csv"
metadata_path = metadata_dir / "telangana_holiday_source_metadata.json"

# --------------------------------------------------
# 2. Save the extracted dataset
# --------------------------------------------------

final_holidays_df.to_csv(
    csv_path,
    index=False,
    encoding="utf-8"
)

# --------------------------------------------------
# 3. Create source metadata
# --------------------------------------------------

source_urls = sorted(
    final_holidays_df["source_url"].dropna().unique().tolist()
)

yearly_counts = {
    str(int(year)): int(count)
    for year, count in (
        final_holidays_df.groupby("year").size().items()
    )
}

metadata = {
    "dataset_name": "Telangana Calendar Holidays 2023-2026",
    "collection_timestamp_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "geographic_scope": "Telangana, India",
    "coverage_start": "2023-01-01",
    "coverage_end": "2026-12-31",
    "record_count": int(len(final_holidays_df)),
    "records_by_year": yearly_counts,
    "source_urls": source_urls,
    "collection_method": (
        "HTML extraction from official Telangana State Portal "
        "calendar pages."
    ),
    "columns": list(final_holidays_df.columns),
    "classification_handling": (
        "Original HTML classes are retained as published. "
        "The values 'fix' and 'opt' have not been independently "
        "verified as legal holiday classifications."
    ),
    "validation": {
        "missing_dates": int(final_holidays_df["date"].isna().sum()),
        "missing_holiday_names": int(
            final_holidays_df["holiday_name"].isna().sum()
        ),
        "duplicate_date_name_records": int(
            final_holidays_df.duplicated(
                subset=["date", "holiday_name"]
            ).sum()
        ),
        "independent_2025_extraction_agreement": True,
    },
    "limitations": [
        (
            "Agreement between two extraction methods does not "
            "independently establish source completeness."
        ),
        (
            "Independent verification against government holiday "
            "orders remains pending."
        ),
        (
            "The 2025 calendar has 49 extracted records; the difference "
            "from other years has not been independently explained."
        ),
        (
            "The original HTML classification labels have not been "
            "independently verified against legal notifications."
        ),
    ],
}

with open(metadata_path, "w", encoding="utf-8") as file:
    json.dump(metadata, file, indent=2, ensure_ascii=False)

print("CSV saved to:", csv_path.resolve())
print("Metadata saved to:", metadata_path.resolve())
print("CSV size (bytes):", csv_path.stat().st_size)
print("Metadata size (bytes):", metadata_path.stat().st_size)
````

````python
# Reload the CSV from disk to verify that it can be read correctly.

reloaded_holidays_df = pd.read_csv(csv_path)

print("Saved CSV rows:", len(reloaded_holidays_df))
print("Saved CSV columns:", reloaded_holidays_df.columns.tolist())

print("\nRecords per year:")
display(
    reloaded_holidays_df.groupby("year")
    .size()
    .rename("record_count")
    .to_frame()
)

with open(metadata_path, "r", encoding="utf-8") as file:
    reloaded_metadata = json.load(file)

print("\nMetadata dataset name:", reloaded_metadata["dataset_name"])
print("Metadata record count:", reloaded_metadata["record_count"])

assert len(reloaded_holidays_df) == len(final_holidays_df)
assert reloaded_metadata["record_count"] == len(final_holidays_df)

print("\nPASS: Both files were saved and reloaded successfully.")
````

````python
# Reconcile holiday files into the single project-level data directory.
from pathlib import Path
import shutil

NOTEBOOK_DIR = Path.cwd().resolve()
PROJECT_ROOT = NOTEBOOK_DIR.parent if NOTEBOOK_DIR.name.lower() == "notebooks" else NOTEBOOK_DIR

source_csv = NOTEBOOK_DIR / "data" / "raw" / "holidays" / "telangana_calendar_holidays_2023_2026.csv"
source_metadata = NOTEBOOK_DIR / "data" / "metadata" / "telangana_holiday_source_metadata.json"

destination_raw_dir = PROJECT_ROOT / "data" / "raw" / "holidays"
destination_metadata_dir = PROJECT_ROOT / "data" / "metadata"
destination_raw_dir.mkdir(parents=True, exist_ok=True)
destination_metadata_dir.mkdir(parents=True, exist_ok=True)

destination_csv = destination_raw_dir / source_csv.name
destination_metadata = destination_metadata_dir / source_metadata.name

# Existing project-level files are canonical; never create timestamped copies.
if destination_csv.exists() and destination_metadata.exists():
    print("Existing project-level holiday files reused; no files moved or duplicated.")
else:
    if source_csv.exists() and not destination_csv.exists():
        shutil.move(str(source_csv), str(destination_csv))
    if source_metadata.exists() and not destination_metadata.exists():
        shutil.move(str(source_metadata), str(destination_metadata))

    if not destination_csv.exists() or not destination_metadata.exists():
        raise FileNotFoundError(
            "Required holiday CSV/metadata missing from both source and project locations. "
            "No duplicate files were created."
        )

    print("Holiday files placed in the project-level data directory.")

print("CSV:", destination_csv)
print("Metadata:", destination_metadata)
````

# Phase 04.4 — Public Holiday Collection, Extraction, Validation, and Storage

## 1. Objective

The objective of this phase was to collect public holiday calendar data for Telangana, India, for the period 2023–2026. Public holidays are included as contextual variables because they may influence consumer food-ordering behaviour, particularly during festivals, regional celebrations, and other holiday periods.

The collected dataset will support the later investigation of relationships between public holidays and hourly food demand across synthetic hyperlocal zones in Hyderabad.

The holiday data was collected independently and stored in the project's raw-data directory. It has not yet been integrated with the food-demand dataset.

## 2. Data Source Selection

The Telangana State Portal was selected as the primary source for the annual holiday calendars.

The following official calendar pages were used:

| Year | Source                                                |
| ---- | ----------------------------------------------------- |
| 2023 | https://www.telangana.gov.in/Downloads/calendar-2023/ |
| 2024 | https://www.telangana.gov.in/downloads/calendar/      |
| 2025 | https://www.telangana.gov.in/downloads/calendar-2025/ |
| 2026 | https://www.telangana.gov.in/downloads/calendar-2026/ |

The calendar pages contain holiday entries organized by month. The source HTML represents holiday dates and names in tables, with date cells and holiday-name cells stored separately.

The HTML class associated with each date cell was retained as a source classification for subsequent verification.

## 3. Data Extraction Methodology

Python was used to retrieve and inspect the official calendar pages. The extraction process involved the following steps:

1. Retrieve each annual calendar page using HTTP requests.
2. Parse the returned HTML using BeautifulSoup.
3. Identify the HTML tables containing calendar dates and holiday names.
4. Extract date and holiday-name pairs from the corresponding table cells.
5. Identify the year associated with each source calendar.
6. Convert the extracted dates into standardized date values.
7. Retain the source HTML classification and table index.
8. Record the source URL for traceability.
9. Combine the yearly records into a consolidated dataset.

The initial text-based extraction approach did not identify the holiday entries because the page content was structured in HTML tables rather than in the expected plain-text format. The extraction method was subsequently adjusted to inspect the table structure directly.

This adjustment enabled the extraction of holiday records from all four annual calendar pages.

## 4. Dataset Structure

The consolidated dataset contains the following columns:

| Column           | Description                                                 |
| ---------------- | ----------------------------------------------------------- |
| `date`           | Standardized holiday date                                   |
| `year`           | Calendar year associated with the record                    |
| `holiday_name`   | Holiday name extracted from the source calendar             |
| `calendar_class` | Source HTML classification, retained as `fix` or `opt`      |
| `source_table`   | Index of the HTML table from which the record was extracted |
| `source_url`     | URL of the source calendar page                             |

The `calendar_class` field preserves the source representation. The labels `fix` and `opt` have not been treated as independently verified legal classifications. Their precise meaning and the official status of individual dates must be confirmed against relevant government notifications before final modeling.

## 5. Extraction Results

The extraction process produced the following record counts:

| Year      | Extracted Records |
| --------- | ----------------: |
| 2023      |                52 |
| 2024      |                52 |
| 2025      |                49 |
| 2026      |                52 |
| **Total** |           **205** |

The consolidated dataset contains 205 records across the four calendar years.

The record counts differ between years. These differences were retained as observed rather than artificially adjusted to produce equal annual totals.

The extracted dataset was subjected to structural and consistency checks.

| -------------------------------------------------------- | ------ |

The source classification counts were also inspected.

| 2023 |    28 |    24 |

| 2025 |    27 |    22 |

These counts describe the extracted HTML date-cell classes. They should not be interpreted as confirmed counts of legally designated general and optional holidays without further verification.

Some holiday names contain explanatory notes referring to other occasions or government holiday provisions. These notes were preserved during extraction to maintain traceability to the original calendar content.

The checks establish that the saved dataset is structurally consistent with the extracted records. They do not independently establish that every source entry is legally authoritative, correctly classified, or suitable as a separate modeling event.

## 8. Limitations and Further Verification

The following limitations must be considered before the holiday data is used for modeling:

1. **Classification verification:** The extracted `fix` and `opt` labels represent HTML classes. Their interpretation must be checked against official holiday notifications.
2. **Official status:** Annual calendar entries should not automatically be treated as legally confirmed general holidays without reviewing the relevant government documentation.
3. **Duplicate dates:** A single date may contain multiple holiday entries. The absence of exact duplicate date-and-name pairs does not mean every date is unique.
4. **Holiday-name normalization:** Original holiday names and explanatory notes have been retained. A separate normalization step may be required for modeling features.
5. **Calendar coverage:** The collection covers the annual calendars for 2023–2026. The actual dates represented in the files must be considered when constructing features for the demand observation period.
6. **Forecasting alignment:** Holiday features will need to be aligned to the demand timestamps and the appropriate local calendar dates in Asia/Kolkata.
7. **Feature engineering:** Separate indicators may later be constructed for holiday dates, holiday categories, and potentially adjacent dates, but these should be defined and justified before model training.

No demand-data integration or model training was performed during this phase.

# Phase 04.5 — Cricket Match Context Collection

## 1. Objective

The objective of this phase is to collect, consolidate, and validate historical cricket-match information that may influence food-ordering demand in Hyderabad, India.

Cricket matches can change consumer routines, viewing behaviour, social gatherings, restaurant activity, and food-delivery preferences. Evening matches and major tournaments may be associated with changes in order volume before, during, or after a match.

Collecting only Indian Premier League (IPL) matches would provide an incomplete representation of cricket-related events. Therefore, this phase will cover multiple categories of cricket matches, including IPL, international cricket hosted in India, major domestic T20 competitions, and international tournaments conducted at Indian venues.

The resulting dataset will provide contextual information for investigating whether cricket events contribute to variations in hourly food demand across synthetic hyperlocal zones in Hyderabad.

The collection will remain independent of the demand dataset until the contextual records have been collected, validated, and documented.

## 2. Scope of Cricket Events

The following event categories will be investigated.

### 2.1 Indian Premier League (IPL)

The dataset will include IPL matches from the selected study period, including league-stage matches and playoff matches.

Where available, the collection will capture match dates, scheduled start times, participating teams, tournament stage, venue, city, and match status.

IPL matches held in Hyderabad will receive particular attention because they may have a more direct relationship with local food demand.

### 2.2 India Men's International Cricket

This category will cover India's men's international matches hosted in India against visiting national teams.

The formats of interest are:

* Test matches
* One Day Internationals (ODIs)
* Twenty20 Internationals (T20Is)

The collection will include bilateral series and international tournament matches, subject to source availability and clear classification.

International matches played in Hyderabad will be identified separately from matches held elsewhere in India.

### 2.3 India Women's International Cricket

Women's international matches hosted in India will be investigated where reliable historical fixture information is available.

The dataset will distinguish women's international cricket from men's international cricket so that their potential associations with demand can be evaluated separately.

### 2.4 Women's Premier League (WPL)

WPL matches will be included as a separate domestic T20 competition category.

The collection will identify tournament dates, participating teams, match venues, scheduled start times, and tournament stages where these details are available.

WPL records will remain distinguishable from IPL records throughout the collection and validation process.

### 2.5 Major International Cricket Tournaments Hosted in India

Major international tournaments will be included when matches are hosted at venues in India.

Examples include ICC tournaments and other major international competitions for which reliable match-level information is available.

Only verified matches will be included. Tournament-level dates alone will not be treated as individual match records.

### 2.6 Other Major Domestic Cricket Events

Other significant domestic cricket competitions may be investigated if reliable match schedules and venue information are available.

Examples may include selected major domestic T20 competitions and other nationally significant cricket events.

This category will be prioritized after the principal event categories have been investigated. Inclusion will depend on data quality, availability, and relevance to the forecasting study.

## 3. Study Period

The initial collection period will cover **1 January 2023 through 31 July 2026**, subject to confirmation against the actual demand-data coverage and the availability of reliable historical cricket records.

The final contextual dataset must not imply that data exists for periods that have not been verified.

If a relevant match or tournament falls outside the final demand observation period, it will be excluded from the primary modeling dataset unless there is a specific methodological reason to retain it.

## 4. Geographical Scope

The geographical scope will include relevant cricket matches hosted in India, with additional attention to Hyderabad.

Matches will be classified into the following geographical groups:

| Geographical category                        | Purpose                                                                                                                          |
| -------------------------------------------- | -------------------------------------------------------------------------------------------------------------------------------- |
| Hyderabad                                    | Primary local cricket-event context                                                                                              |
| Other locations in Telangana, where relevant | Additional regional context                                                                                                      |
| Other cities in India                        | National cricket-event context                                                                                                   |
| Locations outside India                      | Excluded from the primary India-hosted match dataset; considered separately only if required for a national-team match indicator |

Venue and city information will be retained wherever available.

The collection will distinguish matches hosted in Hyderabad from matches hosted elsewhere in India. This distinction is important because the effect of a local stadium event may differ from the effect of a match taking place in another city.

The geographical classification will be based on verified venue information rather than assumptions based on participating teams or tournament names.

## 5. Data Source Selection Strategy

Official and authoritative sources will be prioritized.

Potential sources to investigate include:

* The Board of Control for Cricket in India (BCCI) for India's international fixtures and domestic cricket schedules.
* The official IPL website for IPL schedules, match results, and venues.
* The official WPL website for Women's Premier League fixtures and match information.
* The International Cricket Council (ICC) for international tournaments and relevant match records.
* Other official tournament organizers or cricket-board websites when required.

Sources will be evaluated for historical coverage, match-level detail, reliability, accessibility, and consistency.

Where multiple sources provide the same match information, records will be cross-checked when practical. Secondary sources may be used to investigate gaps, but important records should be verified against authoritative sources wherever possible.

The selected source and source URL will be recorded for traceability.

No source will be treated as complete until its coverage has been investigated.

## 6. Proposed Dataset Structure

A consolidated match-level dataset will be developed. Each record will represent one cricket match.

The following fields are proposed:

| Column               | Description                                                                   |
| -------------------- | ----------------------------------------------------------------------------- |
| `match_id`           | Unique identifier assigned to a match                                         |
| `match_date`         | Verified match date                                                           |

Fields will be populated only when supported by the available evidence. Missing scheduled or actual times will remain missing rather than being estimated without justification.

Additional fields may be introduced if they are necessary for source traceability or subsequent feature engineering.

````python
# Phase 04.5.1 — Download Cricsheet Annual JSON Archives

from pathlib import Path
import requests
import pandas as pd

BASE_URL = "https://cricsheet.org/downloads/"
YEARS = [2023, 2024, 2025, 2026]

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
OUTPUT_DIR = PROJECT_ROOT / "data" / "raw" / "cricket" / "cricsheet"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

session = requests.Session()
session.headers.update({"User-Agent": "FoodDemandForecastingResearch/1.0"})

results = []

for year in YEARS:
    filename = f"{year}_json.zip"
    url = f"{BASE_URL}{filename}"
    output_path = OUTPUT_DIR / filename

    if output_path.is_file() and output_path.stat().st_size > 0:
        results.append({
            "year": year, "source": "Cricsheet", "url": url,
            "file": str(output_path),
            "size_mb": round(output_path.stat().st_size / (1024 * 1024), 2),
            "status": "reused existing archive"
        })
        print(f"Reusing existing {year} archive.")
        continue

    print(f"Downloading {year} archive...")

    try:
        response = session.get(url, timeout=120)
        response.raise_for_status()

        if not response.content.startswith(b"PK"):
            raise ValueError("Downloaded file is not a valid ZIP archive.")

        output_path.write_bytes(response.content)

        results.append({
            "year": year,
            "source": "Cricsheet",
            "url": url,
            "file": str(output_path),
            "size_mb": round(len(response.content) / (1024 * 1024), 2),
            "status": "success"
        })

    except Exception as exc:
        results.append({
            "year": year,
            "source": "Cricsheet",
            "url": url,
            "file": str(output_path),
            "size_mb": None,
            "status": f"failed: {exc}"
        })

download_df = pd.DataFrame(results)

display(download_df)

print("\nSuccessful downloads:",
      (download_df["status"] == "success").sum(),
      "of", len(YEARS))
````

````python
# Phase 04.5.1 — Download and validate Cricsheet JSON archives

from pathlib import Path
import requests
import pandas as pd
import zipfile
import io

BASE_URL = "https://cricsheet.org/downloads/"
YEARS = [2023, 2024, 2025, 2026]

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
OUTPUT_DIR = PROJECT_ROOT / "data" / "raw" / "cricket" / "cricsheet"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

session = requests.Session()
session.headers.update({
    "User-Agent": "Mozilla/5.0 (compatible; FoodDemandResearch/1.0)"
})

results = []

for year in YEARS:
    filename = f"{year}_json.zip"
    url = BASE_URL + filename
    output_path = OUTPUT_DIR / filename

    if output_path.is_file() and output_path.stat().st_size > 0:
        try:
            with zipfile.ZipFile(output_path) as archive:
                json_files = [n for n in archive.namelist() if n.lower().endswith(".json")]
                bad_file = archive.testzip()
                if not json_files or bad_file:
                    raise ValueError("Existing archive failed validation")
            results.append({
                "year": year, "status": "reused existing archive",
                "json_files_in_archive": len(json_files),
                "size_mb": round(output_path.stat().st_size / (1024 * 1024), 2),
                "content_type": "existing local file", "file": str(output_path)
            })
            print(f"Reusing existing validated {year} archive.")
            continue
        except Exception:
            print(f"Existing {year} archive invalid; redownloading it.")

    print(f"Checking {year} archive...")

    try:
        response = session.get(
            url,
            timeout=120,
            allow_redirects=True
        )

        content_type = response.headers.get("Content-Type", "")

        if response.status_code != 200:
            raise ValueError(f"HTTP {response.status_code}")

        # Validate the response as a ZIP archive.
        with zipfile.ZipFile(io.BytesIO(response.content)) as archive:
            json_files = [
                name for name in archive.namelist()
                if name.lower().endswith(".json")
            ]

            if not json_files:
                raise ValueError("ZIP contains no JSON files")

            bad_file = archive.testzip()
            if bad_file:
                raise ValueError(f"Corrupt archive member: {bad_file}")

        output_path.write_bytes(response.content)

        results.append({
            "year": year,
            "status": "success",
            "json_files_in_archive": len(json_files),
            "size_mb": round(len(response.content) / (1024 * 1024), 2),
            "content_type": content_type,
            "file": str(output_path)
        })

    except Exception as exc:
        results.append({
            "year": year,
            "status": f"failed: {exc}",
            "json_files_in_archive": None,
            "size_mb": None,
            "content_type": (
                response.headers.get("Content-Type", "")
                if "response" in locals() else ""
            ),
            "file": None
        })

download_df = pd.DataFrame(results)
display(download_df)

print(
    f"\nSuccessful downloads: "
    f"{(download_df['status'] == 'success').sum()} of {len(YEARS)}"
)
````

````python
# Phase 04.5.1 — Validate Cricket ZIP Archives

from pathlib import Path
from zipfile import ZipFile, BadZipFile
import pandas as pd

# Project root: notebook is inside the notebooks folder
PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

CRICKET_DIR = PROJECT_ROOT / "data" / "raw" / "cricket" / "cricsheet"

ZIP_FILES = [
    CRICKET_DIR / "all_csv2.zip",
    CRICKET_DIR / "ipl_csv2.zip",
]

results = []

for zip_path in ZIP_FILES:
    result = {
        "file": zip_path.name,
        "exists": zip_path.is_file(),
        "size_mb": None,
        "integrity": "Not checked",
        "csv_files": 0,
        "other_files": 0,
        "sample_files": None,
    }

    if zip_path.is_file():
        result["size_mb"] = round(
            zip_path.stat().st_size / (1024 * 1024), 2
        )

        try:
            with ZipFile(zip_path, "r") as archive:
                bad_file = archive.testzip()
                names = [
                    name for name in archive.namelist()
                    if not name.endswith("/")
                ]

                if bad_file:
                    result["integrity"] = f"Corrupt member: {bad_file}"
                else:
                    result["integrity"] = "Valid"

                result["csv_files"] = sum(
                    name.lower().endswith(".csv") for name in names
                )
                result["other_files"] = sum(
                    not name.lower().endswith(".csv") for name in names
                )
                result["sample_files"] = ", ".join(names[:5])

        except BadZipFile:
            result["integrity"] = "Invalid ZIP"

    results.append(result)

display(pd.DataFrame(results))

print("\nExpected folder:", CRICKET_DIR)
````

````python
# Phase 04.5.3 — Parse Cricsheet match metadata correctly

from pathlib import Path
from zipfile import ZipFile
import csv
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

CRICKET_DIR = PROJECT_ROOT / "data" / "raw" / "cricket" / "cricsheet"
OUTPUT_DIR = PROJECT_ROOT / "data" / "processed" / "cricket"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

START_DATE = pd.Timestamp("2023-01-01")
END_DATE = pd.Timestamp("2026-07-31")

def parse_info_csv(archive, member):
    metadata = {}

    with archive.open(member) as raw_file:
        reader = csv.reader(
            line.decode("utf-8-sig") for line in raw_file
        )

        for row in reader:
            # Cricsheet format: info, field_name, field_value
            if len(row) < 3 or row[0].strip() != "info":
                continue

            key = row[1].strip()
            value = row[2].strip()

            if key in metadata:
                if not isinstance(metadata[key], list):
                    metadata[key] = [metadata[key]]
                metadata[key].append(value)
            else:
                metadata[key] = value

    return metadata

def as_text(value):
    if isinstance(value, list):
        return " | ".join(value)
    return value

records = []

for archive_name in ["all_csv2.zip", "ipl_csv2.zip"]:
    archive_path = CRICKET_DIR / archive_name

    if not archive_path.exists():
        raise FileNotFoundError(f"Archive not found: {archive_path}")

    with ZipFile(archive_path) as archive:
        info_files = [
            name for name in archive.namelist()
            if name.lower().endswith("_info.csv")
        ]

        for member in info_files:
            metadata = parse_info_csv(archive, member)

            date_value = metadata.get("date")
            if not date_value:
                continue

            # Some matches have multiple dates; use the first recorded date.
            first_date = (
                date_value[0]
                if isinstance(date_value, list)
                else date_value
            )

            match_date = pd.to_datetime(
                first_date, format="%Y/%m/%d", errors="coerce"
            )

            if pd.isna(match_date):
                continue

            if not START_DATE <= match_date <= END_DATE:
                continue

            teams = metadata.get("team", [])
            if not isinstance(teams, list):
                teams = [teams]

            records.append({
                "match_id": str(metadata.get(
                    "match_id",
                    Path(member).name.replace("_info.csv", "")
                )),
                "match_date": match_date.date().isoformat(),
                "teams": " | ".join(map(str, teams)),
                "match_type": as_text(metadata.get("match_type")),
                "team_type": as_text(metadata.get("team_type")),
                "gender": as_text(metadata.get("gender")),
                "venue": as_text(metadata.get("venue")),
                "city": as_text(metadata.get("city")),
                "event": as_text(metadata.get("event")),
                "source_archive": archive_name,
                "source_file": member,
            })

matches = pd.DataFrame(records)

if matches.empty:
    raise ValueError(
        "No matches found. Check the metadata structure and date fields."
    )

all_matches = (
    matches[matches["source_archive"] == "all_csv2.zip"]
    .drop_duplicates(subset="match_id")
    .copy()
)

ipl_matches = (
    matches[matches["source_archive"] == "ipl_csv2.zip"]
    .drop_duplicates(subset="match_id")
    .copy()
)

all_path = OUTPUT_DIR / "cricket_matches_metadata_2023_2026.csv"
ipl_path = OUTPUT_DIR / "ipl_matches_metadata_2023_2026.csv"

all_matches.to_csv(all_path, index=False)
ipl_matches.to_csv(ipl_path, index=False)

print("Metadata extraction completed.")
print("All-archive matches:", len(all_matches))
print("IPL-archive matches:", len(ipl_matches))
print(
    "All-archive date range:",
    all_matches["match_date"].min(),
    "to",
    all_matches["match_date"].max()
)
print("Saved:", all_path)
print("Saved:", ipl_path)

display(all_matches.head(10))
````

````python
# Phase 04.5.4 — Classify cricket matches for the forecasting project

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data" / "processed" / "cricket"

INPUT_FILE = DATA_DIR / "cricket_matches_metadata_2023_2026.csv"
OUTPUT_FILE = DATA_DIR / "cricket_matches_classified_2023_2026.csv"

matches = pd.read_csv(INPUT_FILE)

# Normalize text for reliable classification.
teams_text = matches["teams"].fillna("").str.casefold()
event_text = matches["event"].fillna("").str.casefold()
venue_text = matches["venue"].fillna("").str.casefold()
city_text = matches["city"].fillna("").str.casefold()
source_text = matches["source_archive"].fillna("").str.casefold()

# India national-team involvement: exact team-name boundaries.
india_pattern = r"(?<![a-z])india(?![a-z])"
matches["india_involved"] = teams_text.str.contains(
    india_pattern, regex=True, na=False
)

# Event categories based on recorded event names.
matches["is_world_cup"] = event_text.str.contains(
    r"world cup|worldcup", regex=True, na=False
)

matches["is_ipl"] = (
    event_text.str.contains(r"\bindian premier league\b|\bipl\b",
                            regex=True, na=False)
    | source_text.eq("ipl_csv2.zip")
)

matches["is_wpl"] = event_text.str.contains(
    r"\bwomen.?s premier league\b|\bwpl\b",
    regex=True, na=False
)

# Identify Hyderabad using recorded venue/city text.
matches["is_hyderabad_match"] = (
    city_text.str.contains(r"\bhyderabad\b", regex=True, na=False)
    | venue_text.str.contains(r"\bhyderabad\b", regex=True, na=False)
)

# Host-country classification needs independent venue verification.
matches["is_india_hosted"] = pd.Series(
    pd.NA, index=matches.index, dtype="boolean"
)

# Preserve an explicit classification status for the next verification step.
matches["host_country_verification"] = "pending"

# Save classified master dataset.
matches.to_csv(OUTPUT_FILE, index=False)

# Concise summary.
summary = pd.DataFrame({
    "classification": [
        "Total matches",
        "India national-team matches",
        "World Cup event matches",
        "IPL matches",
        "WPL matches",
        "Hyderabad venue/city matches",
        "Host country pending verification",
    ],
    "count": [
        len(matches),
        int(matches["india_involved"].sum()),
        int(matches["is_world_cup"].sum()),
        int(matches["is_ipl"].sum()),
        int(matches["is_wpl"].sum()),
        int(matches["is_hyderabad_match"].sum()),
        int(matches["is_india_hosted"].isna().sum()),
    ]
})

print("Phase 04.5.4 completed.")
print("Saved:", OUTPUT_FILE)
display(summary)

print("\nSample classified records:")
display(
    matches.loc[
        matches["india_involved"] | matches["is_hyderabad_match"],
        [
            "match_date", "teams", "match_type", "event",
            "venue", "city", "india_involved",
            "is_world_cup", "is_ipl", "is_wpl",
            "is_hyderabad_match", "is_india_hosted"
        ]
    ].head(10)
)
````

````python
# Phase 04.5.5 — Verify Indian venues and classify host country

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data" / "processed" / "cricket"

INPUT_FILE = DATA_DIR / "cricket_matches_classified_2023_2026.csv"
OUTPUT_FILE = DATA_DIR / "cricket_matches_host_verified_2023_2026.csv"
REVIEW_FILE = DATA_DIR / "cricket_venue_verification_review_2023_2026.csv"

matches = pd.read_csv(INPUT_FILE)

# Recognized Indian cricket cities.
# City matching is a preliminary location signal, not independent proof.
INDIAN_CITIES = {
    "ahmedabad", "amritsar", "bengaluru", "bangalore", "bhubaneswar",
    "chandigarh", "chennai", "coimbatore", "cuttack", "dehradun",
    "dharamshala", "faridabad", "goa", "guwahati", "gwalior",
    "hyderabad", "indore", "jaipur", "jammu", "jamshedpur",
    "kanpur", "kochi", "kolkata", "lucknow", "mohali", "mumbai",
    "nagpur", "new delhi", "delhi", "noida", "patna", "pune",
    "rajkot", "ranchi", "surat", "thiruvananthapuram", "vadodara",
    "visakhapatnam", "vizag"
}

city = matches["city"].fillna("").str.strip().str.casefold()
venue = matches["venue"].fillna("").str.strip().str.casefold()

# Classify using recognized city names, but preserve the evidence and uncertainty.
matches["venue_city_recognized_india"] = city.isin(INDIAN_CITIES)

matches["host_country_status"] = "unverified"

# Recognized Indian cities provide preliminary evidence for review.
matches.loc[
    matches["venue_city_recognized_india"],
    "host_country_status"
] = "probable_india_city_match_review_required"

# Keep the formal host-country field unknown until verified against a reliable source.
matches["host_country"] = pd.NA
matches["is_india_hosted"] = pd.Series(
    pd.NA, index=matches.index, dtype="boolean"
)

# India national-team matches outside the recognized city list need review.
matches["india_match_location_review"] = (
    matches["india_involved"].fillna(False)
    & ~matches["venue_city_recognized_india"]
)

# Hyderabad-specific flag, based on the recorded city or venue.
matches["is_hyderabad_match"] = (
    city.str.contains("hyderabad", regex=False)
    | venue.str.contains("hyderabad", regex=False)
)

# Save updated master dataset and rows requiring venue verification.
matches.to_csv(OUTPUT_FILE, index=False)

review = matches.loc[
    matches["host_country_status"].eq("unverified")
    | matches["india_match_location_review"],
    [
        "match_id", "match_date", "teams", "match_type",
        "event", "venue", "city", "india_involved",
        "venue_city_recognized_india",
        "india_match_location_review",
        "host_country_status"
    ]
].copy()

review.to_csv(REVIEW_FILE, index=False)

summary = pd.DataFrame({
    "metric": [
        "Total matches",
        "Recognized Indian-city matches requiring verification",
        "Matches with unverified host country",
        "Indian-team matches needing location review",
        "Hyderabad matches"
    ],
    "count": [
        len(matches),
        int(matches["venue_city_recognized_india"].sum()),
        int(matches["is_india_hosted"].isna().sum()),
        int(matches["india_match_location_review"].sum()),
        int(matches["is_hyderabad_match"].sum())
    ]
})

print("Phase 04.5.5 completed.")
print("\nSummary:")
display(summary)

print("\nUpdated master dataset:", OUTPUT_FILE)
print("Venue review file:", REVIEW_FILE)

print("\nSample venue records:")
display(
    matches.loc[
        matches["venue_city_recognized_india"]
        | matches["india_match_location_review"],
        [
            "match_date", "teams", "event", "venue", "city",
            "india_involved", "venue_city_recognized_india",
            "host_country_status"
        ]
    ].head(15)
)
````

````python
# Phase 04.5.6 — Final cricket dataset and venue classification

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data" / "processed" / "cricket"

INPUT_FILE = DATA_DIR / "cricket_matches_host_verified_2023_2026.csv"
OUTPUT_FILE = DATA_DIR / "cricket_matches_final_2023_2026.csv"
REVIEW_FILE = DATA_DIR / "cricket_matches_venue_review_2023_2026.csv"

matches = pd.read_csv(INPUT_FILE)

# Normalize venue fields.
city = matches["city"].fillna("").str.strip().str.casefold()
venue = matches["venue"].fillna("").str.strip().str.casefold()
event = matches["event"].fillna("").str.strip().str.casefold()
teams = matches["teams"].fillna("").str.casefold()

# Indian-city evidence from the previous phase.
india_city = matches["venue_city_recognized_india"].fillna(False).astype(bool)

# Identify matches involving the Indian national team.
india_pattern = r"(?<![a-z])india(?![a-z])"
matches["india_involved"] = teams.str.contains(
    india_pattern, regex=True, na=False
)

# Event classification.
matches["is_world_cup"] = event.str.contains(
    r"world cup|worldcup", regex=True, na=False
)
matches["is_ipl"] = (
    event.str.contains(r"\bindian premier league\b|\bipl\b",
                       regex=True, na=False)
    | matches["source_archive"].fillna("").str.casefold().eq("ipl_csv2.zip")
)
matches["is_wpl"] = event.str.contains(
    r"\bwomen.?s premier league\b|\bwpl\b",
    regex=True, na=False
)

# City/venue indicators.
matches["is_hyderabad_match"] = (
    city.str.contains("hyderabad", regex=False)
    | venue.str.contains("hyderabad", regex=False)
)

# This is a preliminary classification only; it is not official verification.
matches["india_host_candidate"] = india_city

matches["host_country_status"] = "needs_external_verification"
matches.loc[
    matches["india_host_candidate"],
    "host_country_status"
] = "indian_city_candidate_not_yet_verified"

# Preserve verified host country as unknown until independently confirmed.
matches["host_country"] = pd.NA
matches["is_india_hosted"] = pd.Series(
    pd.NA, index=matches.index, dtype="boolean"
)

# Separate likely Indian-team matches outside recognized Indian cities for review.
matches["india_match_location_review"] = (
    matches["india_involved"] & ~matches["india_host_candidate"]
)

# Keep one row per source match ID in the master file.
matches = matches.drop_duplicates(subset=["match_id"]).copy()

# Review file: all Indian-city candidates plus India-team matches requiring review.
review = matches.loc[
    matches["india_host_candidate"] | matches["india_match_location_review"],
    [
        "match_id", "match_date", "teams", "match_type", "event",
        "venue", "city", "india_involved", "is_world_cup",
        "is_ipl", "is_wpl", "is_hyderabad_match",
        "india_host_candidate", "host_country_status"
    ]
].copy()

matches.to_csv(OUTPUT_FILE, index=False)
review.to_csv(REVIEW_FILE, index=False)

summary = pd.DataFrame({
    "metric": [
        "Total unique matches",
        "Indian-city candidates requiring verification",
        "India national-team matches",
        "Indian-team matches needing location review",
        "World Cup matches",
        "IPL matches",
        "WPL matches",
        "Hyderabad matches",
        "Host country still unverified"
    ],
    "count": [
        len(matches),
        int(matches["india_host_candidate"].sum()),
        int(matches["india_involved"].sum()),
        int(matches["india_match_location_review"].sum()),
        int(matches["is_world_cup"].sum()),
        int(matches["is_ipl"].sum()),
        int(matches["is_wpl"].sum()),
        int(matches["is_hyderabad_match"].sum()),
        int(matches["is_india_hosted"].isna().sum())
    ]
})

print("Phase 04.5.6 completed.")
display(summary)

print("\nFinal master dataset:", OUTPUT_FILE)
print("Venue review dataset:", REVIEW_FILE)

print("\nSample Hyderabad matches:")
display(
    matches.loc[
        matches["is_hyderabad_match"],
        ["match_date", "teams", "event", "venue", "city"]
    ].head(10)
)
````

````python
# Phase 04.5.7 — Validate Hyderabad match records

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data" / "processed" / "cricket"

INPUT_FILE = DATA_DIR / "cricket_matches_final_2023_2026.csv"
OUTPUT_FILE = DATA_DIR / "hyderabad_cricket_matches_review_2023_2026.csv"

matches = pd.read_csv(INPUT_FILE)

city = matches["city"].fillna("").str.strip().str.casefold()
venue = matches["venue"].fillna("").str.strip().str.casefold()

hyderabad = matches[
    city.str.contains("hyderabad", regex=False)
    | venue.str.contains("hyderabad", regex=False)
].copy()

# Review the actual venue and event combinations.
venue_summary = (
    hyderabad.groupby(["city", "venue", "event", "match_type"], dropna=False)
    .size()
    .reset_index(name="match_count")
    .sort_values(["city", "event", "venue"])
)

hyderabad["venue_verification_status"] = "requires source verification"
hyderabad["host_country"] = pd.NA
hyderabad["is_india_hosted"] = pd.Series(
    pd.NA, index=hyderabad.index, dtype="boolean"
)

hyderabad.to_csv(OUTPUT_FILE, index=False)

print("Hyderabad records found:", len(hyderabad))
print("Unique venues:", hyderabad["venue"].nunique())
print("Unique events:", hyderabad["event"].nunique())
print("Saved review file:", OUTPUT_FILE)

display(venue_summary)
````

````python
# Phase 04.5.8 — Consolidate cricket context dataset

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data" / "processed" / "cricket"

INPUT_FILE = DATA_DIR / "cricket_matches_final_2023_2026.csv"
FINAL_FILE = DATA_DIR / "cricket_context_final_2023_2026.csv"
QUALITY_FILE = DATA_DIR / "cricket_context_quality_report_2023_2026.csv"

matches = pd.read_csv(INPUT_FILE)

# Standardize dates and restrict to the research period.
matches["match_date"] = pd.to_datetime(
    matches["match_date"], errors="coerce"
)

matches = matches[
    matches["match_date"].between(
        "2023-01-01", "2026-07-31", inclusive="both"
    )
].copy()

# Remove duplicate match IDs while retaining the first source record.
matches = matches.drop_duplicates(subset="match_id", keep="first").copy()

# Ensure core context flags exist.
for column in [
    "india_involved", "is_world_cup", "is_ipl",
    "is_wpl", "is_hyderabad_match"
]:
    if column not in matches.columns:
        matches[column] = False
    matches[column] = matches[column].fillna(False).astype(bool)

# Do not infer the host country from team names or city alone.
if "host_country" not in matches.columns:
    matches["host_country"] = pd.NA

if "is_india_hosted" not in matches.columns:
    matches["is_india_hosted"] = pd.Series(
        pd.NA, index=matches.index, dtype="boolean"
    )

# Add useful date-level features for later demand-data integration.
matches["match_year"] = matches["match_date"].dt.year
matches["match_month"] = matches["match_date"].dt.month
matches["match_day_of_week"] = matches["match_date"].dt.day_name()

# Keep the output focused on available match-level context.
preferred_columns = [
    "match_id", "match_date", "match_year", "match_month",
    "match_day_of_week", "teams", "match_type", "team_type",
    "gender", "event", "venue", "city", "india_involved",
    "is_india_hosted", "host_country", "is_world_cup",
    "is_ipl", "is_wpl", "is_hyderabad_match",
    "host_country_status", "source_archive", "source_file"
]

final_columns = [
    column for column in preferred_columns
    if column in matches.columns
]

final_matches = matches[final_columns].sort_values(
    ["match_date", "match_id"]
).copy()

final_matches["match_date"] = final_matches["match_date"].dt.strftime(
    "%Y-%m-%d"
)

final_matches.to_csv(FINAL_FILE, index=False)

# Generate a concise data-quality report.
quality_checks = [
    ("Final match records", len(final_matches)),
    ("Duplicate match IDs", int(final_matches["match_id"].duplicated().sum())),
    ("Invalid/missing dates", int(final_matches["match_date"].isna().sum())),
    ("Missing teams", int(final_matches["teams"].isna().sum())),
    ("Missing event", int(final_matches["event"].isna().sum())),
    ("Missing venue", int(final_matches["venue"].isna().sum())),
    ("Missing city", int(final_matches["city"].isna().sum())),
    ("India-team matches", int(final_matches["india_involved"].sum())),
    ("World Cup matches", int(final_matches["is_world_cup"].sum())),
    ("IPL matches", int(final_matches["is_ipl"].sum())),
    ("WPL matches", int(final_matches["is_wpl"].sum())),
    ("Hyderabad matches", int(final_matches["is_hyderabad_match"].sum())),
    ("Host country still unknown", int(final_matches["host_country"].isna().sum())),
]

quality_report = pd.DataFrame(
    quality_checks, columns=["quality_check", "count"]
)
quality_report.to_csv(QUALITY_FILE, index=False)

print("Phase 04.5.8 completed.")
print("Final dataset:", FINAL_FILE)
print("Quality report:", QUALITY_FILE)
print(
    "Date range:",
    final_matches["match_date"].min(),
    "to",
    final_matches["match_date"].max()
)

display(quality_report)
display(final_matches.head(10))
````

````python
# Phase 04.5.11 — Resolve additional cricket venue locations

from pathlib import Path
import pandas as pd
import re

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data" / "processed" / "cricket"

INPUT_FILE = DATA_DIR / "cricket_context_host_classified_v2_2023_2026.csv"
OUTPUT_FILE = DATA_DIR / "cricket_context_host_classified_v3_2023_2026.csv"
REVIEW_FILE = DATA_DIR / "cricket_host_country_review_v3_2023_2026.csv"

matches = pd.read_csv(INPUT_FILE)

# Additional city-to-country mappings for locations in the review.
# These remain preliminary until validated against reliable venue sources.
EXTRA_COUNTRIES = {
    # Rwanda
    "kigali": "Rwanda",
    "kigali city": "Rwanda",

    # Botswana
    "gaborone": "Botswana",

    # Romania
    "ilfov county": "Romania",
    "bucharest": "Romania",

    # Guyana
    "providence": "Guyana",
    "georgetown": "Guyana",

    # Japan
    "sano": "Japan",

    # Malta
    "marsa": "Malta",

    # Brunei
    "bangi": "Brunei",

    # Nepal
    "kirtipur": "Nepal",
    "kathmandu": "Nepal",

    # Malaysia
    "kuala lumpur": "Malaysia",

    # Thailand
    "bangkok": "Thailand",

    # Indonesia
    "bali": "Indonesia",

    # Oman
    "al amarat": "Oman",

    # Other common cricket locations
    "lagos": "Nigeria",
    "kampala": "Uganda",
    "windhoek": "Namibia",
    "doha": "Qatar",
    "singapore": "Singapore",
    "hong kong": "Hong Kong",
    "amstelveen": "Netherlands",
    "rotterdam": "Netherlands",
    "edinburgh": "Scotland",
    "glasgow": "Scotland",
    "nairobi": "Kenya",
}

def normalize_location(value):
    if pd.isna(value):
        return ""
    return re.sub(r"\s+", " ", str(value).strip().casefold())

if "host_country" not in matches.columns:
    matches["host_country"] = pd.NA

city_norm = matches["city"].map(normalize_location)
extra_map = {
    normalize_location(city): country
    for city, country in EXTRA_COUNTRIES.items()
}

mapped_country = city_norm.map(extra_map)

# Keep existing mappings; fill only previously unresolved countries.
matches["host_country"] = matches["host_country"].where(
    matches["host_country"].notna(), mapped_country
)

matches["host_country_method"] = matches["host_country"].apply(
    lambda value: "preliminary_city_mapping"
    if pd.notna(value) else "unverified"
)

matches["host_country_status"] = matches["host_country"].apply(
    lambda value: "preliminary_mapping_requires_validation"
    if pd.notna(value) else "unverified_city"
)

# Recalculate candidate flags.
matches["is_india_hosted"] = pd.Series(
    pd.NA, index=matches.index, dtype="boolean"
)

mapped = matches["host_country"].notna()
matches.loc[mapped, "is_india_hosted"] = (
    matches.loc[mapped, "host_country"].eq("India")
)

teams = matches["teams"].fillna("").str.casefold()
matches["india_involved"] = teams.str.contains(
    r"(?<![a-z])india(?![a-z])", regex=True, na=False
)

matches["india_team_abroad_candidate"] = (
    matches["india_involved"]
    & matches["host_country"].notna()
    & matches["host_country"].ne("India")
)

matches.to_csv(OUTPUT_FILE, index=False)

unmapped = (
    matches.loc[matches["host_country"].isna()]
    .groupby(["city", "venue"], dropna=False)
    .agg(
        match_count=("match_id", "size"),
        events=("event", lambda values: " | ".join(
            sorted(set(values.dropna().astype(str)))[:8]
        ))
    )
    .reset_index()
    .sort_values("match_count", ascending=False)
)

unmapped.to_csv(REVIEW_FILE, index=False)

summary = pd.DataFrame({
    "metric": [
        "Total matches",
        "Preliminary India-host candidates",
        "Preliminary non-India-host matches",
        "India-team matches outside India (candidates)",
        "Matches with unverified cities",
        "Unmapped city/venue combinations"
    ],
    "count": [
        len(matches),
        int(matches["host_country"].eq("India").sum()),
        int((
            matches["host_country"].notna()
            & matches["host_country"].ne("India")
        ).sum()),
        int(matches["india_team_abroad_candidate"].sum()),
        int(matches["host_country"].isna().sum()),
        len(unmapped)
    ]
})

print("Phase 04.5.11 completed.")
display(summary)

print("\nUpdated dataset:", OUTPUT_FILE)
print("Remaining-location review:", REVIEW_FILE)
display(unmapped.head(20))
````

````python
# Phase 04.5.13 — Prioritize unresolved cricket venues

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data" / "processed" / "cricket"

INPUT_FILE = DATA_DIR / "cricket_context_host_classified_v4_2023_2026.csv"
OUTPUT_FILE = DATA_DIR / "cricket_unmapped_venue_priority_2023_2026.csv"

matches = pd.read_csv(INPUT_FILE)

unmapped = matches.loc[
    matches["host_country"].isna()
].copy()

priority = (
    unmapped.groupby(["city", "venue"], dropna=False)
    .agg(
        match_count=("match_id", "size"),
        first_match_date=("match_date", "min"),
        last_match_date=("match_date", "max"),
        events=("event", lambda s: " | ".join(
            sorted(set(s.dropna().astype(str)))[:5]
        )),
        teams=("teams", lambda s: " | ".join(
            sorted(set(s.dropna().astype(str)))[:5]
        ))
    )
    .reset_index()
    .sort_values(
        ["match_count", "city"],
        ascending=[False, True]
    )
)

priority.insert(
    0, "priority_rank",
    range(1, len(priority) + 1)
)

priority.to_csv(OUTPUT_FILE, index=False)

print("Phase 04.5.13 completed.")
print("Unresolved match records:", len(unmapped))
print("Unresolved city/venue combinations:", len(priority))
print("Saved review file:", OUTPUT_FILE)

display(priority.head(30))
````

````python
# Phase 04.5.16 — Prepare Hyderabad Cricket Context Dataset

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data" / "processed" / "cricket"

INPUT_FILE = DATA_DIR / "cricket_context_host_classified_v5_2023_2026.csv"
OUTPUT_FILE = DATA_DIR / "hyderabad_cricket_context_2023_2026.csv"
SUMMARY_FILE = DATA_DIR / "hyderabad_cricket_context_summary_2023_2026.csv"

START_DATE = "2023-01-01"
END_DATE = "2026-07-31"

matches = pd.read_csv(INPUT_FILE)
matches["match_date"] = pd.to_datetime(
    matches["match_date"], errors="coerce"
)

# Restrict to the project's research period.
matches = matches.loc[
    matches["match_date"].between(START_DATE, END_DATE)
].copy()

# Identify Hyderabad using the city and venue fields.
city = matches["city"].fillna("").str.strip().str.casefold()
venue = matches["venue"].fillna("").str.strip().str.casefold()

hyderabad_mask = (
    city.str.contains("hyderabad", regex=False)
    | venue.str.contains("hyderabad", regex=False)
)

hyderabad = matches.loc[hyderabad_mask].copy()

# Mark these records as requiring final venue/country verification.
hyderabad["hyderabad_location_match"] = True
hyderabad["host_country_verification"] = "pending"

# Save the match-level dataset.
hyderabad.to_csv(OUTPUT_FILE, index=False)

# Summarize matches by event and format.
summary = (
    hyderabad.groupby(["event", "match_type"], dropna=False)
    .size()
    .reset_index(name="match_count")
    .sort_values("match_count", ascending=False)
)

summary.to_csv(SUMMARY_FILE, index=False)

print("Phase 04.5.16 completed.")
print("Research period:", START_DATE, "to", END_DATE)
print("Hyderabad candidate matches:", len(hyderabad))
print("Unique venues:", hyderabad["venue"].nunique())
print("Output dataset:", OUTPUT_FILE)
print("Summary file:", SUMMARY_FILE)

display(summary)
display(
    hyderabad[
        ["match_date", "teams", "match_type", "event", "venue", "city"]
    ].head(15)
)
````

````python
# Phase 04.5.17 — Hyderabad Cricket Dataset Quality Check

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data" / "processed" / "cricket"

INPUT_FILE = DATA_DIR / "hyderabad_cricket_context_2023_2026.csv"
REPORT_FILE = DATA_DIR / "hyderabad_cricket_quality_report_2023_2026.csv"
DUPLICATES_FILE = DATA_DIR / "hyderabad_cricket_duplicate_review_2023_2026.csv"

df = pd.read_csv(INPUT_FILE)
df["match_date"] = pd.to_datetime(df["match_date"], errors="coerce")

required_columns = ["match_date", "teams", "match_type", "event", "venue", "city"]

report_rows = [
    {"check": "Total records", "result": len(df)},
    {"check": "Missing match dates", "result": int(df["match_date"].isna().sum())},
    {"check": "Missing teams", "result": int(df["teams"].isna().sum())},
    {"check": "Missing match type", "result": int(df["match_type"].isna().sum())},
    {"check": "Missing event", "result": int(df["event"].isna().sum())},
    {"check": "Missing venue", "result": int(df["venue"].isna().sum())},
    {"check": "Missing city", "result": int(df["city"].isna().sum())},
    {"check": "Unique venues", "result": int(df["venue"].nunique())},
    {"check": "First match date", "result": df["match_date"].min()},
    {"check": "Last match date", "result": df["match_date"].max()},
]

# Use available match identifiers for duplicate review.
if "match_id" in df.columns:
    duplicate_mask = df.duplicated(subset=["match_id"], keep=False)
else:
    duplicate_mask = df.duplicated(
        subset=["match_date", "teams", "match_type", "venue"],
        keep=False
    )

duplicates = df.loc[duplicate_mask].sort_values("match_date")

report_rows.append({
    "check": "Records involved in potential duplicates",
    "result": len(duplicates)
})

report = pd.DataFrame(report_rows)
report.to_csv(REPORT_FILE, index=False)
duplicates.to_csv(DUPLICATES_FILE, index=False)

print("Phase 04.5.17 completed.")
print("Quality report:", REPORT_FILE)
print("Duplicate review:", DUPLICATES_FILE)

display(report)
display(duplicates.head(10))
````

````python
# Phase 04.5.18 — Prepare Cricket Events for Demand Forecasting

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data" / "processed" / "cricket"

INPUT_FILE = DATA_DIR / "hyderabad_cricket_context_2023_2026.csv"
OUTPUT_FILE = DATA_DIR / "hyderabad_cricket_forecasting_features_2023_2026.csv"
SUMMARY_FILE = DATA_DIR / "hyderabad_cricket_forecasting_summary_2023_2026.csv"

df = pd.read_csv(INPUT_FILE)
df["match_date"] = pd.to_datetime(df["match_date"], errors="coerce")

# Retain the existing match fields and add calendar features.
df["match_year"] = df["match_date"].dt.year
df["match_month"] = df["match_date"].dt.month
df["match_day"] = df["match_date"].dt.day
df["day_of_week"] = df["match_date"].dt.dayofweek
df["is_weekend"] = df["day_of_week"].isin([5, 6])

# Create event-category features from the available source fields.
event_text = df["event"].fillna("").str.lower()
df["is_ipl"] = event_text.str.contains("indian premier league", regex=False)
df["is_world_cup"] = event_text.str.contains("world cup", regex=False)
df["is_international_series"] = (
    ~df["is_ipl"] & ~df["is_world_cup"]
)

# Match-day flag for later hourly demand feature engineering.
df["is_cricket_event_day"] = 1

df = df.sort_values("match_date").reset_index(drop=True)
df.to_csv(OUTPUT_FILE, index=False)

summary = (
    df.groupby(["match_year", "event", "match_type"], dropna=False)
      .size()
      .reset_index(name="match_count")
      .sort_values(["match_year", "match_count"], ascending=[True, False])
)
summary.to_csv(SUMMARY_FILE, index=False)

print("Phase 04.5.18 completed.")
print("Records:", len(df))
print("Date range:", df["match_date"].min(), "to", df["match_date"].max())
print("Forecasting dataset:", OUTPUT_FILE)
print("Event summary:", SUMMARY_FILE)

display(summary)
display(df.head(10))
````

````python
# Phase 04.5.19 — Final Cricket Dataset Validation

from pathlib import Path
import pandas as pd

# Locate the project root from the notebook's working directory
cwd = Path.cwd()
PROJECT_ROOT = next(
    (p for p in [cwd, *cwd.parents]
     if (p / "data" / "processed" / "cricket").is_dir()),
    None
)

if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "Cannot locate the project data folder. "
        "Run this notebook from inside your project directory."
    )

BASE_DIR = PROJECT_ROOT / "data" / "processed" / "cricket"
INPUT_FILE = BASE_DIR / "hyderabad_cricket_forecasting_features_2023_2026.csv"
REPORT_FILE = BASE_DIR / "hyderabad_cricket_final_quality_report_2023_2026.csv"

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"Expected dataset not found: {INPUT_FILE}\n"
        "Check the actual filename in data/processed/cricket."
    )

df = pd.read_csv(INPUT_FILE)

required_columns = ["match_date", "teams", "match_type", "event", "venue", "city"]
missing_columns = [c for c in required_columns if c not in df.columns]

if missing_columns:
    raise ValueError(f"Required columns missing: {missing_columns}")

dates = pd.to_datetime(df["match_date"], errors="coerce")
start_date = pd.Timestamp("2023-01-01")
end_date = pd.Timestamp("2026-07-31")

checks = [
    ("Total records", len(df)),
    ("Missing match dates", int(dates.isna().sum())),
    ("Dates before research period", int((dates < start_date).sum())),
    ("Dates after research period", int((dates > end_date).sum())),
    ("Duplicate rows", int(df.duplicated().sum())),
    ("Missing teams", int(df["teams"].isna().sum())),
    ("Missing match types", int(df["match_type"].isna().sum())),
    ("Missing event names", int(df["event"].isna().sum())),
    ("Missing venues", int(df["venue"].isna().sum())),
    ("Missing cities", int(df["city"].isna().sum())),
    ("Unique venues", int(df["venue"].nunique())),
    ("Unique events", int(df["event"].nunique())),
]

report = pd.DataFrame(checks, columns=["check", "result"])
report.to_csv(REPORT_FILE, index=False)

print("Phase 04.5.19 — Final validation completed")
print("Dataset:", INPUT_FILE)
print("Quality report:", REPORT_FILE)
print("Valid date range:", dates.min(), "to", dates.max())
display(report)
````

# Phase 04.5 — Cricket Context Data Collection and Preparation

## 1. Overview

Phase 04.5 focuses on collecting, classifying, and preparing cricket-match contextual data for the research project:

**Project Title:** *Hyperlocal Food Demand and Surge Forecasting in Indian Food Delivery: A Comparative Evaluation of Statistical, Machine Learning, Deep Learning, and Time-Series Foundation Models*

Cricket matches are included as an external contextual variable because major sporting events may influence food-delivery demand in nearby geographic zones and during relevant time periods.

The research period is **1 January 2023 to 31 July 2026**, with particular attention to matches hosted in Hyderabad and Indian national-team matches played both within India and abroad.

## 2. Data Source

**Source:** Cricsheet

Website: https://cricsheet.org/

Cricsheet provides downloadable cricket match datasets, including match metadata and ball-by-ball information for supported competitions and formats.

### Archives examined

* `all_csv2.zip` — broad collection of supported cricket matches.
* `ipl_csv2.zip` — IPL-specific match collection.

The files were inspected, and their match metadata was extracted for analysis.

**Coverage limitation:** Cricsheet does not necessarily contain every cricket match played worldwide. The resulting dataset must not be described as a complete record of every match.

## 3. Processing Activities

### 3.1 Match Metadata Extraction

Match-level metadata was extracted from the downloaded archives.

The resulting metadata included fields such as:

* Match ID
* Match date
* Teams
* Match format
* Team type
* Gender
* Tournament or event
* Venue
* City
* Source archive
* Source file

The extracted records were restricted to the research period of 1 January 2023 through 31 July 2026.

### 3.2 Match Classification

The records were classified to identify relevant categories, including:

* Indian national-team matches
* World Cup event matches
* Indian Premier League (IPL) matches
* Women's Premier League (WPL) matches
* Matches involving India
* Matches preliminarily mapped to India or outside India
* Matches associated with Hyderabad

The classification was used to support subsequent filtering and contextual analysis.

### 3.3 Venue and Host-Country Review

Venue and city information was reviewed to identify matches potentially hosted in India and to flag records requiring additional location verification.

Several iterations of location mapping were performed, and unresolved city/venue combinations were retained for review.

**Important:** Host-country classification remains provisional. It has not been established that every match has a verified host country.

### 3.4 Hyderabad Match Subset

A Hyderabad-specific dataset was created using the available venue and city metadata.

The dataset contains records associated with the Rajiv Gandhi International Stadium in Hyderabad.

The records cover the following event categories:

| Event                     | Match type | Records |
| ------------------------- | ---------- | ------: |
| Indian Premier League     | T20        |      26 |
| ICC Cricket World Cup     | ODI        |       3 |
| Bangladesh tour of India  | T20        |       1 |
| England tour of India     | Test       |       1 |
| New Zealand tour of India | ODI        |       1 |
| **Total**                 |            |  **32** |

These 32 records represent the Hyderabad matches identified in the current extracted dataset, not necessarily every match held in Hyderabad during the research period.

### 3.5 Forecasting Feature Preparation

A Hyderabad cricket-context dataset was prepared for subsequent use in the forecasting workflow.

The prepared dataset preserves match dates, participating teams, match type, event, venue, and city information, together with the available contextual features.

Cricket-context features can later be aligned with hourly food-order observations to investigate whether match periods coincide with changes in food demand.

No claim is made at this stage that cricket events cause demand increases.

## 4. Final Dataset and Quality Report

### Primary output

`data/processed/cricket/hyderabad_cricket_forecasting_features_2023_2026.csv`

Contains the prepared Hyderabad cricket-context records.

### Summary output

### Quality report

````python
# ============================================================
# PHASE 04.6.3: COLLECT SPORTING EVENT SOURCE PAGES
# ============================================================

from pathlib import Path
import requests
import pandas as pd

# 1. PROJECT PATHS
PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

SOURCE_DIR = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "sporting_events"
    / "source_pages"
)
SOURCE_DIR.mkdir(parents=True, exist_ok=True)

# 2. OFFICIAL SOURCE PAGES
SOURCES = {
    "hockey_tel angana": (
        "https://www.telanganahockey.org/national-events.php"
    ),
    "hockey_india": (
        "https://www.hockeyindia.org/"
    ),
    "pro_kabaddi": (
        "https://www.prokabaddi.com/"
    ),
}

# 3. DOWNLOAD PAGES ONLY WHEN MISSING
session = requests.Session()
session.headers.update({
    "User-Agent": "Mozilla/5.0"
})

results = []

for source_name, url in SOURCES.items():

    filename = source_name.replace(" ", "_") + ".html"
    output_path = SOURCE_DIR / filename

    if output_path.exists():
        print(f"Reusing existing source page: {filename}")
        results.append({
            "source_name": source_name,
            "source_url": url,
            "local_file": str(output_path),
            "status": "reused"
        })
        continue

    try:
        response = session.get(url, timeout=30)
        response.raise_for_status()

        output_path.write_text(
            response.text,
            encoding="utf-8"
        )

        results.append({
            "source_name": source_name,
            "source_url": url,
            "local_file": str(output_path),
            "status": "downloaded"
        })

        print(f"Downloaded: {source_name}")

    except requests.RequestException as exc:
        results.append({
            "source_name": source_name,
            "source_url": url,
            "local_file": "",
            "status": f"failed: {exc}"
        })

        print(f"Could not download {source_name}: {exc}")

# 4. REPORT RESULTS
source_report = pd.DataFrame(results)

print("\nSOURCE COLLECTION REPORT")
display(source_report)

print("Source directory:", SOURCE_DIR)
print("Successful sources:", int(
    source_report["status"].isin(["downloaded", "reused"]).sum()
))
````

````python
# ============================================================
# PHASE 04.6.3: INSPECT DOWNLOADED SPORTING SOURCE PAGES
# ============================================================

from pathlib import Path
from bs4 import BeautifulSoup
import pandas as pd
import re

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

SOURCE_DIR = (
    PROJECT_ROOT / "data" / "raw"
    / "sporting_events" / "source_pages"
)

KEYWORDS = [
    "hyderabad", "secunderabad", "schedule",
    "fixture", "match", "tournament", "2023",
    "2024", "2025", "2026"
]

if not SOURCE_DIR.exists():
    raise FileNotFoundError(f"Source folder not found: {SOURCE_DIR}")

html_files = sorted(SOURCE_DIR.glob("*.html"))

if not html_files:
    raise FileNotFoundError(f"No saved HTML pages found in: {SOURCE_DIR}")

results = []

for file_path in html_files:
    html = file_path.read_text(encoding="utf-8", errors="replace")
    soup = BeautifulSoup(html, "html.parser")

    # Remove navigation and scripts; retain page content
    for tag in soup(["script", "style", "nav", "footer", "header"]):
        tag.decompose()

    lines = [
        re.sub(r"\s+", " ", line).strip()
        for line in soup.get_text("\n").splitlines()
    ]
    lines = [line for line in lines if line]

    matches = [
        line for line in lines
        if any(keyword in line.lower() for keyword in KEYWORDS)
    ]

    for line in matches:
        results.append({
            "source_file": file_path.name,
            "page_text": line
        })

    print(f"{file_path.name}: {len(matches)} relevant text lines")

if not results:
    print("No relevant text found. These pages may require a dedicated schedule endpoint.")
else:
    source_text = pd.DataFrame(results)
    display(source_text.head(40))
    print("Relevant text lines found:", len(source_text))
````

````python
# PHASE 04.6.3 — FILTER HISTORICAL EVENT REFERENCES

import re
import pandas as pd

START_YEAR = 2023
END_YEAR = 2026

# Reuse source_text created by the previous cell
if "source_text" not in globals() or source_text.empty:
    raise ValueError("Run the source-page inspection cell first.")

pattern_year = re.compile(r"\b(?:2023|2024|2025|2026)\b")
pattern_location = re.compile(r"hyderabad|secunderabad|gachibowli", re.I)

historical_candidates = source_text[
    source_text["page_text"].str.contains(
        pattern_year, na=False
    )
    & source_text["page_text"].str.contains(
        pattern_location, na=False, regex=True
    )
].copy()

# Remove current/future dates beyond the study end where a full date is present
historical_candidates = historical_candidates.drop_duplicates()

print("Historical candidate lines:", len(historical_candidates))
display(historical_candidates.head(50))
````

````python
import pandas as pd
from pathlib import Path

# ============================================================
# PROJECT CONFIGURATION
# ============================================================

file_path = Path(
    r"d:\M Tech\Sem_03\Project\AI-Enhanced-Hyperlocal-Food-Demand-Forecasting"
    r"\data\raw\sporting_events\hyderabad_other_sporting_events_2023_2026.csv"
)

START_DATE = pd.Timestamp("2023-01-01")
END_DATE = pd.Timestamp("2026-07-31")

if not file_path.exists():
    raise FileNotFoundError(f"Existing project CSV not found: {file_path}")

df = pd.read_csv(file_path)

required_columns = [
    "event_date",
    "event_end_date",
    "event_name",
    "sport",
    "event_start_time",
    "team_or_participants",
    "venue",
    "city",
    "event_level",
    "source_url",
]

if list(df.columns) != required_columns:
    raise ValueError(
        "The existing CSV schema does not match the expected 10 columns. "
        "No records were added. Check the column names before proceeding."
    )

initial_count = len(df)

# ============================================================
# CONSOLIDATED REMAINING SPORTS EVENTS
# Candidate records supported by published event listings
# ============================================================

remaining_events = [
    # --------------------------------------------------------
    # JUDO
    # --------------------------------------------------------
    {
        "event_date": "2025-11-16",
        "event_end_date": "2025-11-20",
        "event_name": "Sub-Junior National Judo Championships 2025-26",
        "sport": "Judo",
        "event_start_time": None,
        "team_or_participants": "Sub-junior boys and girls",
        "venue": "Lal Bahadur Indoor Stadium",
        "city": "Hyderabad",
        "event_level": "National",
        "source_url": "https://www.onlinejfi.org/event-details.php?id=3641",
    },
    {
        "event_date": "2025-10-22",
        "event_end_date": "2025-10-22",
        "event_name": "Osmania University Inter-College Judo Tournament 2025",
        "sport": "Judo",
        "event_start_time": None,
        "team_or_participants": "University-affiliated college judokas",
        "venue": "M J College of Engineering and Technology, Banjara Hills",
        "city": "Hyderabad",
        "event_level": "Inter-College",
        "source_url": "https://www.lords.ac.in/sports/events-attended/",
    },
    {
        "event_date": "2024-11-07",
        "event_end_date": "2024-11-07",
        "event_name": "Osmania University Inter-College Judo Tournament 2024",
        "sport": "Judo",
        "event_start_time": None,
        "team_or_participants": "University-affiliated college judokas",
        "venue": "ISL Engineering College, Chandrayangutta",
        "city": "Hyderabad",
        "event_level": "Inter-College",
        "source_url": "https://www.lords.ac.in/sports/events-attended/",
    },

    # --------------------------------------------------------
    # TAEKWONDO
    # --------------------------------------------------------
    {
        "event_date": "2025-01-28",
        "event_end_date": "2025-01-30",
        "event_name": (
            "40th National Senior Kyorugi and 13th National Senior "
            "Poomsae Taekwondo Championships 2024-25"
        ),
        "sport": "Taekwondo",
        "event_start_time": None,
        "team_or_participants": "Senior competitors; Kyorugi and Poomsae",
        "venue": "Kotla Vijay Bhaskar Reddy Indoor Stadium, Yousufguda",
        "city": "Hyderabad",
        "event_level": "National",
        "source_url": (
            "https://www.kihapp.com/tournaments/"
            "18291-40th-national-senior-kyorugi-13th-national-senior-poosae-taekwond"
        ),
    },

    # --------------------------------------------------------
    # TABLE TENNIS
    # --------------------------------------------------------
    {
        "event_date": "2024-10-27",
        "event_end_date": "2024-10-27",
        "event_name": "1st Telangana State Veterans Table Tennis Ranking Tournament 2024",
        "sport": "Table Tennis",
        "event_start_time": None,
        "team_or_participants": "Veteran table tennis players",
        "venue": "Little Flower High School, Abids",
        "city": "Hyderabad",
        "event_level": "State",
        "source_url": "https://www.telanganatabletennis.com/Veterans-Events.php",
    },
    {
        "event_date": "2024-12-01",
        "event_end_date": "2024-12-01",
        "event_name": "2nd Telangana State Veterans Table Tennis Ranking Tournament 2024",
        "sport": "Table Tennis",
        "event_start_time": None,
        "team_or_participants": "Veteran table tennis players",
        "venue": "Shree Gujarati Seva Mandal",
        "city": "Secunderabad",
        "event_level": "State",
        "source_url": "https://www.telanganatabletennis.com/Veterans-Events.php",
    },
    {
        "event_date": "2025-01-19",
        "event_end_date": "2025-01-19",
        "event_name": "Veteran Telangana State Table Tennis Championships 2024",
        "sport": "Table Tennis",
        "event_start_time": None,
        "team_or_participants": "Veteran table tennis players",
        "venue": "St. Paul's High School, Hyderguda",
        "city": "Hyderabad",
        "event_level": "State",
        "source_url": "https://www.telanganatabletennis.com/Veterans-Events.php",
    },
    {
        "event_date": "2025-10-15",
        "event_end_date": "2025-10-16",
        "event_name": "Osmania University Inter-College Table Tennis Tournament 2025",
        "sport": "Table Tennis",
        "event_start_time": None,
        "team_or_participants": "University-affiliated college players",
        "venue": "St. Joseph's Degree and PG College, King Koti",
        "city": "Hyderabad",
        "event_level": "Inter-College",
        "source_url": "https://www.lords.ac.in/sports/events-attended/",
    },

    # --------------------------------------------------------
    # HANDBALL
    # --------------------------------------------------------
    {
        "event_date": "2025-09-26",
        "event_end_date": "2025-09-29",
        "event_name": "17th HFI Mini Boys and Girls National Handball Championship",
        "sport": "Handball",
        "event_start_time": None,
        "team_or_participants": "Mini boys and girls national teams",
        "venue": "Nizam College",
        "city": "Hyderabad",
        "event_level": "National",
        "source_url": "https://handballfederationofindia.com/previous-events/",
    },

    # --------------------------------------------------------
    # KARATE
    # --------------------------------------------------------
    {
        "event_date": "2024-11-02",
        "event_end_date": "2024-11-02",
        "event_name": "Osmania University Inter-College Karate Tournament 2024",
        "sport": "Karate",
        "event_start_time": None,
        "team_or_participants": "University-affiliated college karate competitors",
        "venue": "Lords Institute of Engineering and Technology, Himayath Sagar",
        "city": "Hyderabad",
        "event_level": "Inter-College",
        "source_url": "https://www.lords.ac.in/sports/events-attended/",
    },

    # --------------------------------------------------------
    # ARCHERY
    # --------------------------------------------------------
    {
        "event_date": "2025-09-22",
        "event_end_date": "2025-09-22",
        "event_name": "Osmania University National Archery Selection Trials 2025",
        "sport": "Archery",
        "event_start_time": None,
        "team_or_participants": "University archery competitors",
        "venue": "Telangana Sports School, Hakimpet",
        "city": "Hyderabad",
        "event_level": "University Selection Trials",
        "source_url": "https://www.lords.ac.in/sports/events-attended/",
    },
]

new_df = pd.DataFrame(remaining_events)

# Match the existing schema exactly
new_df = new_df[required_columns]

# ============================================================
# NORMALIZE AND VALIDATE DATES
# ============================================================

for col in ["event_date", "event_end_date"]:
    df[col] = pd.to_datetime(df[col], errors="coerce")
    new_df[col] = pd.to_datetime(new_df[col], errors="coerce")

# Retain only records whose dates fall within the study period
valid_date_mask = (
    new_df["event_date"].ge(START_DATE)
    & new_df["event_date"].le(END_DATE)
    & new_df["event_end_date"].ge(new_df["event_date"])
    & new_df["event_end_date"].le(END_DATE)
)

out_of_range_count = int((~valid_date_mask).sum())
new_df = new_df.loc[valid_date_mask].copy()

# Require key fields before insertion
required_values = [
    "event_date",
    "event_end_date",
    "event_name",
    "sport",
    "city",
    "source_url",
]

for col in required_values:
    new_df = new_df[
        new_df[col].notna()
        & new_df[col].astype(str).str.strip().ne("")
    ]

# ============================================================
# DUPLICATE CHECKING
# ============================================================

# Normalize date formats for consistent matching
for col in ["event_date", "event_end_date"]:
    df[col] = pd.to_datetime(df[col]).dt.strftime("%Y-%m-%d")
    new_df[col] = pd.to_datetime(new_df[col]).dt.strftime("%Y-%m-%d")

# Duplicate key based on date, name, sport and venue
dedupe_cols = ["event_date", "event_name", "sport", "venue"]

# Normalize whitespace and case in comparison keys
def make_keys(frame):
    normalized = frame[dedupe_cols].fillna("").astype(str)

    for col in dedupe_cols:
        normalized[col] = (
            normalized[col]
            .str.strip()
            .str.replace(r"\s+", " ", regex=True)
            .str.casefold()
        )

    return normalized.apply(tuple, axis=1)

existing_keys = set(make_keys(df))

new_df = new_df.loc[~make_keys(new_df).isin(existing_keys)].copy()

# Also remove duplicate rows within the proposed batch
new_df = new_df.drop_duplicates(subset=dedupe_cols, keep="first")

# ============================================================
# APPEND AND SAVE
# ============================================================

added_count = len(new_df)

df = pd.concat([df, new_df], ignore_index=True)

# Save to the SAME existing CSV
df.to_csv(file_path, index=False)

print("=" * 60)
print("CONSOLIDATED REMAINING SPORTS COLLECTION")
print("=" * 60)
print("Records before this batch:", initial_count)
print("New records added:", added_count)
print("Out-of-range records excluded:", out_of_range_count)
print("Total records in dataset:", len(df))
print("Sports represented in this batch:",
      ", ".join(sorted(new_df["sport"].dropna().unique())))
print("Saved to:", file_path)

display(
    new_df[
        [
            "event_date",
            "event_end_date",
            "event_name",
            "sport",
            "venue",
            "city",
            "source_url",
        ]
    ].sort_values(["event_date", "sport"])
)

print("\nCurrent record count by sport:")
display(
    df["sport"]
    .fillna("Unknown")
    .value_counts()
    .rename_axis("sport")
    .reset_index(name="record_count")
)
````

````python

import pandas as pd
from pathlib import Path

# ============================================================
# 04.6.4 — DATA QUALITY VALIDATION
# Existing Hyderabad Other Sporting Events Dataset
# ============================================================

file_path = Path(
    r"d:\M Tech\Sem_03\Project\AI-Enhanced-Hyperlocal-Food-Demand-Forecasting"
    r"\data\raw\sporting_events\hyderabad_other_sporting_events_2023_2026.csv"
)

# Load existing dataset
df = pd.read_csv(file_path)

# Define project period
START_DATE = pd.Timestamp("2023-01-01")
END_DATE = pd.Timestamp("2026-07-31")

required_columns = [
    "event_date",
    "event_end_date",
    "event_name",
    "sport",
    "event_start_time",
    "team_or_participants",
    "venue",
    "city",
    "event_level",
    "source_url"
]

print("=" * 65)
print("OTHER SPORTING EVENTS — DATA QUALITY AUDIT")
print("=" * 65)

# 1. Dataset overview
print("\n1. DATASET OVERVIEW")
print("-" * 40)
print(f"Total records: {len(df)}")
print(f"Total columns: {len(df.columns)}")

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    print("ERROR — Missing required columns:", missing_columns)
else:
    print("Required schema: PASS")

    # 2. Check missing values
    print("\n2. MISSING VALUES")
    print("-" * 40)

    missing_counts = df[required_columns].isna().sum()
    print(missing_counts[missing_counts > 0].to_string()
          if missing_counts.any() else "No null values found.")

    # Also detect blank text fields
    text_columns = [
        "event_name", "sport", "venue",
        "city", "event_level", "source_url"
    ]

    blank_counts = {}
    for col in text_columns:
        blank_counts[col] = (
            df[col].astype("string").str.strip().eq("").sum()
        )

    blank_counts = pd.Series(blank_counts)
    blank_counts = blank_counts[blank_counts > 0]

    print("\nBlank text fields:")
    print(blank_counts.to_string() if not blank_counts.empty
          else "No blank text fields found.")

    # 3. Validate dates
    print("\n3. DATE VALIDATION")
    print("-" * 40)

    for col in ["event_date", "event_end_date"]:
        df[col] = pd.to_datetime(df[col], errors="coerce")

    invalid_start = df["event_date"].isna().sum()
    invalid_end = df["event_end_date"].isna().sum()

    reversed_dates = (
        df["event_end_date"].notna()
        & df["event_date"].notna()
        & (df["event_end_date"] < df["event_date"])
    ).sum()

    out_of_period = (
        df["event_date"].notna()
        & (
            (df["event_date"] < START_DATE)
            | (df["event_date"] > END_DATE)
        )
    )

    print(f"Invalid event start dates: {invalid_start}")
    print(f"Invalid event end dates: {invalid_end}")
    print(f"End date before start date: {reversed_dates}")
    print(f"Records starting outside project period: {out_of_period.sum()}")

    # 4. Duplicate checks
    print("\n4. DUPLICATE CHECK")
    print("-" * 40)

    exact_duplicates = df.duplicated().sum()

    event_key = ["event_date", "event_name", "venue"]
    duplicate_event_keys = df.duplicated(
        subset=event_key, keep=False
    )

    print(f"Exact duplicate rows: {exact_duplicates}")
    print(
        "Rows sharing the same event date, name and venue: "
        f"{duplicate_event_keys.sum()}"
    )

    if duplicate_event_keys.any():
        print("\nReview these potential duplicate records:")
        print(
            df.loc[duplicate_event_keys, event_key]
            .sort_values(event_key)
            .to_string(index=False)
        )

    # 5. Source URL audit
    print("\n5. SOURCE URL AUDIT")
    print("-" * 40)

    urls = df["source_url"].fillna("").astype(str).str.strip()

    missing_urls = urls.eq("")
    invalid_url_format = ~urls.str.startswith(
        ("https://", "http://"), na=False
    ) & ~missing_urls

    print(f"Missing source URLs: {missing_urls.sum()}")
    print(f"URLs without http:// or https://: {invalid_url_format.sum()}")

    # 6. Geographic and sport summary
    print("\n6. CITY DISTRIBUTION")
    print("-" * 40)
    print(df["city"].value_counts(dropna=False).to_string())

    print("\n7. RECORD COUNT BY SPORT")
    print("-" * 40)
    print(
        df["sport"].value_counts(dropna=False)
        .to_string()
    )

    # 7. Final audit summary
    print("\n8. AUDIT SUMMARY")
    print("-" * 40)

    checks = {
        "Required columns present": not missing_columns,
        "No invalid start dates": invalid_start == 0,
        "No invalid end dates": invalid_end == 0,
        "No reversed event dates": reversed_dates == 0,
        "No start dates outside project period": out_of_period.sum() == 0,
        "No exact duplicate rows": exact_duplicates == 0,
        "All source URL fields populated": not missing_urls.any(),
    }

    for check, passed in checks.items():
        print(f"{'PASS' if passed else 'REVIEW'} — {check}")

    print("\n" + "=" * 65)
    print("AUDIT COMPLETE — ORIGINAL CSV WAS NOT MODIFIED")
    print("=" * 65)

````

````python

import pandas as pd
import requests
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor, as_completed

# ============================================================
# 04.6.6 — SOURCE URL ACCESSIBILITY CHECK
# Original CSV remains unchanged
# ============================================================

file_path = Path(
    r"d:\M Tech\Sem_03\Project\AI-Enhanced-Hyperlocal-Food-Demand-Forecasting"
    r"\data\raw\sporting_events\hyderabad_other_sporting_events_2023_2026.csv"
)

df = pd.read_csv(file_path)

urls = df["source_url"].dropna().astype(str).str.strip()
unique_urls = sorted(url for url in urls.unique() if url)

print("=" * 65)
print("04.6.6 — SOURCE URL ACCESSIBILITY CHECK")
print("=" * 65)
print(f"Dataset records: {len(df)}")
print(f"Unique URLs to check: {len(unique_urls)}")

def check_url(url):
    headers = {
        "User-Agent": "Mozilla/5.0 (compatible; ResearchDataAudit/1.0)"
    }

    try:
        response = requests.get(
            url,
            headers=headers,
            timeout=15,
            allow_redirects=True
        )

        return {
            "source_url": url,
            "status_code": response.status_code,
            "accessible": 200 <= response.status_code < 400,
            "final_url": response.url,
            "error": ""
        }

    except requests.RequestException as e:
        return {
            "source_url": url,
            "status_code": None,
            "accessible": False,
            "final_url": "",
            "error": str(e)[:180]
        }

results = []

with ThreadPoolExecutor(max_workers=5) as executor:
    futures = {
        executor.submit(check_url, url): url
        for url in unique_urls
    }

    for future in as_completed(futures):
        results.append(future.result())

url_audit = pd.DataFrame(results).sort_values("source_url")

print("\nURL ACCESSIBILITY SUMMARY")
print("-" * 65)
print(f"Accessible URLs: {url_audit['accessible'].sum()}")
print(f"URLs needing review: {(~url_audit['accessible']).sum()}")

print("\nURL CHECK RESULTS")
display(url_audit.reset_index(drop=True))

print("\nNOTE:")
print("An accessible URL does not prove that its event details are accurate.")
print("Blocked requests may also occur for legitimate websites.")
print("\nOriginal CSV remains unchanged.")
````

````python

# ============================================================
# 04.6.7 — REVIEW FAILED SOURCE URLS
# ============================================================

failed_urls = (
    url_audit.loc[
        ~url_audit["accessible"],
        ["source_url", "status_code", "final_url", "error"]
    ]
    .reset_index(drop=True)
)

print("=" * 70)
print("04.6.7 — SOURCE URLS REQUIRING REVIEW")
print("=" * 70)

print(f"URLs requiring review: {len(failed_urls)}")

display(failed_urls)

print("\nNo network requests made in this step.")
print("Original CSV remains unchanged.")
````

````python

# ============================================================
# 04.6.8 — MAP FAILED URLS TO SPORTING EVENTS
# Read-only inspection
# ============================================================

failed_urls = url_audit.loc[
    ~url_audit["accessible"]
].copy()

for number, row in enumerate(
    failed_urls.itertuples(index=False),
    start=1
):
    print(f"\n{'=' * 70}")
    print(f"SOURCE {number}")
    print(f"HTTP status: {row.status_code}")
    print(f"Error: {row.error}")
    print(f"Original URL: {row.source_url}")
    print(f"Final URL: {row.final_url}")

    matching_events = df[
        df["source_url"].astype(str).str.strip()
        == str(row.source_url).strip()
    ]

    print("\nAssociated sporting events:")

    display(
        matching_events[
            [
                "event_date",
                "event_end_date",
                "event_name",
                "sport",
                "venue",
                "city",
                "source_url"
            ]
        ].reset_index(drop=True)
    )

print("\nOriginal CSV remains unchanged.")
````

# Phase 04.6 — Collect Other Sporting Events

## 1. Objective

Collect sporting-event data for Hyderabad and Secunderabad, Telangana, covering **January 1, 2023, to July 31, 2026**, to support the contextual features of the Hyperlocal Food Demand and Surge Forecasting project.

Cricket data is handled separately and is excluded from this phase.

## 2. Dataset Details

| Attribute         | Details                               |
| ----------------- | ------------------------------------- |
| Dataset name      | Hyderabad Other Sporting Events       |
| Total records     | 67                                    |
| Number of columns | 10                                    |
| Geographic scope  | Hyderabad and Secunderabad, Telangana |
| Start date        | January 1, 2023                       |
| End date          | July 31, 2026                         |
| File format       | CSV                                   |
| Collection status | Complete for this phase               |
| Validation status | Preliminary validation completed      |

### Dataset Location

```text
data/raw/sporting_events/
└── hyderabad_other_sporting_events_2023_2026.csv
```

The existing CSV was updated during collection. Validation and inspection steps did not modify the original dataset.

## 3. Dataset Schema

The dataset contains the following 10 columns:

| Column                 | Description                               |
| ---------------------- | ----------------------------------------- |
| `event_date`           | Event start date                          |
| `event_end_date`       | Event end date                            |
| `event_name`           | Name of the sporting event                |
| `sport`                | Sport category                            |
| `event_start_time`     | Event start time, where available         |
| `team_or_participants` | Participating teams or individuals        |
| `venue`                | Event venue                               |
| `city`                 | Recorded city                             |
| `event_level`          | Event classification or competition level |
| `source_url`           | Reference URL associated with the event   |

## 4. Sports Covered

The dataset represents 22 sport categories, including:

* Hockey
* Kabaddi
* Football
* Badminton
* Running
* Basketball
* Volleyball
* Wrestling

## 8. Role in the Forecasting Project

Sporting events may provide contextual information for explaining changes in local food demand.

Potential features derived from this dataset include:

* `is_sporting_event_day`
* `sporting_event_count`
* `sporting_event_duration`
* `sport_category`
* `event_level`
* `is_major_sporting_event`

These features may help the forecasting models capture demand patterns associated with local competitions and sporting gatherings.

However, the presence of a sporting event does not establish that food demand increased. The predictive value of these features must be evaluated using the actual demand data and appropriate backtesting.

Event start times should not be used as verified temporal predictors when they are missing. Event-level features should be designed to work with the information reliably available.

6. The dataset contains selected collected events and should not be interpreted as a complete census of every sporting event held in Hyderabad during the study period.
7. The dataset alone cannot establish the causal effect of sporting events on food demand.

These limitations should be considered during feature engineering, model development, and interpretation of forecasting results.

````python
import re
import requests
import pandas as pd
from bs4 import BeautifulSoup
from pathlib import Path
from datetime import datetime

# ============================================================
# PHASE 04.7.2 — COLLECT HISTORICAL HITEX EVENTS
# Source: Official HITEX Past Events Archive
# Period: 2023-01-01 to 2026-07-31
# ============================================================

SOURCE_URL = "https://hitex.co.in/events/past.html"

PROJECT_DIR = Path(
    r"d:\M Tech\Sem_03\Project\AI-Enhanced-Hyperlocal-Food-Demand-Forecasting"
)

OUTPUT_FILE = (
    PROJECT_DIR
    / "data"
    / "raw"
    / "local_events"
    / "hyderabad_local_events_2023_2026.csv"
)

START_DATE = pd.Timestamp("2023-01-01")
END_DATE = pd.Timestamp("2026-07-31")

response = requests.get(
    SOURCE_URL,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=30
)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

date_pattern = re.compile(
    r"(\d{1,2}\s+[A-Za-z]{3}\s+\d{4})"
    r"(?:\s*-\s*(\d{1,2}\s+[A-Za-z]{3}\s+\d{4}))?"
)

def classify_event(name):
    text = name.lower()

    if any(k in text for k in [
        "food", "mithai", "sweet", "bakery", "horeca",
        "rice & grain", "dairy expo", "poultry india"
    ]):
        return "Food and Culinary"

    if any(k in text for k in [
        "concert", "music", "giggle fest", "soul sante",
        "lala land", "la la land", "dreamhack", "comic con",
        "kids' fair", "kids’ fair", "kite"
    ]):
        return "Entertainment and Public Events"

    if any(k in text for k in [
        "book fair", "literature", "literary", "craft mela",
        "deep mela", "style tatva", "design democracy",
        "art show", "cultural"
    ]):
        return "Cultural and Community"

    if any(k in text for k in [
        "expo", "exhibition", "fair", "show", "summit",
        "conference", "convention", "conclave", "fest",
        "mela", "bazaar", "market", "expo", "expos"
    ]):
        return "Exhibition and Trade Fair"

    return "Other Organised Event"


records = []

# The archive uses event headings followed by date and organizer links.
for heading in soup.find_all(["h2", "h3", "h4"]):
    event_name = heading.get_text(" ", strip=True)

    if not event_name:
        continue

    # Find the next date-bearing link associated with this event.
    date_text = None
    event_url = SOURCE_URL

    heading_link = heading.find("a", href=True)
    if heading_link:
        event_url = requests.compat.urljoin(
            SOURCE_URL, heading_link["href"]
        )

    for link in heading.find_all_next("a", limit=5):
        candidate = link.get_text(" ", strip=True)

        if date_pattern.search(candidate):
            date_text = candidate
            break

        # Stop if the next event heading is reached.
        next_heading = link.find_previous(
            ["h2", "h3", "h4"]
        )
        if next_heading is not None and next_heading is not heading:
            break

    if not date_text:
        continue

    match = date_pattern.search(date_text)
    if not match:
        continue

    try:
        event_start = pd.to_datetime(
            match.group(1), format="%d %b %Y"
        )
        event_end = (
            pd.to_datetime(match.group(2), format="%d %b %Y")
            if match.group(2)
            else event_start
        )
    except (ValueError, TypeError):
        continue

    # Keep events overlapping the requested study period.
    if event_end < START_DATE or event_start > END_DATE:
        continue

    records.append({
        "event_date": event_start.strftime("%Y-%m-%d"),
        "event_end_date": event_end.strftime("%Y-%m-%d"),
        "event_name": event_name,
        "event_category": classify_event(event_name),
        "event_start_time": None,
        "venue": "HITEX Exhibition Centre",
        "city": "Hyderabad",
        "latitude": None,
        "longitude": None,
        "zone_id": None,
        "event_level": None,
        "source_url": event_url
    })

# Build dataset
columns = [
    "event_date",
    "event_end_date",
    "event_name",
    "event_category",
    "event_start_time",
    "venue",
    "city",
    "latitude",
    "longitude",
    "zone_id",
    "event_level",
    "source_url"
]

events_df = pd.DataFrame(records, columns=columns)

# Save collected records
OUTPUT_FILE.parent.mkdir(parents=True, exist_ok=True)
events_df.to_csv(OUTPUT_FILE, index=False, encoding="utf-8-sig")

print("=" * 65)
print("PHASE 04.7.2 — HITEX EVENT COLLECTION")
print("=" * 65)
print(f"Source: {SOURCE_URL}")
print(f"Events collected: {len(events_df)}")
print(f"Output: {OUTPUT_FILE}")

if not events_df.empty:
    print("\nEvents by category:")
    print(events_df["event_category"].value_counts().to_string())

    print("\nSample collected events:")
    print(
        events_df[
            ["event_date", "event_end_date", "event_name", "event_category"]
        ].head(15).to_string(index=False)
    )

print("\nCollection step completed.")
````

````python
# ================================================================
# PHASE 04.7.7
# CONCERTS, ENTERTAINMENT & MAJOR PUBLIC GATHERINGS
#
# Study period: 2023-01-01 to 2026-07-31
# City: Hyderabad, Telangana, India
#
# Includes:
# 1. Telugu music directors and playback singers
# 2. Indian and international music concerts
# 3. Rock, electronic and independent music performances
# 4. Major entertainment and public gatherings
# 5. Previously collected candidate events
#
# Existing dataset is preserved.
# ================================================================

import pandas as pd
from pathlib import Path

# ------------------------------------------------
# 1. EXISTING DATASET PATH
# ------------------------------------------------

file_path = Path(
    r"d:\M Tech\Sem_03\Project\AI-Enhanced-Hyperlocal-Food-Demand-Forecasting"
    r"\data\raw\local_events\hyderabad_local_events_2023_2026.csv"
)

if not file_path.exists():
    raise FileNotFoundError(
        f"Existing local events dataset not found: {file_path}"
    )


# ------------------------------------------------
# 2. SOURCE REFERENCES
# ------------------------------------------------

SOURCE_CONCERT_ARCHIVES = (
    "https://www.concertarchives.org/locations/hyderabad-india"
)

SOURCE_HYDERABAD_TALKIES = (
    "https://hyderabadtalkies.events/past-events/"
)

SOURCE_DSP_OFFICIAL = (
    "https://actcevents.com/devi-sri-prasad-dsp-india-tour-tickets/"
)

SOURCE_KEERAVANI_HITEX = (
    "https://hitex.co.in/events/mm-keeravani-live-in-concert/index.html"
)

SOURCE_ED_SHEERAN = (
    "https://pastdates.edsheeran.com/tour-date/"
    "ramoji-film-city-hyderabad-india-sun-02022025-1200"
)

SOURCE_BRYAN_ADAMS = (
    "https://www.bryanadams.com/index.php/news/india-december-2024"
)

SOURCE_ANIRUDH = (
    "https://www.shreyasgroup.net/event-details/"
    "rockstar-anirudh-live-in-concert-hyderabad"
)

SOURCE_THAMAN_CANCELLED = (
    "https://www.123telugu.com/telugu/news/"
    "heavy-rain-and-storm-halt-thamans-ogtourindia-concert-in-hyderabad.html"
)

SOURCE_SHREYA_2023 = (
    "https://archive.siasat.com/news/"
    "tickets-prices-for-shreya-ghoshals-hyderabad-concert-2573848/"
)

SOURCE_SHREYA_2024 = (
    "https://hyderabadtalkies.events/past-events/"
)

SOURCE_ILLEYARAJA = (
    "https://www.indulgexpress.com/culture/music/2023/"
    "Feb/24/music-maestro-ilaiyaraaja-to-perform-live-music-concert-in-hyderabad-47405.html"
)

SOURCE_ARMAAN_MALIK = (
    "https://www.hindustantimes.com/entertainment/music/"
    "armaan-malik-takes-ed-sheeran-around-hyderabad-to-open-his-x-2025-india-tour-tonight-101738485573121.html"
)

SOURCE_SHREYAS_MEDIA = (
    "https://www.shreyasgroup.net/"
)

SOURCE_TELANGANA_TODAY = (
    "https://telanganatoday.com/"
    "hyderabad-all-ready-to-rock-with-star-studded-concerts-in-upcoming-weeks"
)


# ------------------------------------------------
# 3. CONSOLIDATED CONCERT AND ENTERTAINMENT RECORDS
# ------------------------------------------------
#
# research_status:
#   "Organizer/artist source" = primary source available
#   "Published event report" = published news/event report
#   "Archive candidate" = historical archive listing
#   "Needs confirmation" = venue/date/details need confirmation
#   "Cancelled" = do not treat as a completed attendance surge
#
# Dates below are candidate records, not a claim that every
# Hyderabad concert during the period has been discovered.
# ------------------------------------------------

events = [

    # ============================================================
    # 2023 — TELUGU / INDIAN MUSIC AND MAJOR ENTERTAINMENT
    # ============================================================

    {
        "event_date": "2023-02-25",
        "event_end_date": "2023-02-25",
        "event_name": "Ilaiyaraaja Tribute Concert",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Gachibowli",
        "event_level": "Major music concert",
        "source_url": SOURCE_ILLEYARAJA,
        "research_status": "Published event report",
    },

    {
        "event_date": "2023-02-26",
        "event_end_date": "2023-02-26",
        "event_name": "Ilaiyaraaja Live Concert",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Gachibowli",
        "event_level": "Major music concert",
        "source_url": SOURCE_ILLEYARAJA,
        "research_status": "Published event report",
    },

    {
        "event_date": "2023-03-04",
        "event_end_date": "2023-03-04",
        "event_name": "Martin Garrix Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "HITEX Exhibition Centre",
        "event_level": "International electronic music concert",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2023-04-30",
        "event_end_date": "2023-04-30",
        "event_name": "Shreya Ghoshal Live Concert",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Gachibowli Practice Stadium",
        "event_level": "Major playback singer concert",
        "source_url": SOURCE_SHREYA_2023,
        "research_status": "Published event report",
    },

    {
        "event_date": "2023-09-08",
        "event_end_date": "2023-09-08",
        "event_name": "WWE Superstar Spectacle 2023",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "G. M. C. Balayogi Indoor Stadium",
        "event_level": "Major live sports entertainment",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2023-08-25",
        "event_end_date": "2023-08-25",
        "event_name": "Radio Slave Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "EXT, Jubilee Hills",
        "event_level": "Electronic music performance",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Needs confirmation",
    },

    {
        "event_date": "2023-12-16",
        "event_end_date": "2023-12-16",
        "event_name": "Meca Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Breeze Monkey",
        "event_level": "Electronic music performance",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2023-12-17",
        "event_end_date": "2023-12-17",
        "event_name": "Massano Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": None,
        "event_level": "Electronic music performance",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Venue needs confirmation",
    },

    {
        "event_date": "2023-12-17",
        "event_end_date": "2023-12-17",
        "event_name": "Mali (Maalavika Manoj) Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "EXT, Jubilee Hills",
        "event_level": "Independent music performance",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Needs confirmation",
    },


    # ============================================================
    # 2024 — INDIAN AND INTERNATIONAL CONCERTS
    # ============================================================

    {
        "event_date": "2024-02-18",
        "event_end_date": "2024-02-18",
        "event_name": "Adam Beyer Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Prism",
        "event_level": "Electronic music performance",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2024-02-23",
        "event_end_date": "2024-02-23",
        "event_name": "Anjunadeep, Anyasa, CRi, HANA and Others",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": None,
        "event_level": "Electronic music event",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Venue needs confirmation",
    },

    {
        "event_date": "2024-04-27",
        "event_end_date": "2024-04-27",
        "event_name": "Girish and the Chronicles — Loaded Summer Tour",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Hard Rock Cafe, Hitech City",
        "event_level": "Rock concert",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2024-06-22",
        "event_end_date": "2024-06-22",
        "event_name": "Whiney, Degs, Anais and Others",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Antisocial Hyderabad",
        "event_level": "Electronic music event",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2024-08-10",
        "event_end_date": "2024-08-10",
        "event_name": "Anthracite Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Hard Rock Cafe, Hitech City",
        "event_level": "Rock/metal performance",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2024-10-19",
        "event_end_date": "2024-10-19",
        "event_name": "Rockstar DSP Live — Devi Sri Prasad",
        "event_category": "Entertainment and Public Events",
        "event_start_time": "18:30",
        "venue": "Gachibowli Stadium",
        "event_level": "Major Telugu music director concert",
        "source_url": SOURCE_DSP_OFFICIAL,
        "research_status": "Organizer/artist source",
    },

    {
        "event_date": "2024-11-15",
        "event_end_date": "2024-11-15",
        "event_name": "Candlelight: Tribute to Coldplay",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Radisson Blu Plaza Hotel, Banjara Hills",
        "event_level": "Candlelight music concert",
        "source_url": "https://newsroom.feverup.com/en-IN/246247-experience-the-candlelight-concert-series-now-launching-in-hyderabad/",
        "research_status": "Organizer/artist source",
    },

    {
        "event_date": "2024-11-21",
        "event_end_date": "2024-11-21",
        "event_name": "Ulcerate Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "EXT",
        "event_level": "International metal concert",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2024-11-30",
        "event_end_date": "2024-11-30",
        "event_name": "Shreya Ghoshal Live Concert",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "LB Stadium",
        "event_level": "Major playback singer concert",
        "source_url": SOURCE_SHREYA_2024,
        "research_status": "Organizer archive; date needs confirmation",
    },

    {
        "event_date": "2024-12-07",
        "event_end_date": "2024-12-07",
        "event_name": "Arijit Singh Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "GMR Arena",
        "event_level": "Major playback singer concert",
        "source_url": SOURCE_TELANGANA_TODAY,
        "research_status": "Needs confirmation",
    },

    {
        "event_date": "2024-12-16",
        "event_end_date": "2024-12-16",
        "event_name": "Bryan Adams — So Happy It Hurts Tour",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "GMR Arena",
        "event_level": "International music concert",
        "source_url": SOURCE_BRYAN_ADAMS,
        "research_status": "Artist/official tour source",
    },

    {
        "event_date": "2024-12-29",
        "event_end_date": "2024-12-29",
        "event_name": "Ben Böhmer Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": None,
        "event_level": "Electronic music performance",
        "source_url": SOURCE_TELANGANA_TODAY,
        "research_status": "Venue needs confirmation",
    },


    # ============================================================
    # 2025 — LARGE CONCERTS AND TELUGU MUSIC DIRECTORS
    # ============================================================

    {
        "event_date": "2025-02-02",
        "event_end_date": "2025-02-02",
        "event_name": "Ed Sheeran — Mathematics Tour",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Ramoji Film City",
        "event_level": "Major international concert",
        "source_url": SOURCE_ED_SHEERAN,
        "research_status": "Artist/official tour source",
    },

    {
        "event_date": "2025-02-02",
        "event_end_date": "2025-02-02",
        "event_name": "Armaan Malik — Opening Act for Ed Sheeran",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Ramoji Film City",
        "event_level": "Playback singer / concert opening act",
        "source_url": SOURCE_ARMAAN_MALIK,
        "research_status": "Published event report",
    },

    {
        "event_date": "2025-03-22",
        "event_end_date": "2025-03-22",
        "event_name": "M. M. Keeravaani — NAA TOUR",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "HITEX Exhibition Centre",
        "event_level": "Major Telugu music director concert",
        "source_url": SOURCE_KEERAVANI_HITEX,
        "research_status": "Organizer/venue source",
    },

    {
        "event_date": "2025-03-08",
        "event_end_date": "2025-03-08",
        "event_name": "John Summit Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Quake Arena",
        "event_level": "Electronic music concert",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2025-05-09",
        "event_end_date": "2025-05-09",
        "event_name": "Dimitri Vegas Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Quake Arena",
        "event_level": "Electronic music concert",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2025-05-31",
        "event_end_date": "2025-05-31",
        "event_name": "Girish and the Chronicles Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Hard Rock Cafe, Hitech City",
        "event_level": "Rock concert",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2025-09-21",
        "event_end_date": "2025-09-21",
        "event_name": "Parikrama Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Hard Rock Cafe, Hitech City",
        "event_level": "Rock concert",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2025-09-28",
        "event_end_date": "2025-09-28",
        "event_name": "Deborah De Luca Live",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Quake Arena",
        "event_level": "Electronic music concert",
        "source_url": SOURCE_CONCERT_ARCHIVES,
        "research_status": "Archive candidate",
    },

    {
        "event_date": "2025-11-08",
        "event_end_date": "2025-11-08",
        "event_name": "A. R. Rahman Live in Hyderabad",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": None,
        "event_level": "Major Indian music director concert",
        "source_url": SOURCE_HYDERABAD_TALKIES,
        "research_status": "Organizer archive; venue needs confirmation",
    },


    # ============================================================
    # 2026 — TELUGU MUSIC AND OTHER ENTERTAINMENT
    # ============================================================

    {
        "event_date": "2026-03-21",
        "event_end_date": "2026-03-21",
        "event_name": "Rockstar Anirudh XV — 15 Years With You",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Gachibowli Outdoor Stadium",
        "event_level": "Major Indian music director concert",
        "source_url": SOURCE_ANIRUDH,
        "research_status": "Organizer page; event details should be confirmed",
    },

    {
        "event_date": "2026-06-13",
        "event_end_date": "2026-06-13",
        "event_name": "S. S. Thaman — OG Tour India (Cancelled)",
        "event_category": "Entertainment and Public Events",
        "event_start_time": None,
        "venue": "Boulder Hills, Gachibowli",
        "event_level": "Cancelled concert",
        "source_url": SOURCE_THAMAN_CANCELLED,
        "research_status": "Cancelled",
    },

]


# ------------------------------------------------
# 4. CREATE THE NEW DATAFRAME
# ------------------------------------------------

new_df = pd.DataFrame(events)

# Common fields
new_df["city"] = "Hyderabad"
new_df["latitude"] = None
new_df["longitude"] = None
new_df["zone_id"] = None

# Provisional zone mapping only where the named venue is in Gachibowli.
# Exact mapping should follow your project's geographic zone boundaries.
for idx, row in new_df.iterrows():
    venue = str(row.get("venue", "")).lower()

    if "gachibowli" in venue:
        new_df.loc[idx, "zone_id"] = "Z03"


# ------------------------------------------------
# 5. LOAD THE EXISTING DATASET
# ------------------------------------------------

existing_df = pd.read_csv(file_path)


# ------------------------------------------------
# 6. ALIGN COLUMNS AND PRESERVE EXISTING DATA
# ------------------------------------------------

for col in existing_df.columns:
    if col not in new_df.columns:
        new_df[col] = None

for col in new_df.columns:
    if col not in existing_df.columns:
        existing_df[col] = None

# Keep exactly the same column order
new_df = new_df[existing_df.columns]


# ------------------------------------------------
# 7. APPEND AND REMOVE EXACT DUPLICATES
# ------------------------------------------------

combined_df = pd.concat(
    [existing_df, new_df],
    ignore_index=True
)

# Standardize date fields
for col in ["event_date", "event_end_date"]:
    combined_df[col] = pd.to_datetime(
        combined_df[col],
        errors="coerce"
    ).dt.strftime("%Y-%m-%d")

# Remove duplicate rows by event date, name and venue
combined_df = combined_df.drop_duplicates(
    subset=["event_date", "event_name", "venue"],
    keep="first"
)


# ------------------------------------------------
# 8. SAVE THE UPDATED DATASET
# ------------------------------------------------

combined_df.to_csv(
    file_path,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------
# 9. COLLECTION SUMMARY
# ------------------------------------------------

print("=" * 65)
print("PHASE 04.7.7 — CONSOLIDATED CONCERT COLLECTION COMPLETE")
print("=" * 65)

print(f"Candidate records prepared: {len(events)}")
print(f"Total records in dataset: {len(combined_df)}")
print(f"Output file: {file_path}")

print("\nConcert candidates by year:")
event_dates = pd.to_datetime(
    pd.DataFrame(events)["event_date"],
    errors="coerce"
)

print(
    event_dates.dt.year.value_counts().sort_index().to_string()
)

print("\nCollection completed.")
````

````python
from pathlib import Path
import pandas as pd

print("=" * 75)
print("PHASE 04.7.12 — LOCAL EVENTS DATA QUALITY AUDIT")
print("=" * 75)

file_path = Path(
    r"d:\M Tech\Sem_03\Project\AI-Enhanced-Hyperlocal-Food-Demand-Forecasting"
    r"\data\raw\local_events\hyderabad_local_events_2023_2026.csv"
)

df = pd.read_csv(file_path)

# Standardize date values for analysis only; do not modify the CSV.
df["event_date_parsed"] = pd.to_datetime(
    df["event_date"], errors="coerce"
)

study_start = pd.Timestamp("2023-01-01")
study_end = pd.Timestamp("2026-07-31")

# Duplicate event checks
duplicate_key = ["event_date", "event_name", "venue"]
available_key = [c for c in duplicate_key if c in df.columns]

if len(available_key) == 3:
    duplicate_rows = df.duplicated(
        subset=available_key, keep=False
    )
    duplicate_count = int(duplicate_rows.sum())
else:
    duplicate_rows = pd.Series(False, index=df.index)
    duplicate_count = None

# Missing values
important_columns = [
    "event_date",
    "event_name",
    "event_category",
    "venue",
    "source_url"
]
important_columns = [
    c for c in important_columns if c in df.columns
]
missing_counts = df[important_columns].isna().sum()

# Empty strings also count as missing.
for col in important_columns:
    if df[col].dtype == "object":
        missing_counts[col] += int(
            df[col].fillna("").astype(str).str.strip().eq("").sum()
        )

# Invalid and out-of-period dates
invalid_date_count = int(df["event_date_parsed"].isna().sum())

outside_period = (
    df["event_date_parsed"].notna()
    & (
        (df["event_date_parsed"] < study_start)
        | (df["event_date_parsed"] > study_end)
    )
)

# Research-status review
if "research_status" in df.columns:
    status = (
        df["research_status"]
        .fillna("Missing status")
        .astype(str)
        .str.strip()
    )
    uncertain_mask = status.str.contains(
        r"confirm|verify|unverified|candidate|uncertain|unknown|missing",
        case=False,
        regex=True
    )
    status_counts = status.value_counts(dropna=False)
else:
    uncertain_mask = pd.Series(False, index=df.index)
    status_counts = pd.Series(dtype="int64")

# Category distribution
if "event_category" in df.columns:
    category_counts = (
        df["event_category"]
        .fillna("Missing category")
        .astype(str)
        .value_counts()
    )
else:
    category_counts = pd.Series(dtype="int64")

print(f"\nTotal records: {len(df)}")
print(f"Total columns: {len(df.columns) - 1}")  # Excludes temporary parsed date
print(f"Invalid/missing event dates: {invalid_date_count}")
print(f"Records outside study period: {int(outside_period.sum())}")

if duplicate_count is not None:
    print(f"Rows involved in duplicate groups: {duplicate_count}")
else:
    print("Duplicate check skipped: required columns are missing.")

print("\n--- Missing values in important columns ---")
print(missing_counts.to_string())

print("\n--- Research-status distribution ---")
if len(status_counts):
    print(status_counts.to_string())
else:
    print("No research_status column found.")

print("\n--- Event category distribution ---")
if len(category_counts):
    print(category_counts.to_string())
else:
    print("No event_category column found.")

print("\n--- Records requiring research-status review ---")
print(f"Potentially uncertain records: {int(uncertain_mask.sum())}")

review_columns = [
    c for c in [
        "event_date", "event_name", "event_category",
        "venue", "zone_id", "research_status", "source_url"
    ]
    if c in df.columns
]

review_df = df.loc[uncertain_mask, review_columns].copy()

if len(review_df):
    display(review_df.head(30))
else:
    print("No records were flagged by the research-status text check.")

print("\n--- Duplicate records for review ---")
if duplicate_count:
    display(df.loc[duplicate_rows, review_columns].head(30))
else:
    print("No duplicate groups detected using date + event name + venue.")

print("\nAudit complete. No records were changed or deleted.")
````

````python
from pathlib import Path
import pandas as pd
import shutil

print("=" * 72)
print("PHASE 04.7.12 — STEP 3: UPDATE CONFIRMED VENUES")
print("=" * 72)

file_path = Path(
    r"d:\M Tech\Sem_03\Project\AI-Enhanced-Hyperlocal-Food-Demand-Forecasting"
    r"\data\raw\local_events\hyderabad_local_events_2023_2026.csv"
)

# Confirmed venue updates based on published event listings.
venue_updates = {
    ("2024-12-29", "Ben Böhmer Live"):
        "Quake Arena, Hyderabad",

    ("2025-11-08", "A. R. Rahman Live in Hyderabad"):
        "Ramoji Film City, Hyderabad"
}

if not file_path.exists():
    raise FileNotFoundError(f"Dataset not found: {file_path}")

df = pd.read_csv(file_path)

required_columns = [
    "event_date", "event_name", "venue"
]

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing required columns: {missing_columns}")

# Make a backup before modifying the dataset.
backup_path = file_path.with_name(
    file_path.stem + "_before_venue_updates.csv"
)
shutil.copy2(file_path, backup_path)

df["event_date"] = pd.to_datetime(
    df["event_date"], errors="coerce"
).dt.strftime("%Y-%m-%d")

if "research_status" not in df.columns:
    df["research_status"] = pd.NA

updated_count = 0

for (event_date, event_name), confirmed_venue in venue_updates.items():
    mask = (
        df["event_date"].eq(event_date)
        & df["event_name"].astype(str).str.contains(
            event_name, case=False, regex=False, na=False
        )
    )

    matched_rows = df.index[mask].tolist()

    if len(matched_rows) == 1:
        idx = matched_rows[0]
        df.at[idx, "venue"] = confirmed_venue
        df.at[idx, "research_status"] = (
            "Venue supported by published event listing"
        )
        updated_count += 1
        print(f"UPDATED: {event_name} -> {confirmed_venue}")

    elif len(matched_rows) == 0:
        print(f"NOT FOUND: {event_name} on {event_date}")

    else:
        print(
            f"AMBIGUOUS MATCH: {event_name} on {event_date}; "
            "no changes made to this event."
        )

# Flag the remaining two records without inventing their venues.
unresolved_names = ["Massano Live", "Anjunadeep"]

for name in unresolved_names:
    mask = (
        df["event_name"].astype(str).str.contains(
            name, case=False, regex=False, na=False
        )
        & (
            df["venue"].isna()
            | df["venue"].astype(str).str.strip().eq("")
        )
    )

    df.loc[mask, "research_status"] = (
        "Venue requires further verification"
    )

df.to_csv(file_path, index=False, encoding="utf-8-sig")

missing_venues = (
    df["venue"].isna()
    | df["venue"].astype(str).str.strip().eq("")
)

print("\n" + "=" * 72)
print("VENUE UPDATE SUMMARY")
print("=" * 72)
print(f"Records updated: {updated_count}")
print(f"Records still missing venues: {int(missing_venues.sum())}")
print(f"Dataset records: {len(df)}")
print(f"Backup saved: {backup_path}")
print(f"Updated dataset: {file_path}")
print("\nNo unverified venue names were invented.")
````

````python
from pathlib import Path
import pandas as pd
import shutil

print("=" * 76)
print("PHASE 04.7.13 — FINAL LOCAL EVENTS COLLECTION")
print("=" * 76)

file_path = Path(
    r"d:\M Tech\Sem_03\Project\AI-Enhanced-Hyperlocal-Food-Demand-Forecasting"
    r"\data\raw\local_events\hyderabad_local_events_2023_2026.csv"
)

if not file_path.exists():
    raise FileNotFoundError(f"Dataset not found: {file_path}")

df = pd.read_csv(file_path)

# ------------------------------------------------------------
# 1. ADDITIONAL PUBLIC EVENTS
# ------------------------------------------------------------
# These are additional candidate records based on published
# event information. Existing records are retained.
#
# Note:
# - Dates and venues should be checked against their source URLs.
# - No event is assigned to a study zone without justification.
# - These are additions, not a claim that every Hyderabad event
#   from 2023 through July 2026 has been collected.

events = [
    {
        "event_date": "2023-10-05",
        "event_end_date": "2023-10-15",
        "event_name": "SARAS Fair 2023",
        "event_category": "Public Fair and Community Event",
        "venue": "HMDA Ground, Hyderabad",
        "event_level": "Large public fair",
        "source_url": (
            "https://www.efms.serp.telangana.gov.in/"
            "Livelihoods/SARAS/Views/SarasHome.aspx"
        ),
        "research_status": "Government event record",
        "city": "Hyderabad",
        "event_start_time": "",
        "zone_id": "",
        "latitude": None,
        "longitude": None
    },
    {
        "event_date": "2024-09-26",
        "event_end_date": "2024-10-07",
        "event_name": "SARAS Fair 2024",
        "event_category": "Public Fair and Community Event",
        "venue": "People's Plaza, Necklace Road, Hyderabad",
        "event_level": "Large public fair",
        "source_url": (
            "https://www.efms.serp.telangana.gov.in/"
            "Livelihoods/SARAS/Views/SarasHome.aspx"
        ),
        "research_status": "Government event record",
        "city": "Hyderabad",
        "event_start_time": "",
        "zone_id": "",
        "latitude": None,
        "longitude": None
    },
    {
        "event_date": "2023-10-20",
        "event_end_date": "2023-10-22",
        "event_name": "Hyderabad Metro Fest",
        "event_category": "Cultural and Community",
        "venue": "Ameerpet Metro Station",
        "event_level": "Public cultural event",
        "source_url": (
            "https://www.ltmetro.in/pressreleases/2023/"
            "2023-11-02_hyderabad-metro-rail-metro-fest-"
            "exciting-3-day-metro-fest-for-hyderabad-metro-"
            "rail-commuters"
        ),
        "research_status": "Published event report",
        "city": "Hyderabad",
        "event_start_time": "15:00",
        "zone_id": "",
        "latitude": None,
        "longitude": None
    },
    {
        "event_date": "2023-12-31",
        "event_end_date": "2023-12-31",
        "event_name": "LuLu Mall NYE Fiesta",
        "event_category": "Entertainment and Public Events",
        "venue": "LuLu Mall, Kukatpally",
        "event_level": "Mall public event",
        "source_url": "https://www.hyderabad.lulumall.in/events-3/",
        "research_status": "Mall event archive",
        "city": "Hyderabad",
        "event_start_time": "",
        "zone_id": "",
        "latitude": None,
        "longitude": None
    },
    {
        "event_date": "2024-01-26",
        "event_end_date": "2024-01-27",
        "event_name": "LuLu Mall Rhythm of Republic",
        "event_category": "Entertainment and Public Events",
        "venue": "LuLu Mall, Kukatpally",
        "event_level": "Mall public event",
        "source_url": "https://www.hyderabad.lulumall.in/events-3/",
        "research_status": "Mall event archive",
        "city": "Hyderabad",
        "event_start_time": "18:00",
        "zone_id": "",
        "latitude": None,
        "longitude": None
    },
    {
        "event_date": "2024-03-10",
        "event_end_date": "2024-03-10",
        "event_name": "LuLu Wonder Women",
        "event_category": "Community and Entertainment",
        "venue": "LuLu Mall, Kukatpally",
        "event_level": "Mall public event",
        "source_url": "https://www.hyderabad.lulumall.in/events-3/",
        "research_status": "Mall event archive",
        "city": "Hyderabad",
        "event_start_time": "",
        "zone_id": "",
        "latitude": None,
        "longitude": None
    },
    {
        "event_date": "2025-06-13",
        "event_end_date": "2025-06-13",
        "event_name": "LuLu UIC Fight Night 2",
        "event_category": "Sports and Entertainment",
        "venue": "LuLu Mall, Kukatpally",
        "event_level": "Public sporting entertainment",
        "source_url": "https://www.hyderabad.lulumall.in/events-3/",
        "research_status": "Mall event archive",
        "city": "Hyderabad",
        "event_start_time": "",
        "zone_id": "",
        "latitude": None,
        "longitude": None
    },
    {
        "event_date": "2025-05-22",
        "event_end_date": "2025-05-22",
        "event_name": "LuLu Mall Summer Games",
        "event_category": "Sports and Entertainment",
        "venue": "LuLu Mall, Kukatpally",
        "event_level": "Mall public event",
        "source_url": "https://www.hyderabad.lulumall.in/events-3/",
        "research_status": "Mall event archive",
        "city": "Hyderabad",
        "event_start_time": "",
        "zone_id": "",
        "latitude": None,
        "longitude": None
    }
]

new_df = pd.DataFrame(events)

# ------------------------------------------------------------
# 2. STANDARDIZE EXISTING DATA AND PREPARE SCHEMA
# ------------------------------------------------------------

if "event_date" not in df.columns:
    raise ValueError("The existing dataset has no event_date column.")

if "event_name" not in df.columns:
    raise ValueError("The existing dataset has no event_name column.")

if "venue" not in df.columns:
    raise ValueError("The existing dataset has no venue column.")

df["event_date"] = pd.to_datetime(
    df["event_date"], errors="coerce"
).dt.strftime("%Y-%m-%d")

new_df["event_date"] = pd.to_datetime(
    new_df["event_date"], errors="coerce"
).dt.strftime("%Y-%m-%d")

new_df["event_end_date"] = pd.to_datetime(
    new_df["event_end_date"], errors="coerce"
).dt.strftime("%Y-%m-%d")

# Restrict additions to the study period.
new_df = new_df[
    new_df["event_date"].between("2023-01-01", "2026-07-31")
].copy()

# Preserve existing columns and add new columns only when needed.
for col in new_df.columns:
    if col not in df.columns:
        df[col] = pd.NA

for col in df.columns:
    if col not in new_df.columns:
        new_df[col] = pd.NA

new_df = new_df[df.columns]

# ------------------------------------------------------------
# 3. DEDUPLICATE
# ------------------------------------------------------------

key_cols = ["event_date", "event_name", "venue"]

for col in key_cols:
    df[col] = df[col].fillna("").astype(str).str.strip()
    new_df[col] = new_df[col].fillna("").astype(str).str.strip()

existing_keys = set(
    df[key_cols].itertuples(index=False, name=None)
)

new_df["_dedupe_key"] = list(
    new_df[key_cols].itertuples(index=False, name=None)
)

new_df = new_df[
    ~new_df["_dedupe_key"].isin(existing_keys)
].copy()

new_df = new_df.drop_duplicates(subset=key_cols)
new_df = new_df.drop(columns=["_dedupe_key"])

# ------------------------------------------------------------
# 4. BACK UP AND SAVE
# ------------------------------------------------------------

backup_path = file_path.with_name(
    file_path.stem + "_before_phase_04_7_13.csv"
)

shutil.copy2(file_path, backup_path)

added_count = len(new_df)

df = pd.concat([df, new_df], ignore_index=True)

df.to_csv(file_path, index=False, encoding="utf-8-sig")

print("\n" + "=" * 76)
print("PHASE 04.7.13 COMPLETED")
print("=" * 76)
print(f"Candidate records prepared : {len(events)}")
print(f"New records added          : {added_count}")
print(f"Existing records retained  : {len(df) - added_count}")
print(f"Total records now          : {len(df)}")
print(f"Backup saved               : {backup_path}")
print(f"Updated dataset            : {file_path}")
print("\nCollection cell completed.")
````

````python
from pathlib import Path
import pandas as pd
import requests
import time
from urllib.parse import urlparse

print("=" * 80)
print("PHASE 04.7 — FINAL SOURCE AND DATE VERIFICATION")
print("=" * 80)

# ------------------------------------------------------------
# 1. PATHS AND STUDY PERIOD
# ------------------------------------------------------------

base_dir = Path(
    r"d:\M Tech\Sem_03\Project\AI-Enhanced-Hyperlocal-Food-Demand-Forecasting"
)

file_path = (
    base_dir
    / "data"
    / "raw"
    / "local_events"
    / "hyderabad_local_events_2023_2026.csv"
)

report_path = (
    base_dir
    / "data"
    / "raw"
    / "local_events"
    / "hyderabad_local_events_final_verification_report.csv"
)

START_DATE = pd.Timestamp("2023-01-01")
END_DATE = pd.Timestamp("2026-07-31")

if not file_path.exists():
    raise FileNotFoundError(f"Dataset not found: {file_path}")

df = pd.read_csv(file_path)

print(f"\nDataset loaded: {len(df)} records")

# ------------------------------------------------------------
# 2. CHECK REQUIRED COLUMNS
# ------------------------------------------------------------

required_columns = [
    "event_date",
    "event_name",
    "venue",
    "source_url"
]

for col in required_columns:
    if col not in df.columns:
        df[col] = pd.NA

if "research_status" not in df.columns:
    df["research_status"] = pd.NA

# ------------------------------------------------------------
# 3. DATE VALIDATION
# ------------------------------------------------------------

parsed_dates = pd.to_datetime(
    df["event_date"],
    errors="coerce"
)

date_missing_or_invalid = parsed_dates.isna()

outside_period = (
    parsed_dates.notna()
    & (
        (parsed_dates < START_DATE)
        | (parsed_dates > END_DATE)
    )
)

# Check event_end_date if available.
if "event_end_date" in df.columns:
    parsed_end_dates = pd.to_datetime(
        df["event_end_date"],
        errors="coerce"
    )

    end_date_present = (
        df["event_end_date"].notna()
        & df["event_end_date"].astype(str).str.strip().ne("")
    )

    invalid_end_date = (
        end_date_present
        & parsed_end_dates.isna()
    )

    end_before_start = (
        parsed_dates.notna()
        & parsed_end_dates.notna()
        & (parsed_end_dates < parsed_dates)
    )

    end_outside_period = (
        parsed_end_dates.notna()
        & (
            (parsed_end_dates < START_DATE)
            | (parsed_end_dates > END_DATE)
        )
    )
else:
    parsed_end_dates = pd.Series(
        pd.NaT, index=df.index
    )
    invalid_end_date = pd.Series(
        False, index=df.index
    )
    end_before_start = pd.Series(
        False, index=df.index
    )
    end_outside_period = pd.Series(
        False, index=df.index
    )

# ------------------------------------------------------------
# 4. MISSING FIELD CHECKS
# ------------------------------------------------------------

def missing_text(series):
    return (
        series.isna()
        | series.astype(str).str.strip().eq("")
    )

missing_event_name = missing_text(df["event_name"])
missing_venue = missing_text(df["venue"])
missing_source = missing_text(df["source_url"])

# ------------------------------------------------------------
# 5. DUPLICATE CHECK
# ------------------------------------------------------------

duplicate_key = [
    "event_date",
    "event_name",
    "venue"
]

for col in duplicate_key:
    df[col] = df[col].fillna("").astype(str).str.strip()

duplicate_mask = df.duplicated(
    subset=duplicate_key,
    keep=False
)

# ------------------------------------------------------------
# 6. RESEARCH-STATUS REVIEW
# ------------------------------------------------------------

status = (
    df["research_status"]
    .fillna("")
    .astype(str)
    .str.strip()
)

status_missing = status.eq("")

uncertain_status = status.str.contains(
    r"needs confirmation|need confirmation|"
    r"requires verification|requires further verification|"
    r"unverified|candidate|to confirm|"
    r"venue requires|date requires|"
    r"needs verification|verify final",
    case=False,
    regex=True,
    na=False
)

# ------------------------------------------------------------
# 7. SOURCE URL FORMAT CHECK
# ------------------------------------------------------------

def valid_http_url(value):
    if pd.isna(value):
        return False

    value = str(value).strip()

    if not value:
        return False

    try:
        parsed = urlparse(value)
        return (
            parsed.scheme in ("http", "https")
            and bool(parsed.netloc)
        )
    except Exception:
        return False

source_url_format_invalid = ~df["source_url"].apply(
    valid_http_url
)

# ------------------------------------------------------------
# 8. CHECK SOURCE URL ACCESSIBILITY
# ------------------------------------------------------------
# This checks accessibility only, not whether the page confirms
# the specific event date or venue.
#
# To avoid excessive requests, each unique URL is checked once.
# Network errors are recorded as inconclusive.

unique_urls = (
    df.loc[~source_url_format_invalid, "source_url"]
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
)

url_results = {}

session = requests.Session()
session.headers.update({
    "User-Agent": "Mozilla/5.0 (compatible; ResearchDataAudit/1.0)"
})

print(f"\nChecking {len(unique_urls)} unique source URLs...")
print("This may take a few minutes depending on website response times.")

for i, url in enumerate(unique_urls, start=1):
    try:
        response = session.get(
            url,
            timeout=12,
            allow_redirects=True,
            stream=True
        )

        status_code = response.status_code
        final_url = response.url

        if 200 <= status_code < 400:
            result = "Accessible — content still needs event verification"
        elif status_code == 404:
            result = "Page not found — investigate source"
        elif status_code in (401, 403):
            result = "Access restricted — manual verification needed"
        else:
            result = f"HTTP {status_code} — investigate source"

        response.close()

        url_results[url] = {
            "source_http_status": status_code,
            "source_access_result": result,
            "source_final_url": final_url
        }

    except requests.RequestException as exc:
        url_results[url] = {
            "source_http_status": pd.NA,
            "source_access_result": (
                "Inconclusive — network or website error: "
                + type(exc).__name__
            ),
            "source_final_url": pd.NA
        }

    if i % 25 == 0:
        print(f"Checked {i}/{len(unique_urls)} unique URLs")

# ------------------------------------------------------------
# 9. BUILD VERIFICATION REPORT
# ------------------------------------------------------------

report = df.copy()

report["parsed_event_date"] = parsed_dates
report["parsed_event_end_date"] = parsed_end_dates

report["date_missing_or_invalid"] = date_missing_or_invalid
report["event_date_outside_study_period"] = outside_period
report["invalid_end_date"] = invalid_end_date
report["end_date_before_start_date"] = end_before_start
report["event_end_date_outside_study_period"] = end_outside_period

report["missing_event_name"] = missing_event_name
report["missing_venue"] = missing_venue
report["missing_source_url"] = missing_source
report["invalid_source_url_format"] = source_url_format_invalid

report["duplicate_record"] = duplicate_mask
report["missing_research_status"] = status_missing
report["uncertain_research_status"] = uncertain_status

report["source_http_status"] = pd.NA
report["source_access_result"] = "Not checked"
report["source_final_url"] = pd.NA

for idx, row in report.iterrows():
    url = str(row["source_url"]).strip()

    if url in url_results:
        report.at[idx, "source_http_status"] = (
            url_results[url]["source_http_status"]
        )
        report.at[idx, "source_access_result"] = (
            url_results[url]["source_access_result"]
        )
        report.at[idx, "source_final_url"] = (
            url_results[url]["source_final_url"]
        )
    elif bool(row["invalid_source_url_format"]):
        report.at[idx, "source_access_result"] = (
            "Missing or invalid source URL"
        )

# Determine which records require manual review.
review_flags = [
    "date_missing_or_invalid",
    "event_date_outside_study_period",
    "invalid_end_date",
    "end_date_before_start_date",
    "event_end_date_outside_study_period",
    "missing_event_name",
    "missing_venue",
    "missing_source_url",
    "invalid_source_url_format",
    "duplicate_record",
    "uncertain_research_status"
]

report["requires_manual_review"] = report[review_flags].any(axis=1)

report["verification_note"] = report.apply(
    lambda row: "; ".join(
        label for column, label in [
            ("date_missing_or_invalid", "Invalid or missing event date"),
            ("event_date_outside_study_period", "Event date outside study period"),
            ("invalid_end_date", "Invalid event end date"),
            ("end_date_before_start_date", "End date precedes start date"),
            ("event_end_date_outside_study_period", "End date outside study period"),
            ("missing_event_name", "Missing event name"),
            ("missing_venue", "Missing venue"),
            ("missing_source_url", "Missing source URL"),
            ("invalid_source_url_format", "Invalid source URL format"),
            ("duplicate_record", "Duplicate key"),
            ("missing_research_status", "Missing research status"),
            ("uncertain_research_status", "Research status indicates uncertainty")
        ]
        if bool(row.get(column, False))
    ),
    axis=1
)

# ------------------------------------------------------------
# 10. EXPORT REPORT AND PRINT SUMMARY
# ------------------------------------------------------------

report.to_csv(
    report_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 80)
print("FINAL VERIFICATION REPORT")
print("=" * 80)

print(f"Total records reviewed              : {len(report)}")
print(f"Invalid/missing event dates         : {int(date_missing_or_invalid.sum())}")
print(f"Event dates outside study period    : {int(outside_period.sum())}")
print(f"Invalid end dates                   : {int(invalid_end_date.sum())}")
print(f"End dates before start dates        : {int(end_before_start.sum())}")
print(f"Events missing names                : {int(missing_event_name.sum())}")
print(f"Events missing venues               : {int(missing_venue.sum())}")
print(f"Events missing source URLs          : {int(missing_source.sum())}")
print(f"Invalid source URL formats          : {int(source_url_format_invalid.sum())}")
print(f"Rows in duplicate groups            : {int(duplicate_mask.sum())}")
print(f"Missing research-status labels      : {int(status_missing.sum())}")
print(f"Uncertain research-status labels    : {int(uncertain_status.sum())}")
print(f"Records requiring manual review     : {int(report['requires_manual_review'].sum())}")

access_results = pd.Series(
    [value["source_access_result"] for value in url_results.values()]
).value_counts()

print("\n--- Unique source URL accessibility results ---")
print(access_results.to_string())

print("\nVerification report saved to:")
print(report_path)

print("\nOriginal event dataset was not modified.")
print("Source accessibility does not establish that event dates are correct.")
print("Manual review of the report is required before declaring verification complete.")
````

````python
from pathlib import Path
import pandas as pd
import shutil

print("=" * 75)
print("PHASE 04.7 — FINAL DATE-BOUNDARY REVIEW")
print("=" * 75)

file_path = Path(
    r"d:\M Tech\Sem_03\Project\AI-Enhanced-Hyperlocal-Food-Demand-Forecasting"
    r"\data\raw\local_events\hyderabad_local_events_2023_2026.csv"
)

if not file_path.exists():
    raise FileNotFoundError(f"Dataset not found: {file_path}")

df = pd.read_csv(file_path)

# Preserve a backup before updating metadata.
backup_path = file_path.with_name(
    file_path.stem + "_before_final_boundary_review.csv"
)
shutil.copy2(file_path, backup_path)

# Parse dates without changing the original date values.
start_dates = pd.to_datetime(df["event_date"], errors="coerce")

if "event_end_date" in df.columns:
    end_dates = pd.to_datetime(df["event_end_date"], errors="coerce")
else:
    end_dates = start_dates.copy()
    df["event_end_date"] = df["event_date"]

study_end = pd.Timestamp("2026-07-31")

# Identify events that start within the study period but end afterward.
boundary_mask = (
    start_dates.between(
        pd.Timestamp("2023-01-01"),
        study_end
    )
    & end_dates.notna()
    & (end_dates > study_end)
)

# Add a review column without overwriting existing notes.
if "final_boundary_review" not in df.columns:
    df["final_boundary_review"] = ""

df.loc[boundary_mask, "final_boundary_review"] = (
    "Event extends beyond study end; retain only in-period dates "
    "when constructing contextual features."
)

# Save without deleting records or altering original dates.
df.to_csv(file_path, index=False, encoding="utf-8-sig")

print(f"Total records retained       : {len(df)}")
print(f"Events extending past cutoff : {int(boundary_mask.sum())}")
print(f"Study period end             : {study_end.date()}")
print(f"Backup saved                 : {backup_path}")
print(f"Updated dataset              : {file_path}")

if boundary_mask.any():
    print("\nEvents flagged for boundary review:")
    display(
        df.loc[
            boundary_mask,
            [
                "event_date",
                "event_end_date",
                "event_name",
                "final_boundary_review"
            ]
        ]
    )

print("\nNo event records were deleted.")
````

````python
import requests
from bs4 import BeautifulSoup
from pathlib import Path
from datetime import datetime
import pandas as pd

url = "https://echallan.tspolice.gov.in/htp/"

response = requests.get(url, timeout=30)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

# Extract the published page text
page_text = soup.get_text("\n", strip=True)

# Save the source snapshot for traceability
output_dir = Path("data/raw/traffic")
output_dir.mkdir(parents=True, exist_ok=True)

snapshot_path = output_dir / "hyderabad_traffic_police_source_snapshot.txt"
snapshot_path.write_text(page_text, encoding="utf-8")

# Store the actual source snapshot; do not invent individual records
df = pd.DataFrame([{
    "collection_timestamp": datetime.now().astimezone().isoformat(),
    "source_url": url,
    "http_status": response.status_code,
    "page_title": soup.title.get_text(strip=True) if soup.title else None,
    "source_text": page_text
}])

output_csv = output_dir / "hyderabad_traffic_police_source_snapshot.csv"
df.to_csv(output_csv, index=False, encoding="utf-8-sig")

print("Source retrieved successfully.")
print("HTTP status:", response.status_code)
print("Text length:", len(page_text))
print("Saved:", snapshot_path)
print("Saved:", output_csv)
````

````python
from pathlib import Path
from datetime import datetime
import pandas as pd
import re

# Load the saved official-source snapshot
source_file = Path("data/raw/traffic/hyderabad_traffic_police_source_snapshot.txt")
text = source_file.read_text(encoding="utf-8")

# Identify advisory sections using headings present in the source
pattern = re.compile(
    r"(?:TRAFFIC\s+ADVISORY|TRAFFIC\s+ALERT)\s*:?",
    re.IGNORECASE
)

matches = list(pattern.finditer(text))
records = []

for i, match in enumerate(matches):
    start = match.start()
    end = matches[i + 1].start() if i + 1 < len(matches) else len(text)
    advisory_text = text[start:end].strip()

    # Keep only sections containing meaningful text
    if len(advisory_text) < 40:
        continue

    records.append({
        "collection_timestamp": datetime.now().astimezone().isoformat(),
        "advisory_text": advisory_text,
        "source_url": "https://echallan.tspolice.gov.in/htp/"
    })

# Save only text extracted from the genuine source
output_dir = Path("data/raw/traffic")
output_dir.mkdir(parents=True, exist_ok=True)

output_file = output_dir / "hyderabad_traffic_advisories.csv"

df = pd.DataFrame(records)

if not df.empty:
    df = df.drop_duplicates(subset=["advisory_text"])
    df.to_csv(output_file, index=False, encoding="utf-8-sig")
    print("Extracted advisories:", len(df))
    print("Saved to:", output_file)
    print("\nFirst extracted records:")
    print(df[["advisory_text"]].head(3).to_string(index=False))
else:
    print("No identifiable advisory sections found.")
    print("The source snapshot remains unchanged.")
````

````python
from pathlib import Path
from datetime import datetime
import pandas as pd
import re

source_file = Path("data/raw/traffic/hyderabad_traffic_police_source_snapshot.txt")
text = source_file.read_text(encoding="utf-8")

# Identify actual advisory headings
pattern = re.compile(
    r"(?im)^\s*(?:TRAFFIC\s+ADVISORY|TRAFFIC\s+ALERT)\s*:?\s*"
)

matches = list(pattern.finditer(text))
records = []

for i, match in enumerate(matches):
    start = match.start()
    end = matches[i + 1].start() if i + 1 < len(matches) else len(text)
    advisory = text[start:end].strip()

    # Remove generic website navigation and unrelated footer text
    advisory = re.split(
        r"(?i)Citizen Services|Pay Challan|Emergency & Important Contacts",
        advisory
    )[0].strip()

    if len(advisory) < 60:
        continue

    records.append({
        "collection_timestamp": datetime.now().astimezone().isoformat(),
        "advisory_text": advisory,
        "source_url": "https://echallan.tspolice.gov.in/htp/"
    })

output_file = Path("data/raw/traffic/hyderabad_traffic_advisories.csv")

df = pd.DataFrame(records).drop_duplicates(subset=["advisory_text"])
df.to_csv(output_file, index=False, encoding="utf-8-sig")

print("Candidate advisory sections:", len(df))
print("Saved to:", output_file)

for i, row in df.iterrows():
    print(f"\n--- Candidate {i + 1} ---")
    print(row["advisory_text"][:1000])
````

````python
from pathlib import Path
from datetime import datetime
import pandas as pd
import re

source_file = Path("data/raw/traffic/hyderabad_traffic_police_source_snapshot.txt")
text = source_file.read_text(encoding="utf-8")

# Identify titled traffic advisories in the source text
pattern = re.compile(
    r"(?:🚦\s*)?(TRAFFIC\s+ADVISORY\s*[|–-]\s*[^\\n]+|"
    r"Traffic Advisory\s*[-–]\s*[^\\n]+)",
    re.IGNORECASE
)

matches = list(pattern.finditer(text))
records = []

for i, match in enumerate(matches):
    start = match.start()
    end = matches[i + 1].start() if i + 1 < len(matches) else len(text)
    section = text[start:end].strip()

    # Exclude empty or navigation-only sections
    if len(section) < 60:
        continue

    records.append({
        "advisory_title": match.group(1).strip(),
        "advisory_text": section,
        "collection_timestamp": datetime.now().astimezone().isoformat(),
        "source_url": "https://echallan.tspolice.gov.in/htp/"
    })

output_dir = Path("data/raw/traffic")
output_dir.mkdir(parents=True, exist_ok=True)
output_file = output_dir / "hyderabad_traffic_advisories.csv"

df = pd.DataFrame(records)

if not df.empty:
    df = df.drop_duplicates(subset=["advisory_title", "advisory_text"])
    df.to_csv(output_file, index=False, encoding="utf-8-sig")
    print("Advisory candidates extracted:", len(df))
    print("Saved:", output_file)
    print(df[["advisory_title"]].to_string(index=False))
else:
    print("No matching advisory titles found.")
    print("Original source snapshot remains unchanged.")
````
