The Open-Meteo Historical Weather API provides historical weather information using reanalysis datasets, including ERA5 and ERA5-Land.

Potential hourly variables include:

* Relative humidity.
* Precipitation.
* Rainfall.
* Cloud cover.
* Wind speed.
* Weather codes.

**Potential application:** Historical weather analysis and experiments evaluating the association between weather conditions and synthetic food demand.

**Limitation:** Historical reanalysis data represent reconstructed weather conditions. They do not establish which weather forecasts were available when a historical demand prediction would have been made.

#### Source B: Open-Meteo Historical Forecast API

The Open-Meteo Historical Forecast API provides archived weather forecasts, with historical coverage beginning around 2022 according to its documentation.

Official documentation:
https://open-meteo.com/en/docs/historical-forecast-api

Potential hourly variables include:

* Temperature.
* Relative humidity.
* Precipitation probability.
* Precipitation.
* Rainfall.
* Cloud cover.
* Wind speed.
* Weather codes.

**Potential application:** Historical forecasting experiments that use weather forecast information rather than actual future weather observations.

**Limitation:** Available forecast dates, model runs, lead times, and variables must be verified for the required study period. A continuous hourly forecast archive does not necessarily preserve every original forecast run and lead time.

#### Source C: Open-Meteo Weather Forecast API

The Open-Meteo Weather Forecast API provides current and future weather forecasts for specified geographical coordinates.

Official documentation:
https://open-meteo.com/en/docs

Potential hourly variables include:

* Temperature.
* Relative humidity.
* Precipitation probability.
* Precipitation.
* Cloud cover.
* Wind speed.
* Weather codes.

**Potential application:** Supplying weather forecasts to the live demand forecasting workflow.

**Limitation:** Forecast availability and horizon depend on the selected API configuration and weather models. Current forecasts cannot substitute for archived forecasts in historical backtesting.

```python
import requests
import pandas as pd

# Provisional city-level test location for Hyderabad.
# This is a connectivity-test point, not a validated coordinate
# for any individual hyperlocal zone.
latitude = 17.3850
longitude = 78.4867

# Test one day within the documented project period.
start_date = "2026-08-30"
end_date = "2026-08-30"

url = "https://archive-api.open-meteo.com/v1/archive"

params = {
    "latitude": latitude,
    "longitude": longitude,
    "start_date": start_date,
    "end_date": end_date,
    "hourly": [
        "temperature_2m",
        "relative_humidity_2m",
        "precipitation",
        "rain",
        "wind_speed_10m",
        "cloud_cover"
    ],
    "timezone": "Asia/Kolkata"
}

try:
    response = requests.get(url, params=params, timeout=30)

    print("HTTP status:", response.status_code)
    response.raise_for_status()

    weather_data = response.json()

    print("\nAPI response received successfully.")
    print("Requested location:", latitude, longitude)
    print("Returned coordinates:",
          weather_data.get("latitude"),
          weather_data.get("longitude"))
    print("Timezone:", weather_data.get("timezone"))
    print("Elevation:", weather_data.get("elevation"))

    hourly_data = weather_data.get("hourly", {})
    weather_df = pd.DataFrame(hourly_data)

    print("\nReturned variables:")
    print(weather_df.columns.tolist())

    print("\nDataset shape:", weather_df.shape)

    if not weather_df.empty:
        print("\nFirst five records:")
        display(weather_df.head())

        print("\nLast timestamp returned:",
              weather_df["time"].iloc[-1])

        print("\nMissing values by variable:")
        print(weather_df.isna().sum())

except requests.exceptions.RequestException as error:
    print("API request failed:", error)

except (ValueError, KeyError) as error:
    print("Response validation failed:", error)
```

## 04.2.7 — Historical Weather Data Coverage and Completeness Assessment

### Objective

The objective of this subsection is to evaluate the temporal coverage, hourly continuity, and data completeness of the Open-Meteo Historical Weather API for the project's study period.

The required study period is January 1, 2023, through July 31, 2026. The selected location is Hyderabad, Telangana, India, using the `Asia/Kolkata` timezone.

### Data Source

- **Provider:** Open-Meteo
- **API:** Historical Weather API
- **Endpoint:** https://archive-api.open-meteo.com/v1/archive
- **Requested Latitude:** 17.3850
- **Requested Longitude:** 78.4867
- **Timezone:** Asia/Kolkata

### Variables

The following hourly weather variables will be examined:

1. Temperature at 2 metres
2. Relative humidity
3. Precipitation
4. Rainfall
5. Wind speed at 10 metres
6. Cloud cover

### Assessment Procedure

The assessment will:

1. Request historical weather data in manageable monthly batches.
2. record successful and unsuccessful API requests.
3. Count the returned hourly records.
4. Compare the returned timestamps with the expected hourly timeline.
5. Identify missing timestamps and missing variable values.
6. Check for duplicate timestamps.
7. Summarize coverage by month.
8. Determine whether the retrieved data is suitable for further processing.

### Expected Outcome

The assessment will provide evidence of the API's availability and completeness over the requested period.

A successful API response alone will not be considered sufficient proof of complete coverage. Actual timestamps, record counts, duplicate records, missing values, and failed requests will be examined.

### Important Limitations

The API returns gridded historical weather estimates, not independent weather observations for each synthetic delivery zone.

The initial assessment uses one Hyderabad coordinate. The returned grid-cell coordinates will be recorded because they may differ from the requested coordinates.

This assessment evaluates historical weather data only. It does not establish that the same information was available as a forecast at the original demand-prediction time.

```python
# ============================================================
# 04.2.7 — Historical Weather Coverage Assessment
# ============================================================

import time
import requests
import pandas as pd
from IPython.display import display

# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

API_URL = "https://archive-api.open-meteo.com/v1/archive"

LATITUDE = 17.3850
LONGITUDE = 78.4867
TIMEZONE = "Asia/Kolkata"

START_DATE = "2023-01-01"
END_DATE = "2026-07-31"

WEATHER_VARIABLES = [
    "temperature_2m",
    "relative_humidity_2m",
    "precipitation",
    "rain",
    "wind_speed_10m",
    "cloud_cover",
]

EXPECTED_COLUMNS = ["time"] + WEATHER_VARIABLES

session = requests.Session()

# ------------------------------------------------------------
# 2. Generate monthly request periods
# ------------------------------------------------------------

start = pd.Timestamp(START_DATE)
end = pd.Timestamp(END_DATE)

month_starts = pd.date_range(
    start=start.replace(day=1),
    end=end.replace(day=1),
    freq="MS"
)

monthly_periods = []

for month_start in month_starts:
    month_end = month_start + pd.offsets.MonthEnd(0)

    period_start = max(month_start, start)
    period_end = min(month_end, end)

    monthly_periods.append(
        (
            period_start.strftime("%Y-%m-%d"),
            period_end.strftime("%Y-%m-%d")
        )
    )

print("Assessment period:", START_DATE, "to", END_DATE)
print("Monthly API requests:", len(monthly_periods))
print("Weather variables:", len(WEATHER_VARIABLES))

# ------------------------------------------------------------
# 3. Retrieve data month by month
# ------------------------------------------------------------

monthly_data = []
request_log = []

for period_start, period_end in monthly_periods:

    params = {
        "latitude": LATITUDE,
        "longitude": LONGITUDE,
        "start_date": period_start,
        "end_date": period_end,
        "hourly": WEATHER_VARIABLES,
        "timezone": TIMEZONE,
    }

    try:
        response = session.get(
            API_URL,
            params=params,
            timeout=60
        )

        response.raise_for_status()
        payload = response.json()

        if "hourly" not in payload or "time" not in payload["hourly"]:
            raise ValueError("API response does not contain hourly timestamps.")

        month_df = pd.DataFrame(payload["hourly"])

        missing_columns = [
            column for column in EXPECTED_COLUMNS
            if column not in month_df.columns
        ]

        if missing_columns:
            raise ValueError(
                f"Missing expected columns: {missing_columns}"
            )

        month_df = month_df[EXPECTED_COLUMNS].copy()
        month_df["time"] = pd.to_datetime(month_df["time"])

        monthly_data.append(month_df)

        request_log.append({
            "period_start": period_start,
            "period_end": period_end,
            "status": "SUCCESS",
            "http_status": response.status_code,
            "records": len(month_df),
            "returned_latitude": payload.get("latitude"),
            "returned_longitude": payload.get("longitude"),
            "error": None,
        })

        print(
            f"SUCCESS | {period_start} to {period_end} "
            f"| {len(month_df):,} hourly records"
        )

    except Exception as exc:
        request_log.append({
            "period_start": period_start,
            "period_end": period_end,
            "status": "FAILED",
            "http_status": (
                response.status_code
                if "response" in locals()
                else None
            ),
            "records": 0,
            "returned_latitude": None,
            "returned_longitude": None,
            "error": str(exc),
        })

        print(
            f"FAILED  | {period_start} to {period_end} "
            f"| {type(exc).__name__}: {exc}"
        )

    # Small pause between requests
    time.sleep(0.2)

# ------------------------------------------------------------
# 4. Build the assessment tables
# ------------------------------------------------------------

request_log_df = pd.DataFrame(request_log)

successful_months = request_log_df[
    request_log_df["status"] == "SUCCESS"
]

failed_months = request_log_df[
    request_log_df["status"] == "FAILED"
]

print("\n--- Request Summary ---")
print("Total monthly requests:", len(request_log_df))
print("Successful requests:", len(successful_months))
print("Failed requests:", len(failed_months))

display(request_log_df)

# ------------------------------------------------------------
# 5. Combine successful monthly responses
# ------------------------------------------------------------

if not monthly_data:
    raise RuntimeError(
        "No monthly weather data was retrieved. "
        "Inspect request_log_df before proceeding."
    )

weather_coverage_df = pd.concat(
    monthly_data,
    ignore_index=True
)

weather_coverage_df = weather_coverage_df.sort_values(
    "time"
).reset_index(drop=True)

print("\n--- Retrieved Data Summary ---")
print("Shape:", weather_coverage_df.shape)
print("First timestamp:", weather_coverage_df["time"].min())
print("Last timestamp:", weather_coverage_df["time"].max())

print("\n--- Missing Values by Column ---")
display(weather_coverage_df.isna().sum().to_frame("missing_values"))

# ------------------------------------------------------------
# 6. Check duplicate timestamps
# ------------------------------------------------------------

duplicate_count = weather_coverage_df["time"].duplicated().sum()

print("\n--- Duplicate Timestamp Check ---")
print("Duplicate timestamps:", duplicate_count)

# ------------------------------------------------------------
# 7. Check expected hourly timeline
# ------------------------------------------------------------

expected_timeline = pd.date_range(
    start=pd.Timestamp(START_DATE),
    end=pd.Timestamp(END_DATE) + pd.Timedelta(hours=23),
    freq="h"
)

actual_timeline = pd.DatetimeIndex(
    weather_coverage_df["time"].drop_duplicates().sort_values()
)

missing_timestamps = expected_timeline.difference(actual_timeline)
unexpected_timestamps = actual_timeline.difference(expected_timeline)

print("\n--- Hourly Timeline Check ---")
print("Expected hourly timestamps:", len(expected_timeline))
print("Unique returned timestamps:", len(actual_timeline))
print("Missing timestamps:", len(missing_timestamps))
print("Unexpected timestamps:", len(unexpected_timestamps))

if len(missing_timestamps) > 0:
    print("\nFirst 20 missing timestamps:")
    print(missing_timestamps[:20].tolist())

if len(unexpected_timestamps) > 0:
    print("\nFirst 20 unexpected timestamps:")
    print(unexpected_timestamps[:20].tolist())

# ------------------------------------------------------------
# 8. Summarize actual monthly coverage
# ------------------------------------------------------------

weather_coverage_df["month"] = (
    weather_coverage_df["time"].dt.to_period("M").astype(str)
)

monthly_coverage = (
    weather_coverage_df
    .groupby("month")
    .agg(
        returned_records=("time", "size"),
        unique_timestamps=("time", "nunique"),
    )
    .reset_index()
)

monthly_coverage["expected_records"] = (
    monthly_coverage["month"]
    .apply(
        lambda month: len(
            pd.date_range(
                start=pd.Period(month, freq="M").start_time,
                end=pd.Period(month, freq="M").end_time.floor("h"),
                freq="h"
            )
        )
    )
)

monthly_coverage["record_difference"] = (
    monthly_coverage["returned_records"]
    - monthly_coverage["expected_records"]
)

print("\n--- Monthly Coverage Summary ---")
display(monthly_coverage)

# ------------------------------------------------------------
# 9. Final preliminary assessment
# ------------------------------------------------------------

all_requests_succeeded = len(failed_months) == 0
no_duplicate_timestamps = duplicate_count == 0
no_missing_timestamps = len(missing_timestamps) == 0
no_missing_values = weather_coverage_df[
    WEATHER_VARIABLES
].isna().sum().sum() == 0

print("\n--- Preliminary Assessment ---")
print("All monthly requests succeeded:", all_requests_succeeded)
print("No duplicate timestamps:", no_duplicate_timestamps)
print("No missing timestamps:", no_missing_timestamps)
print("No missing weather values:", no_missing_values)

print(
    "\nNote: This assessment checks the returned data. "
    "Review monthly coverage and failed requests before "
    "deciding whether the dataset is ready for integration."
)
```

## 04.2.10 — Retrieve and Save the Historical Weather Dataset

### Objective
Retrieve the verified historical weather series from the Open-Meteo Historical Weather API and save the data, metadata, and coverage reports in the project repository.

### Configuration
- Period: January 1, 2023 – July 31, 2026
- Location: Hyderabad, India
- Timezone: Asia/Kolkata
- Resolution: Hourly
- Retrieval method: Monthly API requests

### Output Files
1. `data/raw/weather/hyderabad_historical_weather.csv`
2. `data/metadata/weather_source_metadata.json`
3. `results/weather_coverage/monthly_coverage.csv`
4. `results/weather_coverage/request_log.csv`

### Data Integrity
The retrieval process will:
- Retrieve data in monthly batches.
- Validate expected columns and timestamps.
- Check missing values, duplicate timestamps, and hourly continuity.
- Preserve the coordinates and metadata returned by the API.
- Keep temporary analysis columns out of the raw weather CSV.
- Avoid overwriting existing files.
- Reopen the saved CSV and verify its contents.

### Research Limitation
This dataset represents historical weather estimates for one selected grid location. It will initially be used for historical analysis, not treated as archived weather forecasts available at past prediction times.

```python
# ============================================================
# 04.2.10 — Retrieve and Save Historical Weather Dataset
# ============================================================

import json
import time
import requests
import pandas as pd
from datetime import datetime, timezone
from pathlib import Path
from IPython.display import display

# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

API_URL = "https://archive-api.open-meteo.com/v1/archive"

LATITUDE = 17.3850
LONGITUDE = 78.4867
TIMEZONE = "Asia/Kolkata"

START_DATE = "2023-01-01"
END_DATE = "2026-07-31"

WEATHER_VARIABLES = [
    "temperature_2m",
    "relative_humidity_2m",
    "precipitation",
    "rain",
    "wind_speed_10m",
    "cloud_cover",
]

OUTPUT_CSV = (
    project_root / "data" / "raw" / "weather"
    / "hyderabad_historical_weather.csv"
)

OUTPUT_METADATA = (
    project_root / "data" / "metadata"
    / "weather_source_metadata.json"
)

OUTPUT_MONTHLY = (
    project_root / "results" / "weather_coverage"
    / "monthly_coverage.csv"
)

OUTPUT_LOG = (
    project_root / "results" / "weather_coverage"
    / "request_log.csv"
)

output_paths = [
    OUTPUT_CSV,
    OUTPUT_METADATA,
    OUTPUT_MONTHLY,
    OUTPUT_LOG,
]

existing_files = [str(path) for path in output_paths if path.exists()]

if existing_files:
    # Keep existing outputs intact; write this run to a new set of filenames.
    run_suffix = datetime.now().strftime("%Y%m%d_%H%M%S")
    candidate_paths = [
        path.with_name(f"{path.stem}_{run_suffix}{path.suffix}")
        for path in output_paths
    ]

    counter = 1
    while any(path.exists() for path in candidate_paths):
        run_suffix = (
            f"{datetime.now().strftime('%Y%m%d_%H%M%S')}_{counter}"
        )
        candidate_paths = [
            path.with_name(f"{path.stem}_{run_suffix}{path.suffix}")
            for path in output_paths
        ]
        counter += 1

    OUTPUT_CSV, OUTPUT_METADATA, OUTPUT_MONTHLY, OUTPUT_LOG = candidate_paths
    output_paths = candidate_paths

    print("Existing outputs were left unchanged.")
    print("This run will save to:")
    for path in output_paths:
        print(f"  {path}")

# ------------------------------------------------------------
# 2. Prepare monthly periods
# ------------------------------------------------------------

start = pd.Timestamp(START_DATE)
end = pd.Timestamp(END_DATE)

month_starts = pd.date_range(
    start=start.replace(day=1),
    end=end.replace(day=1),
    freq="MS",
)

monthly_periods = []

for month_start in month_starts:
    month_end = month_start + pd.offsets.MonthEnd(0)

    period_start = max(month_start, start)
    period_end = min(month_end, end)

    monthly_periods.append((
        period_start.strftime("%Y-%m-%d"),
        period_end.strftime("%Y-%m-%d"),
    ))

# ------------------------------------------------------------
# 3. Retrieve monthly weather data
# ------------------------------------------------------------

session = requests.Session()
monthly_data = []
request_log = []
location_metadata = None

for period_start, period_end in monthly_periods:
    params = {
        "latitude": LATITUDE,
        "longitude": LONGITUDE,
        "start_date": period_start,
        "end_date": period_end,
        "hourly": WEATHER_VARIABLES,
        "timezone": TIMEZONE,
    }

    try:
        response = session.get(
            API_URL,
            params=params,
            timeout=60,
        )
        response.raise_for_status()
        payload = response.json()

        if "hourly" not in payload:
            raise ValueError("Response does not contain hourly data.")

        month_df = pd.DataFrame(payload["hourly"])

        expected_columns = ["time"] + WEATHER_VARIABLES
        missing_columns = [
            column for column in expected_columns
            if column not in month_df.columns
        ]

        if missing_columns:
            raise ValueError(
                f"Missing columns: {missing_columns}"
            )

        month_df = month_df[expected_columns].copy()
        month_df["time"] = pd.to_datetime(month_df["time"])

        # Ensure each response contains timestamps only
        # within the requested monthly period.
        lower_bound = pd.Timestamp(period_start)
        upper_bound = pd.Timestamp(period_end) + pd.Timedelta(hours=23)

        if not month_df["time"].between(
            lower_bound, upper_bound
        ).all():
            raise ValueError(
                "Returned timestamps fall outside the requested period."
            )

        monthly_data.append(month_df)

        location_metadata = {
            "returned_latitude": payload.get("latitude"),
            "returned_longitude": payload.get("longitude"),
            "elevation_m": payload.get("elevation"),
            "timezone": payload.get("timezone"),
            "timezone_abbreviation": payload.get(
                "timezone_abbreviation"
            ),
            "utc_offset_seconds": payload.get("utc_offset_seconds"),
        }

        request_log.append({
            "period_start": period_start,
            "period_end": period_end,
            "status": "SUCCESS",
            "http_status": response.status_code,
            "records": len(month_df),
            "error": None,
        })

        print(
            f"SUCCESS | {period_start} to {period_end}"
            f" | {len(month_df):,} records"
        )

    except Exception as exc:
        request_log.append({
            "period_start": period_start,
            "period_end": period_end,
            "status": "FAILED",
            "http_status": (
                response.status_code
                if "response" in locals()
                else None
            ),
            "records": 0,
            "error": str(exc),
        })

        print(
            f"FAILED | {period_start} to {period_end}"
            f" | {type(exc).__name__}: {exc}"
        )

    time.sleep(0.2)

# ------------------------------------------------------------
# 4. Check request success before saving
# ------------------------------------------------------------

request_log_df_final = pd.DataFrame(request_log)

if (
    len(request_log_df_final) != len(monthly_periods)
    or (request_log_df_final["status"] != "SUCCESS").any()
):
    raise RuntimeError(
        "One or more monthly requests failed. "
        "No final dataset was saved. Inspect request_log_df_final."
    )

if not monthly_data:
    raise RuntimeError("No weather data was retrieved.")

# ------------------------------------------------------------
# 5. Combine and validate data
# ------------------------------------------------------------

weather_final_df = (
    pd.concat(monthly_data, ignore_index=True)
    .sort_values("time")
    .reset_index(drop=True)
)

if weather_final_df["time"].duplicated().any():
    raise ValueError("Duplicate timestamps detected.")

if weather_final_df[WEATHER_VARIABLES].isna().any().any():
    raise ValueError(
        "Missing weather values detected. "
        "Investigate before saving the final dataset."
    )

expected_timeline = pd.date_range(
    start=pd.Timestamp(START_DATE),
    end=pd.Timestamp(END_DATE) + pd.Timedelta(hours=23),
    freq="h",
)

actual_timeline = pd.DatetimeIndex(weather_final_df["time"])

missing_timestamps = expected_timeline.difference(actual_timeline)
unexpected_timestamps = actual_timeline.difference(expected_timeline)

if len(missing_timestamps) or len(unexpected_timestamps):
    raise ValueError(
        "Hourly coverage validation failed. "
        f"Missing={len(missing_timestamps)}, "
        f"unexpected={len(unexpected_timestamps)}."
    )

# ------------------------------------------------------------
# 6. Create monthly coverage report
# ------------------------------------------------------------

weather_final_df["month"] = (
    weather_final_df["time"].dt.to_period("M").astype(str)
)

monthly_coverage_df_final = (
    weather_final_df
    .groupby("month")
    .agg(
        returned_records=("time", "size"),
        unique_timestamps=("time", "nunique"),
    )
    .reset_index()
)

monthly_coverage_df_final["expected_records"] = (
    monthly_coverage_df_final["month"].apply(
        lambda month: len(
            pd.date_range(
                start=pd.Period(month, freq="M").start_time,
                end=pd.Period(month, freq="M").end_time.floor("h"),
                freq="h",
            )
        )
    )
)

monthly_coverage_df_final["record_difference"] = (
    monthly_coverage_df_final["returned_records"]
    - monthly_coverage_df_final["expected_records"]
)

# ------------------------------------------------------------
# 7. Create metadata
# ------------------------------------------------------------

metadata = {
    "provider": "Open-Meteo",
    "api_endpoint": API_URL,
    "retrieval_timestamp_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "requested_location": {
        "latitude": LATITUDE,
        "longitude": LONGITUDE,
    },
    "returned_location": location_metadata,
    "timezone": TIMEZONE,
    "start_date": START_DATE,
    "end_date": END_DATE,
    "temporal_resolution": "Hourly",
    "record_count": int(len(weather_final_df)),
    "unique_timestamp_count": int(
        weather_final_df["time"].nunique()
    ),
    "duplicate_timestamp_count": int(
        weather_final_df["time"].duplicated().sum()
    ),
    "missing_timestamp_count": int(len(missing_timestamps)),
    "unexpected_timestamp_count": int(len(unexpected_timestamps)),
    "missing_values_by_variable": {
        variable: int(weather_final_df[variable].isna().sum())
        for variable in WEATHER_VARIABLES
    },
    "variables": WEATHER_VARIABLES,
    "units": {
        "temperature_2m": "°C",
        "relative_humidity_2m": "%",
        "precipitation": "mm",
        "rain": "mm",
        "wind_speed_10m": "km/h",
        "cloud_cover": "%",
    },
    "limitations": [
        "Weather data represents one selected grid location.",
        "Historical weather estimates are not archived forecasts.",
        "Returned coordinates may differ from requested coordinates.",
    ],
    "source_documentation": (
        "https://open-meteo.com/en/docs/historical-weather-api"
    ),
}

# ------------------------------------------------------------
# 8. Save outputs without overwriting
# ------------------------------------------------------------

# Remove temporary analysis column from raw output.
weather_to_save = weather_final_df[
    ["time"] + WEATHER_VARIABLES
].copy()

# Use exclusive creation mode so existing files cannot be overwritten.
# If a write fails, a partial output may remain and must be inspected.
with OUTPUT_CSV.open("x", encoding="utf-8", newline="") as file:
    weather_to_save.to_csv(file, index=False)

with OUTPUT_METADATA.open("x", encoding="utf-8") as file:
    json.dump(metadata, file, indent=4, ensure_ascii=False)

with OUTPUT_MONTHLY.open("x", encoding="utf-8", newline="") as file:
    monthly_coverage_df_final.to_csv(file, index=False)

with OUTPUT_LOG.open("x", encoding="utf-8", newline="") as file:
    request_log_df_final.to_csv(file, index=False)

# ------------------------------------------------------------
# 9. Reopen saved CSV and verify
# ------------------------------------------------------------

weather_saved_df = pd.read_csv(
    OUTPUT_CSV,
    parse_dates=["time"],
)

print("\n--- Saved Dataset Verification ---")
print("Saved CSV:", OUTPUT_CSV.relative_to(project_root))
print("Saved metadata:", OUTPUT_METADATA.relative_to(project_root))
print("Saved monthly report:", OUTPUT_MONTHLY.relative_to(project_root))
print("Saved request log:", OUTPUT_LOG.relative_to(project_root))

print("\nOriginal data shape:", weather_to_save.shape)
print("Saved data shape:", weather_saved_df.shape)
print("Saved columns:", weather_saved_df.columns.tolist())
print("Saved first timestamp:", weather_saved_df["time"].min())
print("Saved last timestamp:", weather_saved_df["time"].max())
print("Saved duplicate timestamps:", weather_saved_df["time"].duplicated().sum())
print("Saved missing values:")
display(weather_saved_df.isna().sum().to_frame("missing_values"))

print("\n--- Monthly Coverage Report (First 5 Rows) ---")
display(monthly_coverage_df_final.head())

print("\n--- Request Log Summary ---")
display(
    request_log_df_final["status"]
    .value_counts()
    .rename_axis("status")
    .to_frame("count")
)

print("\nHistorical weather retrieval and saving completed.")
```

## 04.3.4 — Historical Weather API Pilot Test

### Objective

This subsection tests whether the Open-Meteo Historical Weather API can return hourly weather estimates for multiple geographical locations associated with the five Hyderabad demand zones.

The pilot will examine API accessibility, returned coordinates, hourly variable availability, missing values, and spatial differences between locations.

### Important Data Distinction

Open-Meteo's Historical Weather API provides historical weather estimates derived from reanalysis datasets and weather models. These estimates must not be described as direct weather-station observations.

The pilot is intended to evaluate spatial resolution and API behaviour. The test coordinates are provisional representatives for testing only; they do not define official zone boundaries.

### Pilot Procedure

1. Submit a historical weather request for provisional coordinates associated with the five zones.
2. Retrieve temperature, relative humidity, precipitation, rainfall, wind speed, and cloud cover.
3. Record the coordinates returned by the API.
4. Check whether different requested locations resolve to different weather grid coordinates.
5. Compare hourly rainfall values across locations.
6. Validate timestamps, missing values, and response completeness.

### Decision Rule

Successful API responses alone will not establish suitability for hyperlocal forecasting. Final selection will depend on geographical relevance, spatial resolution, historical coverage, data provenance, and the ability to represent meaningful local weather variation.


```python
# Phase 04.3.4 — Open-Meteo API Spatial Resolution Pilot

import requests
import pandas as pd
from IPython.display import display

# --------------------------------------------------
# 1. API configuration
# --------------------------------------------------

API_URL = "https://archive-api.open-meteo.com/v1/archive"

TEST_DATE = "2026-08-30"

# Provisional representative points for API testing only.
# These are NOT verified zone boundaries or official station locations.

test_locations = {
    "Z01_Kondapur": {
        "latitude": 17.4650,
        "longitude": 78.3560,
    },
    "Z02_Hitech_City": {
        "latitude": 17.4430,
        "longitude": 78.3770,
    },
    "Z03_Gachibowli": {
        "latitude": 17.4400,
        "longitude": 78.3480,
    },
    "Z04_Madhapur": {
        "latitude": 17.4480,
        "longitude": 78.3910,
    },
    "Z05_Kukatpally": {
        "latitude": 17.4930,
        "longitude": 78.3990,
    },
}

weather_variables = [
    "temperature_2m",
    "relative_humidity_2m",
    "precipitation",
    "rain",
    "wind_speed_10m",
    "cloud_cover",
]

# --------------------------------------------------
# 2. Prepare multi-location API request
# --------------------------------------------------

params = {
    "latitude": ",".join(
        str(location["latitude"])
        for location in test_locations.values()
    ),
    "longitude": ",".join(
        str(location["longitude"])
        for location in test_locations.values()
    ),
    "start_date": TEST_DATE,
    "end_date": TEST_DATE,
    "hourly": ",".join(weather_variables),
    "timezone": "Asia/Kolkata",
    "timeformat": "iso8601",
}

print("Requesting historical weather data...")
print("Test date:", TEST_DATE)
print("Number of requested locations:", len(test_locations))

response = requests.get(
    API_URL,
    params=params,
    timeout=120,
)

print("HTTP status:", response.status_code)
response.raise_for_status()

payload = response.json()

# --------------------------------------------------
# 3. Normalize the API response
# --------------------------------------------------

# Multiple coordinates normally produce a list of
# location responses. Handle both response structures.

if isinstance(payload, dict):
    if "hourly" in payload:
        location_responses = [payload]
    else:
        raise ValueError(
            "Unexpected API response structure. "
            f"Returned keys: {list(payload.keys())}"
        )

elif isinstance(payload, list):
    location_responses = payload

else:
    raise TypeError(
        f"Unexpected response type: {type(payload)}"
    )

if len(location_responses) != len(test_locations):
    print(
        "WARNING: Requested location count and "
        "returned response count differ."
    )

# --------------------------------------------------
# 4. Build location metadata and hourly records
# --------------------------------------------------

metadata_rows = []
weather_frames = []

for location_name, location_response in zip(
    test_locations.keys(),
    location_responses,
):
    hourly = location_response.get("hourly", {})

    if "time" not in hourly:
        print(f"WARNING: No hourly timestamps for {location_name}")
        continue

    frame = pd.DataFrame(hourly)
    frame.insert(0, "zone_test_id", location_name)

    requested = test_locations[location_name]

    metadata_rows.append({
        "zone_test_id": location_name,
        "requested_latitude": requested["latitude"],
        "requested_longitude": requested["longitude"],
        "returned_latitude": location_response.get("latitude"),
        "returned_longitude": location_response.get("longitude"),
        "elevation_m": location_response.get("elevation"),
        "timezone": location_response.get("timezone"),
        "hourly_records": len(frame),
    })

    weather_frames.append(frame)

if not weather_frames:
    raise ValueError("No hourly weather records were returned.")

weather_test = pd.concat(
    weather_frames,
    ignore_index=True,
)

location_metadata = pd.DataFrame(metadata_rows)

# --------------------------------------------------
# 5. Inspect returned coordinates
# --------------------------------------------------

print("\n--- Requested vs. Returned Coordinates ---")

display(location_metadata)

# --------------------------------------------------
# 6. Check timestamps and missing values
# --------------------------------------------------

print("\n--- Record Counts by Test Location ---")

display(
    weather_test.groupby("zone_test_id")
    .agg(
        record_count=("time", "size"),
        first_timestamp=("time", "min"),
        last_timestamp=("time", "max"),
    )
    .reset_index()
)

print("\n--- Missing Values by Test Location ---")

missing_summary = (
    weather_test.groupby("zone_test_id")[weather_variables]
    .apply(lambda group: group.isna().sum())
)

display(missing_summary)

# --------------------------------------------------
# 7. Compare hourly rainfall
# --------------------------------------------------

print("\n--- Hourly Rainfall Comparison ---")

rainfall_comparison = weather_test.pivot(
    index="time",
    columns="zone_test_id",
    values="rain",
)

display(rainfall_comparison)

# --------------------------------------------------
# 8. Check unique returned grid coordinates
# --------------------------------------------------

unique_grid_cells = location_metadata[
    ["returned_latitude", "returned_longitude"]
].drop_duplicates()

print("\nNumber of requested locations:", len(test_locations))
print("Number of successful location responses:", len(location_metadata))
print("Number of distinct returned grid coordinates:", len(unique_grid_cells))

print("\n--- Distinct Returned Grid Coordinates ---")
display(unique_grid_cells.reset_index(drop=True))
```

```python
# Phase 04.3.6 — Nearby Weather Station Discovery
# Compatible with the installed Meteostat 2.1.5 interface

import pandas as pd
from IPython.display import display
from meteostat import Point, stations

# --------------------------------------------------
# 1. Provisional zone reference coordinates
# --------------------------------------------------

zone_locations = {
    "Z01_Kondapur": (17.4650, 78.3560),
    "Z02_Hitech_City": (17.4430, 78.3770),
    "Z03_Gachibowli": (17.4400, 78.3480),
    "Z04_Madhapur": (17.4480, 78.3910),
    "Z05_Kukatpally": (17.4930, 78.3990),
}

# These coordinates are search references only.
# They are not verified zone boundaries or station locations.

SEARCH_RADIUS_METERS = 30000
MAX_STATIONS_PER_ZONE = 100

station_records = []

# --------------------------------------------------
# 2. Search for nearby stations
# --------------------------------------------------

for zone_id, (latitude, longitude) in zone_locations.items():

    print(f"\nSearching near {zone_id}...")

    point = Point(latitude, longitude)

    try:
        nearby = stations.nearby(
            point,
            radius=SEARCH_RADIUS_METERS,
            limit=MAX_STATIONS_PER_ZONE,
        )

        if nearby.empty:
            print("No stations returned.")
            continue

        print("Returned station IDs:", len(nearby))

        # The nearby search returns station IDs.
        # Retrieve metadata for each candidate station.

        for station_id in nearby.index:

            try:
                station = stations.meta(str(station_id))

                if station is None:
                    continue

                station_records.append({
                    "search_zone": zone_id,
                    "station_id": station.id,
                    "station_name": station.name,
                    "country": station.country,
                    "region": station.region,
                    "latitude": station.latitude,
                    "longitude": station.longitude,
                    "elevation_m": station.elevation,
                    "timezone": station.timezone,
                })

            except Exception as exc:
                print(
                    f"Metadata lookup failed for {station_id}: {exc}"
                )

    except Exception as exc:
        print(f"Search failed for {zone_id}: {exc}")

# --------------------------------------------------
# 3. Build station inventory
# --------------------------------------------------

if not station_records:
    raise RuntimeError(
        "No station metadata was retrieved. "
        "Check the error messages and Meteostat connectivity."
    )

station_inventory = pd.DataFrame(station_records)

# Remove duplicate station-zone records, if any.
station_inventory = station_inventory.drop_duplicates(
    subset=["search_zone", "station_id"]
)

# --------------------------------------------------
# 4. Calculate distance from each provisional reference
# --------------------------------------------------

import numpy as np

def haversine_km(lat1, lon1, lat2, lon2):
    """Calculate great-circle distance in kilometres."""

    lat1, lon1, lat2, lon2 = np.radians(
        [lat1, lon1, lat2, lon2]
    )

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    return 6371 * 2 * np.arcsin(np.sqrt(a))

station_inventory["distance_km"] = station_inventory.apply(
    lambda row: haversine_km(
        zone_locations[row["search_zone"]][0],
        zone_locations[row["search_zone"]][1],
        row["latitude"],
        row["longitude"],
    ) if pd.notna(row["latitude"])
      and pd.notna(row["longitude"])
      else np.nan,
    axis=1,
)

station_inventory = station_inventory.sort_values(
    ["search_zone", "distance_km"]
).reset_index(drop=True)

# --------------------------------------------------
# 5. Display the discovered station inventory
# --------------------------------------------------

print("\n--- Nearby Weather Station Inventory ---")

display(station_inventory)

print("\n--- Number of Stations Found per Search Zone ---")

display(
    station_inventory.groupby("search_zone")
    .agg(
        candidate_stations=("station_id", "nunique"),
        nearest_station_km=("distance_km", "min"),
    )
    .reset_index()
)

# --------------------------------------------------
# 6. Identify stations appearing in multiple searches
# --------------------------------------------------

print("\n--- Stations Appearing in Multiple Zone Searches ---")

shared_stations = (
    station_inventory.groupby("station_id")
    .agg(
        station_name=("station_name", "first"),
        number_of_zone_searches=("search_zone", "nunique"),
        zones=("search_zone", lambda x: ", ".join(sorted(set(x)))),
    )
    .reset_index()
)

display(
    shared_stations[
        shared_stations["number_of_zone_searches"] > 1
    ]
)
```

```python
# Phase 04.3.10 — Assess Zone Coordinates for IMERG Evaluation

import pandas as pd
from itertools import combinations
from math import radians, sin, cos, sqrt, atan2
from IPython.display import display

# Provisional coordinates used in our previous station investigation.
# These are search reference points, NOT verified zone boundaries.

zone_locations = {
    "Z01_Kondapur": {
        "latitude": 17.4650,
        "longitude": 78.3560,
    },
    "Z02_Hitech_City": {
        "latitude": 17.4430,
        "longitude": 78.3770,
    },
    "Z03_Gachibowli": {
        "latitude": 17.4400,
        "longitude": 78.3480,
    },
    "Z04_Madhapur": {
        "latitude": 17.4480,
        "longitude": 78.3910,
    },
    "Z05_Kukatpally": {
        "latitude": 17.4930,
        "longitude": 78.3990,
    },
}

# --------------------------------------------------
# 1. Display zone reference coordinates
# --------------------------------------------------

zone_coordinates = pd.DataFrame([
    {
        "zone_id": zone_id,
        "latitude": values["latitude"],
        "longitude": values["longitude"],
    }
    for zone_id, values in zone_locations.items()
])

print("--- Provisional Zone Coordinates ---")
display(zone_coordinates)

# --------------------------------------------------
# 2. Calculate geographical distances between zones
# --------------------------------------------------

def haversine_km(lat1, lon1, lat2, lon2):
    """Calculate great-circle distance between two coordinates."""

    earth_radius_km = 6371.0

    lat1, lon1, lat2, lon2 = map(
        radians,
        [lat1, lon1, lat2, lon2],
    )

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        sin(dlat / 2) ** 2
        + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
    )

    return 2 * earth_radius_km * atan2(
        sqrt(a),
        sqrt(1 - a),
    )

distance_records = []

for zone_a, zone_b in combinations(zone_locations.keys(), 2):

    point_a = zone_locations[zone_a]
    point_b = zone_locations[zone_b]

    distance_records.append({
        "zone_a": zone_a,
        "zone_b": zone_b,
        "distance_km": round(
            haversine_km(
                point_a["latitude"],
                point_a["longitude"],
                point_b["latitude"],
                point_b["longitude"],
            ),
            3,
        ),
    })

zone_distances = pd.DataFrame(distance_records)

print("\n--- Distances Between Zone Reference Points ---")
display(zone_distances)

print("\n--- Distance Summary ---")
display(
    zone_distances["distance_km"]
    .describe()
    .to_frame("distance_km")
)

# --------------------------------------------------
# 3. Record nominal IMERG grid spacing for comparison
# --------------------------------------------------

IMERG_GRID_SPACING_DEGREES = 0.1

print("\n--- IMERG Spatial Resolution Reference ---")
print("Nominal grid spacing:", IMERG_GRID_SPACING_DEGREES, "degrees")
print("Approximate nominal spatial resolution: 10 km")
print(
    "Note: Actual ground distance represented by 0.1 degrees "
    "of longitude depends on latitude."
)
print(
    "This cell does not assign zones to IMERG grid cells. "
    "That requires checking the actual product grid coordinates."
)
```

```python
# Phase 04.3.12 — IMERG Catalogue and Availability Check

import requests
import pandas as pd
from IPython.display import display

CMR_URL = "https://cmr.earthdata.nasa.gov/search/granules.json"

# Hyderabad bounding box:
# west, south, east, north
hyderabad_bbox = "78.30,17.35,78.45,17.52"

# Check representative dates across the research period.
test_periods = [
    ("2023-01-01T00:00:00Z", "2023-01-02T23:59:59Z"),
    ("2024-06-01T00:00:00Z", "2024-06-02T23:59:59Z"),
    ("2025-12-01T00:00:00Z", "2025-12-02T23:59:59Z"),
    ("2026-07-01T00:00:00Z", "2026-07-02T23:59:59Z"),
]

headers = {
    "User-Agent": "FoodDemandForecastingResearch/1.0"
}

results = []
sample_granules = []

for start_time, end_time in test_periods:
    params = {
        "short_name": "GPM_3IMERGHH",
        "version": "07",
        "temporal": f"{start_time},{end_time}",
        "bounding_box": hyderabad_bbox,
        "page_size": 5,
        "sort_key": "-start_date",
    }

    record = {
        "start_requested": start_time,
        "end_requested": end_time,
        "http_status": None,
        "granules_returned": None,
        "error": None,
    }

    try:
        response = requests.get(
            CMR_URL,
            params=params,
            headers=headers,
            timeout=60,
        )

        record["http_status"] = response.status_code
        response.raise_for_status()

        payload = response.json()
        granules = payload.get("feed", {}).get("entry", [])

        record["granules_returned"] = len(granules)

        for granule in granules:
            sample_granules.append({
                "requested_period_start": start_time,
                "granule_id": granule.get("id"),
                "granule_start": granule.get("time_start"),
                "granule_end": granule.get("time_end"),
                "collection_concept_id": granule.get("collection_concept_id"),
                "title": granule.get("title"),
                "links": [
                    link.get("href")
                    for link in granule.get("links", [])
                    if link.get("href")
                ],
            })

    except Exception as exc:
        record["error"] = f"{type(exc).__name__}: {exc}"

    results.append(record)

availability_df = pd.DataFrame(results)

print("=== IMERG Catalogue Search Results ===")
display(availability_df)

granules_df = pd.DataFrame(sample_granules)

print("\n=== Sample Granule Metadata ===")

if not granules_df.empty:
    display(
        granules_df[
            [
                "requested_period_start",
                "granule_id",
                "granule_start",
                "granule_end",
                "collection_concept_id",
                "title",
            ]
        ]
    )

    print("\n=== Example Access Links ===")
    for _, row in granules_df.head(3).iterrows():
        print(f"\nGranule: {row['granule_id']}")
        for link in row["links"][:5]:
            print(link)
else:
    print(
        "No granule metadata was returned. "
        "Review the HTTP status and error details before drawing conclusions."
    )
```

## 04.3.17 — IMERG Data File Access Validation

### Objective

Verify whether a real NASA GPM IMERG Early Run HDF5 granule can be accessed and retrieved from NASA's data service.

### Approach

Extract the HTTPS data link from the granule metadata returned by NASA CMR. Send a small streaming request to inspect the HTTP status, final response host, content type, and file signature before saving the file.

### Validation Requirements

* Use the actual URL returned by NASA's catalogue.
* Do not assume HTTP 200 means the response is a valid HDF5 file.
* Detect authentication redirects and HTML login pages.
* Avoid printing authentication credentials.
* Save the sample only after verifying that the response begins with the HDF5 file signature.

### Expected Outcome

Determine whether the sample granule can be downloaded and validated for subsequent inspection of precipitation variables and spatial coordinates.


```python
# Phase 04.3.17 — IMERG HDF5 File Access and Validation

import requests
from pathlib import Path
from urllib.parse import urlparse

# Find the actual GES DISC data-file URL.
data_links = [
    link.get("href")
    for link in first_granule.get("links", [])
    if link.get("href", "").startswith(
        "https://data.gesdisc.earthdata.nasa.gov/data/"
    )
]

if not data_links:
    raise ValueError(
        "No direct GES DISC data-file link was found."
    )

sample_url = data_links[0]

print("Selected granule:", first_granule.get("id"))
print("Data host:", urlparse(sample_url).netloc)
print("File name:", Path(urlparse(sample_url).path).name)

# Request only the beginning of the file to avoid unnecessary downloads.
try:
    with requests.get(
        sample_url,
        headers={
            "User-Agent": "FoodDemandForecastingResearch/1.0",
            "Range": "bytes=0-511",
        },
        stream=True,
        timeout=60,
        allow_redirects=True,
    ) as response:

        print("\nHTTP status:", response.status_code)
        print("Final host:", urlparse(response.url).netloc)
        print("Content type:", response.headers.get("Content-Type"))
        print("Content length:", response.headers.get("Content-Length"))
        print("Content range:", response.headers.get("Content-Range"))

        # Read only a small prefix.
        prefix = next(response.iter_content(chunk_size=512), b"")

        print("Bytes received:", len(prefix))

        # HDF5 files begin with this eight-byte signature.
        hdf5_signature = b"\x89HDF\r\n\x1a\n"

        if prefix.startswith(hdf5_signature):
            print("HDF5 signature: VALID")
        elif response.status_code in (401, 403):
            print(
                "Authentication required or access denied. "
                "Use the official NASA Earthdata login workflow."
            )
        elif prefix.lstrip().startswith(b"<"):
            print(
                "The response appears to contain HTML, not an HDF5 file. "
                "Inspect the authentication or error response."
            )
        else:
            print(
                "HDF5 signature not confirmed. "
                "Check the response and file-access method."
            )

except requests.RequestException as exc:
    print("File-access request failed:", type(exc).__name__, str(exc))
```
