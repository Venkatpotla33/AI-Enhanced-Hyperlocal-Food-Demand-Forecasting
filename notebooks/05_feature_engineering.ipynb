# 6. Forecasting Data Leakage Prevention

Temporal leakage prevention is a critical requirement.

The following rules will be maintained:

1. Lag features must use only previous demand observations.
2. Rolling statistics must exclude the current target observation.
3. Future actual demand must never be used as an input.
4. Historical weather observations must not be treated automatically as future weather information.
5. Future event information may be used only when it would genuinely be available at the forecast origin.
6. Feature selection must respect chronological training and validation boundaries.
7. Any transformations requiring fitted parameters must be fitted using training data only.

The forecasting feature design must account for the information available at the actual prediction time.

````python

# --------------------------------------------------
# PHASE 05.1 – HOURLY TEMPORAL COVERAGE VALIDATION
# Cell 6: Verify complete hourly observations
# --------------------------------------------------

print("=" * 70)
print("HOURLY TEMPORAL COVERAGE VALIDATION")
print("=" * 70)

# Parse timestamps separately without modifying df
ts = pd.to_datetime(df["timestamp"], errors="coerce")

# Expected hourly timeline
expected_hours = pd.date_range(
    start="2023-01-01 00:00:00",
    end="2026-07-31 23:00:00",
    freq="h"
)

print("\n1. EXPECTED STUDY COVERAGE")
print("-" * 70)
print("Expected days:", len(expected_hours) // 24)
print("Expected hours per zone:", len(expected_hours))
print("Expected total observations:", len(expected_hours) * 5)

# Zone-wise coverage
print("\n2. ZONE-WISE HOURLY COVERAGE")
print("-" * 70)

coverage_results = []

for zone in sorted(df["zone_name"].dropna().unique()):
    zone_data = df.loc[df["zone_name"] == zone]
    actual_hours = pd.DatetimeIndex(
        ts.loc[zone_data.index].dropna().unique()
    ).sort_values()

    missing_hours = expected_hours.difference(actual_hours)
    unexpected_hours = actual_hours.difference(expected_hours)

    coverage_results.append({
        "Zone": zone,
        "Records": len(zone_data),
        "Unique_Hours": len(actual_hours),
        "Missing_Hours": len(missing_hours),
        "Unexpected_Hours": len(unexpected_hours)
    })

coverage_df = pd.DataFrame(coverage_results)

display(coverage_df)

# Overall key checks
print("\n3. OVERALL COVERAGE CHECK")
print("-" * 70)

print("Total unique timestamps:", ts.nunique())
print("Invalid timestamps:", ts.isna().sum())
print("Expected timestamps per zone:", len(expected_hours))
print(
    "Total observed zone-time combinations:",
    df.groupby("zone_name")["timestamp"].nunique().sum()
)

print("\n" + "=" * 70)
print("HOURLY COVERAGE VALIDATION COMPLETED")
print("=" * 70)
````
