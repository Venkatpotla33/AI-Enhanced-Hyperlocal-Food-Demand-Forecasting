# 6. Forecasting Data Leakage Prevention

Temporal leakage prevention is a critical requirement.

The following rules will be maintained:

1. Lag features must use only previous demand observations.
2. Rolling statistics must exclude the current target observation.
3. Future actual demand must never be used as an input.
4. Historical weather observations must not be treated automatically as future weather information.
5. Future event information may be used only when it would genuinely be available at the forecast origin.
6. Feature selection must respect chronological training and validation boundaries.
7. Any transformations requiring fitted parameters must be fitted using training data only.

The forecasting feature design must account for the information available at the actual prediction time.

````python

# ============================================================
# PHASE 05.2.2 – WEATHER AND FOOD DEMAND RELATIONSHIP
# Cell 6: Correlation and Demand Association Analysis
# ============================================================

# 1. Aggregate zone-level demand into total hourly demand
hourly_demand = (
    df.groupby("timestamp", as_index=False)["order_count"]
      .sum()
      .rename(columns={"order_count": "total_hourly_demand"})
)

hourly_demand["timestamp"] = pd.to_datetime(
    hourly_demand["timestamp"]
)

# 2. Prepare unique hourly weather observations
hourly_weather = weather_unique[
    ["timestamp"] + weather_columns
].copy()

hourly_weather["timestamp"] = pd.to_datetime(
    hourly_weather["timestamp"]
)

# 3. Merge hourly demand with unique weather observations
weather_demand = hourly_demand.merge(
    hourly_weather,
    on="timestamp",
    how="inner",
    validate="one_to_one"
)

print("=" * 75)
print("WEATHER AND FOOD DEMAND RELATIONSHIP ANALYSIS")
print("=" * 75)

print("\n1. MERGED DATASET VERIFICATION")
print("-" * 75)
print("Hourly observations:", len(weather_demand))
print("Expected observations:", 31392)
print("Missing values:", weather_demand.isna().sum().sum())

# 4. Correlation analysis
print("\n2. PEARSON CORRELATION WITH HOURLY DEMAND")
print("-" * 75)

pearson_corr = (
    weather_demand[weather_columns + ["total_hourly_demand"]]
    .corr(method="pearson")["total_hourly_demand"]
    .drop("total_hourly_demand")
    .sort_values(ascending=False)
)

display(pearson_corr.to_frame("Pearson_Correlation").round(4))

print("\n3. SPEARMAN CORRELATION WITH HOURLY DEMAND")
print("-" * 75)

spearman_corr = (
    weather_demand[weather_columns + ["total_hourly_demand"]]
    .corr(method="spearman")["total_hourly_demand"]
    .drop("total_hourly_demand")
    .sort_values(ascending=False)
)

display(spearman_corr.to_frame("Spearman_Correlation").round(4))

# 5. Scatter plots for selected weather variables
selected_weather = [
    "temperature_2m",
    "relative_humidity_2m",
    "rain",
    "wind_speed_10m"
]

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

for ax, column in zip(axes.flatten(), selected_weather):

    ax.scatter(
        weather_demand[column],
        weather_demand["total_hourly_demand"],
        alpha=0.15,
        s=8
    )

    ax.set_title(f"{column} vs Total Hourly Demand")
    ax.set_xlabel(column)
    ax.set_ylabel("Total Hourly Demand")
    ax.grid(alpha=0.3)

plt.suptitle(
    "Weather Conditions vs Hourly Food Demand",
    fontsize=15
)

plt.tight_layout()
plt.show()

print("\n" + "=" * 75)
print("WEATHER-DEMAND RELATIONSHIP ANALYSIS COMPLETED")
print("=" * 75)
````

````python

# ============================================================
# PHASE 05.2.2 – WEATHER CATEGORY VS FOOD DEMAND
# Cell 7: Conditional Demand Comparison
# ============================================================

weather_category_data = weather_demand.copy()

# 1. Temperature categories
weather_category_data["temperature_category"] = pd.cut(
    weather_category_data["temperature_2m"],
    bins=[-float("inf"), 20, 25, 30, float("inf")],
    labels=["Cool", "Moderate", "Warm", "Hot"]
)

# 2. Rain categories
weather_category_data["rain_category"] = (
    weather_category_data["rain"]
    .gt(0)
    .map({True: "Rainy", False: "Dry"})
)

# 3. Humidity categories
weather_category_data["humidity_category"] = pd.cut(
    weather_category_data["relative_humidity_2m"],
    bins=[-float("inf"), 40, 60, 80, float("inf")],
    labels=["Low", "Moderate", "High", "Very High"]
)

print("=" * 75)
print("WEATHER CATEGORY VS FOOD DEMAND")
print("=" * 75)

# 4. Summary function
def weather_demand_summary(data, category_column):

    summary = (
        data.groupby(category_column, observed=False)
        ["total_hourly_demand"]
        .agg(
            Observation_Count="count",
            Mean_Demand="mean",
            Median_Demand="median",
            Std_Demand="std"
        )
        .round(2)
    )

    return summary

# 5. Temperature comparison
print("\n1. TEMPERATURE CATEGORY VS DEMAND")
print("-" * 75)

temperature_demand_summary = weather_demand_summary(
    weather_category_data,
    "temperature_category"
)

display(temperature_demand_summary)

# 6. Rain comparison
print("\n2. RAINY VS DRY CONDITIONS")
print("-" * 75)

rain_demand_summary = weather_demand_summary(
    weather_category_data,
    "rain_category"
)

display(rain_demand_summary)

# 7. Humidity comparison
print("\n3. HUMIDITY CATEGORY VS DEMAND")
print("-" * 75)

humidity_demand_summary = weather_demand_summary(
    weather_category_data,
    "humidity_category"
)

display(humidity_demand_summary)

# 8. Visualization
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

temperature_demand_summary["Mean_Demand"].plot(
    kind="bar", ax=axes[0]
)
axes[0].set_title("Mean Demand by Temperature")
axes[0].set_xlabel("Temperature Category")
axes[0].set_ylabel("Mean Hourly Demand")
axes[0].tick_params(axis="x", rotation=30)

rain_demand_summary["Mean_Demand"].plot(
    kind="bar", ax=axes[1]
)
axes[1].set_title("Mean Demand: Rainy vs Dry")
axes[1].set_xlabel("Rain Category")
axes[1].set_ylabel("Mean Hourly Demand")
axes[1].tick_params(axis="x", rotation=0)

humidity_demand_summary["Mean_Demand"].plot(
    kind="bar", ax=axes[2]
)
axes[2].set_title("Mean Demand by Humidity")
axes[2].set_xlabel("Humidity Category")
axes[2].set_ylabel("Mean Hourly Demand")
axes[2].tick_params(axis="x", rotation=30)

plt.suptitle(
    "Hourly Food Demand Across Weather Conditions",
    fontsize=15
)

plt.tight_layout()
plt.show()

print("\n" + "=" * 75)
print("WEATHER CATEGORY DEMAND ANALYSIS COMPLETED")
print("=" * 75)
````

````python

# ============================================================
# PHASE 05.2.2 – WEATHER-DEMAND ASSOCIATION WITH TIME CONTROLS
# Cell 8: Exploratory Multiple Regression
# ============================================================

import statsmodels.formula.api as smf

# 1. Prepare analysis data
adjusted_weather_data = weather_demand.copy()

adjusted_weather_data["timestamp"] = pd.to_datetime(
    adjusted_weather_data["timestamp"]
)

adjusted_weather_data["hour"] = (
    adjusted_weather_data["timestamp"].dt.hour
)

adjusted_weather_data["day_of_week"] = (
    adjusted_weather_data["timestamp"].dt.dayofweek
)

# 2. Define regression formula
# Exclude precipitation because it duplicates rain in this dataset.
# Use temperature rather than apparent temperature to avoid
# including two closely related temperature measurements together.

formula = """
total_hourly_demand ~ temperature_2m
+ relative_humidity_2m
+ rain
+ wind_speed_10m
+ cloud_cover
+ C(hour)
+ C(day_of_week)
"""

print("=" * 75)
print("WEATHER-DEMAND REGRESSION WITH TEMPORAL CONTROLS")
print("=" * 75)

# 3. Fit exploratory OLS regression
weather_model = smf.ols(
    formula=formula,
    data=adjusted_weather_data
).fit(
    cov_type="HAC",
    cov_kwds={"maxlags": 24}
)

# 4. Model summary
print("\n1. REGRESSION MODEL SUMMARY")
print("-" * 75)

print("Observations:", int(weather_model.nobs))
print("R-squared:", round(weather_model.rsquared, 4))
print("Adjusted R-squared:", round(weather_model.rsquared_adj, 4))
print("Model F-statistic:", round(weather_model.fvalue, 3))
print("Model probability:", weather_model.f_pvalue)

# 5. Extract weather coefficients
weather_coefficients = pd.DataFrame({
    "Coefficient": weather_model.params,
    "Robust_Std_Error": weather_model.bse,
    "P_Value": weather_model.pvalues
})

weather_coefficients = weather_coefficients.loc[
    weather_columns[:0] + [
        "temperature_2m",
        "relative_humidity_2m",
        "rain",
        "wind_speed_10m",
        "cloud_cover"
    ]
].round(5)

print("\n2. WEATHER COEFFICIENTS")
print("-" * 75)

display(weather_coefficients)

# 6. Confidence intervals
print("\n3. WEATHER COEFFICIENT CONFIDENCE INTERVALS")
print("-" * 75)

confidence_intervals = weather_model.conf_int().loc[
    weather_coefficients.index
].round(5)

confidence_intervals.columns = [
    "Lower_95_CI",
    "Upper_95_CI"
]

display(confidence_intervals)

print("\n" + "=" * 75)
print("TEMPORALLY CONTROLLED REGRESSION COMPLETED")
print("=" * 75)
````

````python

# ============================================================
# PHASE 05.3.1 - WEATHER RELATIONSHIP ANALYSIS
# WITH TEMPORAL CONTROLS AND HAC STANDARD ERRORS
# ============================================================

import pandas as pd
import statsmodels.formula.api as smf

print("=" * 75)
print("PHASE 05.3.1 - WEATHER RELATIONSHIP ANALYSIS")
print("=" * 75)

# 1. Verify required columns
required_columns = [
    "timestamp", "order_count", "hour", "day_of_week", "month",
    "temperature_2m", "relative_humidity_2m",
    "rain", "wind_speed_10m", "cloud_cover"
]

missing_columns = [col for col in required_columns if col not in df.columns]

if missing_columns:
    raise ValueError(f"Missing required columns: {missing_columns}")

# 2. Create timestamp-level dataset
# Demand is summed across all five zones.
# Weather and calendar variables are retained once per timestamp.

hourly_data = (
    df.groupby("timestamp", as_index=False)
      .agg(
          order_count=("order_count", "sum"),
          hour=("hour", "first"),
          day_of_week=("day_of_week", "first"),
          month=("month", "first"),
          temperature_2m=("temperature_2m", "first"),
          relative_humidity_2m=("relative_humidity_2m", "first"),
          rain=("rain", "first"),
          wind_speed_10m=("wind_speed_10m", "first"),
          cloud_cover=("cloud_cover", "first")
      )
)

hourly_data["timestamp"] = pd.to_datetime(hourly_data["timestamp"])
hourly_data["year"] = hourly_data["timestamp"].dt.year

print("\n1. HOURLY DATA VERIFICATION")
print("-" * 75)
print("Hourly observations:", len(hourly_data))
print("Unique timestamps:", hourly_data["timestamp"].nunique())
print("Study start:", hourly_data["timestamp"].min())
print("Study end:", hourly_data["timestamp"].max())

# 3. Fit temporally controlled regression
# Rain is used instead of precipitation to avoid including
# two highly overlapping weather measurements.

formula = """
order_count ~ temperature_2m
            + relative_humidity_2m
            + rain
            + wind_speed_10m
            + cloud_cover
            + C(hour)
            + C(day_of_week)
            + C(month)
            + C(year)
"""

model = smf.ols(formula=formula, data=hourly_data).fit(
    cov_type="HAC",
    cov_kwds={"maxlags": 24}
)

# 4. Display model summary
print("\n2. HAC-ROBUST REGRESSION RESULTS")
print("-" * 75)
print("Observations:", int(model.nobs))
print("R-squared:", round(model.rsquared, 4))
print("Adjusted R-squared:", round(model.rsquared_adj, 4))
print("HAC maximum lags: 24")

# 5. Extract weather coefficients
weather_features = [
    "temperature_2m",
    "relative_humidity_2m",
    "rain",
    "wind_speed_10m",
    "cloud_cover"
]

weather_results = pd.DataFrame({
    "Coefficient": model.params[weather_features],
    "HAC_Std_Error": model.bse[weather_features],
    "P_Value": model.pvalues[weather_features],
    "Lower_95_CI": model.conf_int().loc[weather_features, 0],
    "Upper_95_CI": model.conf_int().loc[weather_features, 1]
})

weather_results = weather_results.round(5)

print("\n3. WEATHER FEATURE STATISTICAL RESULTS")
print("-" * 75)
print(weather_results.to_string())

print("\n" + "=" * 75)
print("PHASE 05.3.1 WEATHER ANALYSIS COMPLETED")
print("=" * 75)
````

````python

# ============================================================
# PHASE 05.3.2 - HOLIDAY AND EVENT RELATIONSHIP ANALYSIS
# ZONE-LEVEL PANEL REGRESSION
# ============================================================

import pandas as pd
import statsmodels.formula.api as smf

print("=" * 75)
print("PHASE 05.3.2 - ZONE-LEVEL HOLIDAY AND EVENT ANALYSIS")
print("=" * 75)

event_features = [
    "is_holiday",
    "is_cricket_event",
    "is_local_event"
]

required_columns = [
    "timestamp", "zone_name", "order_count",
    "hour", "day_of_week", "month"
] + event_features

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

# 1. Prepare zone-level panel without aggregating away events
event_panel = df[required_columns].copy()

event_panel["timestamp"] = pd.to_datetime(
    event_panel["timestamp"]
)

event_panel["year"] = event_panel["timestamp"].dt.year

# 2. Verify event variation across zones
print("\n1. EVENT INDICATOR DISTRIBUTIONS")
print("-" * 75)

for feature in event_features:
    print(f"\n{feature}")
    print(event_panel[feature].value_counts(dropna=False).to_string())

print("\nLOCAL EVENT COUNTS BY ZONE")
print("-" * 75)
print(
    event_panel.groupby("zone_name")["is_local_event"]
    .agg(["count", "sum", "mean"])
    .to_string()
)

# 3. Identify indicators with variation
variable_features = [
    feature for feature in event_features
    if event_panel[feature].nunique(dropna=True) > 1
]

if not variable_features:
    raise ValueError("No event indicators have sufficient variation.")

# 4. Zone-level regression with temporal and zone controls
formula = (
    "order_count ~ "
    + " + ".join(variable_features)
    + " + C(zone_name)"
    + " + C(hour)"
    + " + C(day_of_week)"
    + " + C(month)"
    + " + C(year)"
)

model_event = smf.ols(
    formula=formula,
    data=event_panel
).fit(
    cov_type="cluster",
    cov_kwds={"groups": event_panel["timestamp"]}
)

# 5. Extract event results
available_features = [
    feature for feature in variable_features
    if feature in model_event.params.index
]

event_results = pd.DataFrame({
    "Coefficient": model_event.params.loc[available_features],
    "Clustered_Std_Error": model_event.bse.loc[available_features],
    "P_Value": model_event.pvalues.loc[available_features],
    "Lower_95_CI": model_event.conf_int().loc[available_features, 0],
    "Upper_95_CI": model_event.conf_int().loc[available_features, 1]
}).round(5)

print("\n2. ZONE-LEVEL PANEL REGRESSION")
print("-" * 75)
print("Observations:", int(model_event.nobs))
print("Zones:", event_panel["zone_name"].nunique())
print("Unique timestamps:", event_panel["timestamp"].nunique())
print("R-squared:", round(model_event.rsquared, 4))
print("Adjusted R-squared:", round(model_event.rsquared_adj, 4))
print("Standard errors: Clustered by timestamp")

print("\n3. EVENT FEATURE STATISTICAL RESULTS")
print("-" * 75)
print(event_results.to_string())

print("\n" + "=" * 75)
print("PHASE 05.3.2 ZONE-LEVEL EVENT ANALYSIS COMPLETED")
print("=" * 75)
````

## 3. Phase 05.3.1 – Weather Relationship Analysis

### 3.1 Methodology

Weather variables were analysed against hourly food demand using:

- Hourly observation and timestamp verification.
- HAC-robust regression.
- Newey–West heteroskedasticity and autocorrelation-consistent standard errors.
- Maximum HAC lag of 24 hours.
- Coefficient, p-value, and 95% confidence interval analysis.

|---|---:|---:|---:|

| Rain | -0.13742 | 0.65825 | 0.83463 |
| Wind Speed | 0.16557 | 0.09165 | 0.07083 |
| Cloud Cover | -0.01749 | 0.01337 | 0.19089 |

### 3.3 Model Summary

- Observations: 31,392
- R-squared: 0.9818
- Adjusted R-squared: 0.9817
- Maximum HAC lag: 24 hours

### 3.4 Interpretation

The estimated weather coefficients were not statistically significant at the 5% significance level. Their 95% confidence intervals included zero.

Temperature and wind speed showed relatively smaller p-values than the other weather variables, but the results do not establish statistically significant weather relationships after the temporal controls used in this model.

The high R-squared should be interpreted in the context of the complete regression specification, including its temporal controls. It should not be attributed to weather variables alone.
